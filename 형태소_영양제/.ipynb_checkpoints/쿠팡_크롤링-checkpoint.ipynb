```python
# -*- coding: utf-8 -*-
"""
Coupang 검색 크롤러 (정확 필드만 저장)
- 결과 컬럼: 상품명, 가격(최종), 정가, 배송비, 별점, 리뷰수, 1정당가격
- 신형(li.ProductUnit_*) / 구형(li.search-product) 동시 대응
- 숫자형 컬럼은 정수 보장(없으면 0), 별점은 float
"""

import re, json, time, urllib.parse
from datetime import datetime
from typing import Any, Dict, List, Optional, Tuple

import pandas as pd

import undetected_chromedriver as uc
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.remote.webdriver import WebElement
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.common.exceptions import TimeoutException, StaleElementReferenceException

# =========================== 설정 ===========================
RAW_URL = "https://www.coupang.com/np/search?q=영양제&channel=user&page=1"
MAX_PAGES = 3             # 수집 페이지 수
MAX_PRODUCTS = 120        # None이면 제한 없음
DO_DETAIL = True          # 상세 진입해 1정당가격/별점/리뷰수 보강
HEADLESS = False
WIN_SIZE = "1920,1080"
SHORT, MID, LONG = 3, 7, 12
THROTTLE = 0.8            # 페이지 간 인터벌

# =========================== 유틸/정규식 ===========================
ONLY_INT = re.compile(r"^[0-9,]+$")
STAR_WIDTH = re.compile(r"width\s*:\s*([0-9.]+)\s*%")
UNIT_PAT = re.compile(r"(?:1\s*정|1\s*개\s*당|1\s*포\s*당|1\s*캡슐\s*당|1\s*정\s*당)\s*([0-9,]+)\s*원")

def normalize_url(raw: str, page: int) -> str:
    sp = urllib.parse.urlsplit(raw)
    qs = urllib.parse.parse_qs(sp.query)
    q = qs.get("q", [""])[0]
    if q.startswith("search?"):
        q = q.split("search?", 1)[1]
    if not q: q = "영양제"
    q_enc = urllib.parse.quote(q, safe="")
    channel = qs.get("channel", ["user"])[0]
    return f"https://www.coupang.com/np/search?q={q_enc}&channel={channel}&page={page}"

def to_int(s: Optional[str]) -> int:
    if not s: return 0
    try:
        return int(re.sub(r"[^\d]", "", str(s)))
    except:
        return 0

def make_driver() -> uc.Chrome:
    opts = uc.ChromeOptions()
    if HEADLESS: opts.add_argument("--headless=new")
    opts.add_argument("--disable-dev-shm-usage"); opts.add_argument("--no-sandbox")
    opts.add_argument("--lang=ko_KR"); opts.add_argument(f"--window-size={WIN_SIZE}")
    opts.add_argument("--start-maximized"); opts.add_argument("--disable-blink-features=AutomationControlled")
    opts.add_argument("--user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/141.0.0.0 Safari/537.36")
    service = Service(ChromeDriverManager().install())
    d = uc.Chrome(service=service, options=opts)
    d.set_page_load_timeout(60)
    return d

def wait_cards(d):
    # 하나의 콤마 CSS로 “둘 중 아무거나” 대기 (presence_of_any_elements_located 미사용)
    css = "li.search-product, ul#product-list li, li[class*='ProductUnit_productUnit__']"
    WebDriverWait(d, LONG).until(EC.presence_of_element_located((By.CSS_SELECTOR, css)))

# ===================== JSON-LD(상세) ======================
def parse_json_ld(html: str) -> Dict[str, Any]:
    out: Dict[str, Any] = {}
    for m in re.finditer(r'<script[^>]+type="application/ld\+json"[^>]*>(.*?)</script>', html,
                         flags=re.DOTALL|re.IGNORECASE):
        try: obj = json.loads(m.group(1).strip())
        except: continue
        objs = obj if isinstance(obj, list) else [obj]
        for o in objs:
            if not isinstance(o, dict) or o.get("@type") != "Product": continue
            ag = o.get("aggregateRating") or {}
            if isinstance(ag, dict):
                if "reviewCount" in ag: out.setdefault("review_count", to_int(ag["reviewCount"]))
                if "ratingValue" in ag:
                    try: out.setdefault("rating", float(str(ag["ratingValue"]).replace(",", "")))
                    except: pass
            off = o.get("offers")
            price = None
            if isinstance(off, dict): price = off.get("price")
            elif isinstance(off, list):
                for oo in off:
                    if isinstance(oo, dict) and oo.get("price"): price = oo["price"]; break
            if price is not None: out.setdefault("price", to_int(price))
    return out

# ===================== 카드 파싱 ==========================
def get_url(it: WebElement) -> str:
    a = it.find_elements(By.CSS_SELECTOR, "a.search-product-link")
    if a: 
        href = a[0].get_attribute("href") or ""
        if href: return href
    a2 = it.find_elements(By.CSS_SELECTOR, "a[href*='/vp/products/'], a[href*='/np/products/']")
    return a2[0].get_attribute("href") if a2 else ""

def get_name(it: WebElement) -> str:
    # 1) 구형
    nm = it.find_elements(By.CSS_SELECTOR, "div.name")
    if nm:
        t = (nm[0].text or "").strip()
        if t: return t
    # 2) 신형: productName 영역에서 한글 긴 텍스트 선택
    candidates = []
    for sel in [
        "div[class*='productName'] div, div[class*='productName'] a, div[class*='productName'] span",
        "strong, span, em"
    ]:
        for e in it.find_elements(By.CSS_SELECTOR, sel):
            t = (e.text or "").strip()
            if not t: continue
            if "원" in t or "할인" in t or "적립" in t:  # 가격/혜택 제거
                continue
            if len(t) >= 10 and re.search("[가-힣]", t):
                candidates.append(t)
    if candidates:
        return max(candidates, key=len)
    # 3) 마지막 보루: 링크 title/aria-label
    a = it.find_elements(By.CSS_SELECTOR, "a[href*='/vp/products/'], a[href*='/np/products/']")
    if a:
        for attr in ("title", "aria-label"):
            v = a[0].get_attribute(attr) or ""
            if v and len(v) >= 6: return v.strip()
    return ""

def get_price_and_orig(it: WebElement) -> Tuple[int, int]:
    price = 0
    # 구형: strong.price-value
    pv = it.find_elements(By.CSS_SELECTOR, "strong.price-value")
    if pv:
        price = to_int(pv[0].text)
    if price == 0:
        # 신형: 텍스트에서 “####원” 중 취소선(del)이 아닌 첫 값
        # 먼저 취소선 안의 값 수집
        dels = [to_int(el.text) for el in it.find_elements(By.CSS_SELECTOR, "del") if el.text]
        del_set = set([v for v in dels if v])
        for s in it.find_elements(By.CSS_SELECTOR, "div, span, strong"):
            t = (s.text or "").strip()
            if not t or "원" not in t: continue
            if ONLY_INT.match(t.replace("원", "").replace(",", "").strip()):
                v = to_int(t)
                if v and v not in del_set:
                    price = v; break
    # 정가(취소선)
    orig = 0
    for sel in ["del.price-old", "span.price > del", "div.price-amount.original-price-amount", "del"]:
        els = it.find_elements(By.CSS_SELECTOR, sel)
        if els and els[0].text:
            orig = to_int(els[0].text); break
    return price, orig

def get_shipping_fee(it: WebElement) -> int:
    # “배송비 ####원” 뱃지/문구
    badge = it.find_elements(By.CSS_SELECTOR, "[class*='TextBadge_feePrice']")
    if badge:
        v = to_int(badge[0].text)
        if v: return v
    for s in it.find_elements(By.CSS_SELECTOR, "span, div"):
        t = (s.text or "").strip()
        if "배송비" in t and "원" in t:
            v = to_int(t)
            if v: return v
    # 무료배송이면 0
    if any("무료배송" in (x.text or "") for x in it.find_elements(By.CSS_SELECTOR, "span, div")):
        return 0
    return 0

def get_unit_price(it: WebElement) -> int:
    for s in it.find_elements(By.CSS_SELECTOR, "span, div"):
        t = (s.text or "").strip()
        if not t: continue
        m = UNIT_PAT.search(t)
        if m: return to_int(m.group(1))
    return 0

def get_rating_and_count(it: WebElement) -> Tuple[float, int]:
    # 구형
    rating = None
    em = it.find_elements(By.CSS_SELECTOR, "em.rating")
    if em and em[0].text:
        try: rating = float(em[0].text.strip())
        except: rating = None
    cnt = 0
    rc = it.find_elements(By.CSS_SELECTOR, "span.rating-total-count")
    if rc and rc[0].text:
        cnt = to_int(rc[0].text)
    # 신형
    if rating is None:
        stars = it.find_elements(By.CSS_SELECTOR, "[class*='ProductRating_star']")
        if stars:
            style = stars[0].get_attribute("style") or ""
            m = STAR_WIDTH.search(style)
            if m:
                pct = float(m.group(1))
                rating = round(5.0 * pct / 100.0, 2)
    if cnt == 0:
        rc2 = it.find_elements(By.CSS_SELECTOR, "[class*='ProductRating_ratingCount']")
        if rc2 and rc2[0].text:
            cnt = to_int(rc2[0].text)
    return (rating if rating is not None else 0.0), cnt

def parse_card(it: WebElement) -> Dict[str, Any]:
    name = get_name(it)
    if not name: return {}
    price, orig = get_price_and_orig(it)
    ship = get_shipping_fee(it)
    unitp = get_unit_price(it)
    rating, rcount = get_rating_and_count(it)

    # 상세 보강(옵션)
    if DO_DETAIL and (unitp == 0 or (rating == 0.0 and rcount == 0)):
        url = get_url(it)
        if url:
            try:
                it.parent.execute_script("window.open('','_blank');")
                it.parent.switch_to.window(it.parent.window_handles[-1])
                it.parent.get(url)
                WebDriverWait(it.parent, LONG).until(EC.presence_of_element_located((By.CSS_SELECTOR, "body")))
                time.sleep(0.8)
                # 1정당가격
                body = (it.parent.find_element(By.TAG_NAME, "body").text or "")
                m = UNIT_PAT.search(body)
                if m: unitp = unitp or to_int(m.group(1))
                # JSON-LD로 별점/리뷰/가격 보강
                j = parse_json_ld(it.parent.page_source)
                if rating == 0.0 and j.get("rating") is not None:
                    try: rating = float(j["rating"])
                    except: pass
                if rcount == 0 and j.get("review_count") is not None:
                    rcount = int(j["review_count"])
                if price == 0 and j.get("price") is not None:
                    price = int(j["price"])
            except Exception:
                pass
            finally:
                if len(it.parent.window_handles) > 1:
                    it.parent.close()
                    it.parent.switch_to.window(it.parent.window_handles[0])

    return {
        "상품명": name,
        "가격(최종)": int(price or 0),
        "정가": int(orig or 0),
        "배송비": int(ship or 0),
        "별점": float(rating or 0.0),
        "리뷰수": int(rcount or 0),
        "1정당가격": int(unitp or 0),
    }

# =========================== 메인 =============================
def main():
    d = make_driver()
    rows: List[Dict[str, Any]] = []
    try:
        taken = 0
        for page in range(1, MAX_PAGES + 1):
            url = normalize_url(RAW_URL, page)
            print("[GO]", url)
            d.get(url)
            try:
                wait_cards(d)
            except TimeoutException:
                print("[WARN] 카드 대기 타임아웃")
            time.sleep(1.0)

            # lazy 보정 스크롤
            for _ in range(5):
                d.execute_script("window.scrollBy(0, 1200);"); time.sleep(0.25)

            cards = d.find_elements(By.CSS_SELECTOR, "li.search-product, ul#product-list li, li[class*='ProductUnit_productUnit__']")
            for i in range(len(cards)):
                try:
                    it = cards[i]
                    rec = parse_card(it)
                except StaleElementReferenceException:
                    fresh = d.find_elements(By.CSS_SELECTOR, "li.search-product, ul#product-list li, li[class*='ProductUnit_productUnit__']")
                    if i < len(fresh):
                        rec = parse_card(fresh[i])
                    else:
                        continue
                if not rec: continue
                rows.append(rec)
                taken += 1
                if MAX_PRODUCTS and taken >= MAX_PRODUCTS:
                    break
            if MAX_PRODUCTS and taken >= MAX_PRODUCTS:
                break
            time.sleep(THROTTLE)
    finally:
        d.quit()

    df = pd.DataFrame(rows, columns=["상품명","가격(최종)","정가","배송비","별점","리뷰수","1정당가격"])
    # 숫자/타입 보정
    for col in ["가격(최종)","정가","배송비","리뷰수","1정당가격"]:
        df[col] = pd.to_numeric(df[col], errors="coerce").fillna(0).astype(int)
    df["별점"] = pd.to_numeric(df["별점"], errors="coerce").fillna(0).astype(float)

    ts = datetime.now().strftime("%Y%m%d_%H%M")
    out_csv = f"coupang_clean_{ts}.csv"
    out_xlsx = f"coupang_clean_{ts}.xlsx"
    df.to_csv(out_csv, index=False, encoding="utf-8-sig")
    try:
        df.to_excel(out_xlsx, index=False)
    except Exception as e:
        print("[WARN] 엑셀 저장 실패:", e)
    print(f"[DONE] 저장: {out_csv} / {out_xlsx} (행 {len(df)})")

if __name__ == "__main__":
    main()
```
