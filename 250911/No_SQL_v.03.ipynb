```python
import requests
from bs4 import BeautifulSoup

for page_num in range(10) : 
    if page_num == 0 :
        res = requests.get("https://davelee-fun.github.io/")
    else :
        res = requests.get(f"https://davelee-fun.github.io/page{str(page_num + 1)}")
    soup = BeautifulSoup(res.content, "html.parser")
    data = soup.select("div.card-body")
    for item in data :
        category = item.select_one("h2.card-title").get_text().replace("관련 상품 추천","").strip()
    for item in data :
        product = item.select_one("h4.card-text").get_text().replace("상품명:","").strip()
        print(category, product)
```

```python
products = collection.find({}, {"_id": 0, "product": 1})
for product in products :
    print(product["product"])
```

```python
Top_5 = [
    {"$match": {"category" : "매트리스커버"}}]

tops = list(collection.aggregate(Top_5))
for top in tops :
    print(top["product"])
```

```python
products = collection.find({"category": "매트릭스커버"}, {"_id":0,"product":1})
for product in products:
    print(product)
```

```python
# 각 카테고리별 상품 수
Top_5 = [
    {"$group": {"_id" : "$category", "count": {"$sum": 1}}}
    ,{"$project": {"_id":0, "category": "_id", "count": 1}}]

tops = list(collection.aggregate(Top_5))
print(tops)
```

```python
# 상품명의 "핑크"가 포함된 상품을 조회주세요

for movie in collection.find({"product" :{"$regex":"핑크", "$options" : "i"}}) :
    print(movie["product"])
```
