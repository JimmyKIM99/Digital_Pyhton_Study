###. SQL 개념 정의
1. 데이터 (Data)
2. 데이터베이스
3. 스키마 (Schema)
4. 테이블 (table)
5. SQL, No SQL

1. Data
- 수집 가능한 값, 사실, 기록물 등의 원초적인 정보
- 숫자, 문자, 상태(True, False)
- 데이터 그 자체로는 의미가 부족 -> 구조화가 되어야 데이터는 비로소 가치를 갖게 됨

2. 데이터 베이스 (Database = DB)
- 데이터를 구조화 시켜주는 저장소
- 여러가지 다양한 데이터를 구조적이고 체계적으로 저장 및 관리할 수 있도록 해주는 저장소 (물리적 구조를 논리적으로 나눠놓은 것)
- 관계형 데이터베이스 : (RDBMS -> SQL / Oracle // MySQL)
- 비관계형 데이터베이스 : (DMBS -> No SQL / MongoDB)

3. 스키마 (Schema)
- 데이터베이스를 어떻게 구조화할 것인지를 정의해놓은 일종의 설계도
- 데이터베이스 내부에 존재하는 논리적 구조

4. 테이블 (table)
- 관계형 데이터베이스 안에 저장된 실제 데이터의 저장 단위
- 구성 요소 : 열(Column) => 속성(attribute) / 행(Row) => 레코드(record)

5. SQL (Structured Query Language)
- 관계형 데이터베이스를 정의, 조작, 제어 할 수 있도록 도와주는 표준 언어
- 종류
    - DDL(Data Definition Language) : CREATE, DROP, ALTER
    - DML(Data Manipulation Language) : SELECT, INSERT, UPDATE, DELETE
    - DCL(Data Control Language) : GRANT, REVOKE

- 데이터 : 가장 하위 개념, 원초적인 정보 요약
- 테이블 : 해당 데이터를 담은 (*구조화) 틀
- 스키마 : 이러한 테이블 간  관계를 정의한 설계도면
- 데이터베이슨 : 스키마의 개념 속에 존재하는 테이블이 포함된 상위 구조 및 개념
- SQL : 이 모든 구조를 만들고, 조작하고, 제어할 수 있도록 해주는 언어

### SELECT 문법 :
- LIMIT : 가져올 수 있는 데이터 총 일부만 가져오고 싶을 때
- COUNT (*집계함수) : 데이터의 갯수를 세고 싶을 때
- DISTINCT : 특정 값을 가져올 때, 중복된 값을 출력하고 싶지 않을 떄
- SUM : 해당 컬럼값 합계
- AVG : 해당 컬럼값의 평균값
- MAX : 해당 컬럼값의 최대값
- MIN : 해당 컬럼값의 최솟값
- GROUP BY : 특정 컬럼값을 기반으로 그룹핑하기
  ex) SELECT rating FROM film GROUP BY rating -> 필름 테이블에서 rating값을 찾아오는데, 해당 rating의 값을 그룹화해서 찾아와라
  ex) SELECT COUNT(*) FROM film GROUP BY rating -> 각 rating 값의 종류별로 몇개의 데이터가 있는지 확인
  ex) SELECT COUNT(*) FROM film WHERE 조건문 GROUP BY rating -> 특정 조건에 맞는 데이터 중 rating 값의 종류별로 몇 개의 데이터가 있는지 확인
- ODER BY : 특정 컬럼값을 기준으로 데이터를 정렬 (*오름차순 = ASC || 내림차순 = DESC)
  if DESC, ASC 중 어떤 값도 입력하지 않았다면, 디폴트 값은 ASC

### SQL 중급 단계
1) FOREIGN KEY (외래키) <-> PRIMARY KEY (고유키)
2) HAVING
   -HAVING 절을 활용해서 조건 비교!!
   -독립 테이블 내 GROUP BY를 직접 상태에서 집계함수가 포함된 조건식을 사용할 때, HAVING 절을 사용해야함!
3) JOIN : 서로다른 테이블을 하나의 연결된 테이블로 사용 (실무에서 많이 사용됨!! / 복수의 테이블 관리 -> 각각의 테이블간 연결고리 FOREIGN KEY -> 테이블간 연결)
   - INNER JOIN : 서로 다른 테이블간 매칭되는 데이터에 한에서 해당 데이터를 기준으로 연결
     (*JOIN 함수 사용시 default가 INNER JOIN임)
     서로  다른 테이블을 하나로 연결하기 위해 -> ON 전치사
   - 신문법 JOIN -> USING()으로 대체 가능
   - OUTER JOIN : 서로 매칭되지 않는 데이터도 모두 찾아옴
     LEFT OUTER JOIN : 먼저 언급된(*Left data-부모데이터를 기준으로 다른 테이블을 연결 / 부모 데이터와 매칭되지 않는 자식데이터도 가져옴)
     RIGHT OUTER JOIN : 자식 데이터를 기준으로 다른 테이블을 연결, 마찬가지로 매칭되지 않은 데이터를 누락시키지 않고 모두 찾아옴
4) SubQuery
   -SQL문 안에 포함되어있는 SQL문 => 중첩 SQL문
5) INDEX
   -파이썬 > 리스트 > 리스트 자료구조 내부에 있는 아이템들에게 각각 인덱스값을 부여
   -SQL구문을 활용해 컬럼 안에 입력되어있는 데이터들에게 개별적인 인덱스 값을 부여 -> 값을 서칭해서 찾아오는 빈도 잦음 -> 너무 많은 시간, 과정
   -인덱스 종류
       -클러스터형 인덱스 : 영어 사전과 같은 형태로 데이터를 정렬 (ex. PRIMARY KEY, FOREIGN KEY, UNIQUE)
       -보조 인덱스 : 책 끝부분에 있는 찾아보기 인덱스 페이지(색인)
       (ex. ALTER TABLE usertbl ADD CONSTRAINT TESTDate UNIQUE(mDATE);)
       (ex. CREATE INDEX idx_birth ON usertbl(birthYear);)
       (ex. ALTER TABLE usertbl ADD INDEX index_addr(addr);)
       (ex. ALTER TABLE usertbl DROP INDEX index_addr; -> 보조 인덱스 삭제 방법)
       (ex. SHOW INDEX FROM customers; -> 인덱스 상태 확인)

- 다양한 SQL 함수
    - 문자열 함수 :
      1) LENGTH(string) 문자의 길이를 반환하는 함수
      2) UPPER(string) 특정 문자를 대문자로 변환시켜주는 함수
      3) LOWER(string) 특정 문자를 소문자로 변환시켜주는 함수
      4) CONCAT(string1, string2 ...) 복수의 문자열을 하나로 연결하는 함수
      5) SUBSTRING() 일부 문자열만 추출해주는 함수 (인자값1, 시작, 몇개)
    - 날짜 / 시간 함수 :
      1) NOW() : 현재 날짜와 시간을 반환하는 함수
      2) CURDATE() : 현재 날짜를 반환하는 함수
      3) CURTIME() : 현재 시간을 반환하는 함수
      4) DATE_ADD(date, INTERVAL unit(t)) : 특정 날짜에서 간격을 추가해주는 함수
      5) DATE_SUB(date, INTERVAL unit(t)) : 특정 날짜에서 간격을 빼주는 함수
         - unit type : MONTH / DAY / HOUR / MINUTE / SECOND
      6) EXTRACT(추출해오고자 하는 값 = field FROM 추출해 오고자 하는 대상 = source)
      7) YEAR(), MONTH(), DAY(), HOUR(), MINUTE(), SECOND()
      8) DAYOFWEEK 해당 날짜의 요일 반환, 숫자값으로
      9) DATE_FORMAT() : 형식지정자로 값을 받아서 요일을 출력할 수 있음 -> %w, %a
          - %y : 연도의 마지막 두자리
          - %Y : 4자리 연도
          - %M : 영문 월 JUNE
          - %m : 월을 두자리 숫자로 표시 01 02
          - %c : 월을 한자리 숫자로 표기
          - %D : 일을 두자리 숫자 + 영문 접미사
          - %d : 일을 두자리 숫자 01 02
          - %H : 시간을 24시간 두자리
          - %h : 시간을 12시간 형식의 두자리 표현
          - %i : 분을 두자리 숫자로
          - %s : 초를 두자리 숫자로
      10) TIMESTAMPDIFF(unit, start_datetime, end_datetime) 두 날짜 혹은 시간 값의 차이를 계산할 때 사용하는 함수
          - unit type : MONTH / DAY / HOUR / MINUTE / SECOND
      11) ABS(number) : Absolute 절댓값
      12) CEIL(number) 올림처리
      13) FLOOR(number) 바텀처리
      14) ROUND(number) 반올림처리 -> ROUND(숫자A, 숫자B) -> 숫자B가 최종적인 반올림 결과값 자릿수가 되도록!
      15) SQRT(number) 해당 숫자의 제곱근을 반환하는 함수

### SQL 고급문법
- 고급 윈도우 : SQL 쿼리문 안에서 데이터 집합을 세분화하여 각 부분에 대한 계산을 수행하는 함수
  1) RANK() : 순위를 매기는데, 동일한 값이 있을 경우 같은 순위 부여, 다음 순위는 건너뜀
  2) DENSE_RANK : 순위를 매기는데, 동일한 값이 있을 경우, 같은 순위 부여 후 다음 순위 건너뜀X
  3) ROW_NUMBER : 순위와 상관없이 각 행에 고유한 번호를 부여하는 것
     > 위 3개의 함수는 공통적으로 OVER절 사용
     > ORDER BY 절을 통해 순위를 매길 기준을 정의!!
- 윈도우 함수에서 많이 사용되는 구문
  1) PARTITION BY : 특정 컬럼을 기준으로 부분집합으로서 분할 때 사용
     - COUNT(*)
     - SUM()
     - AVG()
     - MIN()
     - MAX()
     - RANK()
     - DENSE()
     - ROW_NUMBER()
     - LEAD(column, n, default value) : 현재 행을 기준으로 n행 뒤의 값을 가져올 때 사용
     - LAG(column, n, default value) : 현재 행을 기준으로 n행 앞의 값을 가져올 떄 사용
     - FIRST_VALUE() : 파티션을 통해 부분집합화 되어진 그룹을 기준으로 첫번째 값을 가져올 때 사용
     - LAST_VALUE() : 파티션을 통해 부분집합화 되어진 그룹을 기준으로 마지막번째 값을 가져올 때 사용
  3) ORDER BY : 특정 부분집합 내 정렬 기준을 정의하고자 할 때 사용
  4) ROWS : 특정 부분집합의 범위를 설정할 때 사용
     - 특정 부분집합의 행을 물리적으로 해석
     - UNBOUNDED PRECEDING : 앞에서 파티션을 통해 설정된 부분집합의 첫번째 행부터 시작
     - UNBOUNDED FOLLOWING : 앞에서 파티션을 통해 설정된 부분집합의 마지막번째 행까지
     - CURRENT ROW : 현재 행을 포함해라
     - n PRECEDING/FOLLOWING : n번째 행 앞이나 뒤
  6) RANGE : 정렬하고자 하는 기준값과 동일한 값을 가진 모든 행을 하나의 그룹으로 정의
     - 특정 부분집합 내 동일한 값을 가지고 있는 요소를 하나의 행으로 정의)
  7) SUM(revenue) : 일반적인 집계함수의 기능, 만약 GROUP BY절과 함꼐 사용한다면 해당 그룹의 revenue 값을 계산!
  8) SUM(revenue) OVER() : 윈도우 함수로서 기능, OVER()는 특정 파티션이 내부에서 정의된 상태라면, 해당 파티션을 기준으로 revenue 값을 집계합니다
     but, OVER() 안에 특정 파티션이 정의되지 않는다면 전체 결과 세트에 대해서 SUM() 함수를 적용하라는 의미
  9) PERCENT_RANK : 행의 백분위 순위를 계산 => 0~1까지, 행의 값을 나눠 순위를 매길때 사용
  10) CUME_DIST() : 행의 누적분포를 계산 => 0부터 1까지
  11) NTILE(n) : 행을 n개의 그룹으로 분할합니다
