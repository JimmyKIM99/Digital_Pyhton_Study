```python
# fetchall() : 해당 테이블 내 모든 데이터를 가져옵니다
# fetchone() : 매칭되는 해당 값 중 첫번쨰 값을 가져옵니다
# > 이미 앞에서 먼저 찾아온 값이 존재한다면, 해당 값 이후부터 값을 찾아옵니다
# fetchmany() : 인자값으로 작성된 숫자만큼 데이터를 가져옵니다
#
```

```python
SQL = "SELECT * FROM product"
cursor.execute(SQL)
row = cursor.fetchone()
print(row)
```

```python
row = cursor.fetchmany(2)
print(row)
#이미 찾은 값이 존재한다면, 다음 값부터 찾음
```

```python
rowsALL = cursor.fetchall()
for row in rowsALL :
    print(row)
```
