###. SQL 개념 정의
1. 데이터 (Data)
2. 데이터베이스
3. 스키마 (Schema)
4. 테이블 (table)
5. SQL, No SQL

1. Data
- 수집 가능한 값, 사실, 기록물 등의 원초적인 정보
- 숫자, 문자, 상태(True, False)
- 데이터 그 자체로는 의미가 부족 -> 구조화가 되어야 데이터는 비로소 가치를 갖게 됨

2. 데이터 베이스 (Database = DB)
- 데이터를 구조화 시켜주는 저장소
- 여러가지 다양한 데이터를 구조적이고 체계적으로 저장 및 관리할 수 있도록 해주는 저장소 (물리적 구조를 논리적으로 나눠놓은 것)
- 관계형 데이터베이스 : (RDBMS -> SQL / Oracle // MySQL)
- 비관계형 데이터베이스 : (DMBS -> No SQL / MongoDB)

3. 스키마 (Schema)
- 데이터베이스를 어떻게 구조화할 것인지를 정의해놓은 일종의 설계도
- 데이터베이스 내부에 존재하는 논리적 구조

4. 테이블 (table)
- 관계형 데이터베이스 안에 저장된 실제 데이터의 저장 단위
- 구성 요소 : 열(Column) => 속성(attribute) / 행(Row) => 레코드(record)

5. SQL (Structured Query Language)
- 관계형 데이터베이스를 정의, 조작, 제어 할 수 있도록 도와주는 표준 언어
- 종류
    - DDL(Data Definition Language) : CREATE, DROP, ALTER
    - DML(Data Manipulation Language) : SELECT, INSERT, UPDATE, DELETE
    - DCL(Data Control Language) : GRANT, REVOKE

- 데이터 : 가장 하위 개념, 원초적인 정보 요약
- 테이블 : 해당 데이터를 담은 (*구조화) 틀
- 스키마 : 이러한 테이블 간  관계를 정의한 설계도면
- 데이터베이슨 : 스키마의 개념 속에 존재하는 테이블이 포함된 상위 구조 및 개념
- SQL : 이 모든 구조를 만들고, 조작하고, 제어할 수 있도록 해주는 언어

### DDL
- CREATE DATABASE dbname;
- CREATE DATABASE IF NO EXISTS dbname;
- SHOW DATABASE;
- CREATE TABLE tablename;
- DROP TABLE IF EXISTS tablename;
- ALTER TABLE tablename ADD COLUMN attributename attributetype;
- ALTER TABLE tablename MODIFY COLUMN attributename atrributetype;
- ALTER TABLE tablename CHANGE COLUMN arttributename1 attribute2 arttributetype;
- ALTER TABLE tablename DROP COLUMN arttributename;

### DML
- INSERT INTO tablename VALUES (Field Value1, Field Value2, ...);
- INSERT INTO tablename (Field1, Field2, Field3) VALUES (Field Value1, Field Value2, Field Value3);
- INSERT INTO tablename (Field1, Field2, Field3) VALUES (Field Value1, Field Value2, Field Value3), (Field Value4, Field Value5, Field Value6);
- DESC tablename
- SELECT * FROM tablename;
- SELECT column FROM tablename;
- SELECT column1, column2 FROM tablename;
- SELECT * FROM tablename WHERE column = value;
- "=" / ">" / "<" / "!="
- SELECT * FROM tablename WHERE column >= value1 AND, OR value2;
- SELECT * FROM tablename WHERE attribute LIKE "%value%"
- SELECT * FROM tablename WHERE attribute LIKE "_value_"
- UPDATE tablename SET column = value;
- UPDATE tablename SET column = value, column = value WHERE primarykey = value;
- DELETE FROM tablename WHERE primarykey = value;
