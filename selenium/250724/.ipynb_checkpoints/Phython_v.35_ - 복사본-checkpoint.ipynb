```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.keys import Keys
import time

service = Service(ChromeDriverManager().install())
options = Options()
driver = webdriver.Chrome(service = service, options = options)
driver.get("https://davelee-fun.github.io/")

for _ in range(5) :
    elements = driver.find_elements(By.CSS_SELECTOR, "h4.card-text")
    for element in elements :
        print(element.text)
    driver.execute_script("window.scrollTo(0, document.body.scrollHeight)")
    elements = driver.find_elements(By.CSS_SELECTOR, ".ml-1.ml-1")
    elements[-1].click()
    time.sleep(10)



driver.quit()
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options
from webdriver_manager.chrome import ChromeDriverManager
from urllib.request import urlretrieve
import time


service = Service(ChromeDriverManager().install())
options = Options()
options.add_argument("headless")
options.add_argument("window-size=1920x1080")
options.add_argument("--disable-gpu")
options.add_argument("lang=ko_KR")


driver = webdriver.Chrome(service = service, options = options)
driver.get("https://davelee-fun.github.io/")

elements = driver.find_elements(By.CSS_SELECTOR, "div.maxthumb img")
sources = list()

for element in elements :
    sources.append(element.get_attribute("src"))

for index, source in enumerate(sources) :
    urlretrieve(source, f"IMG/{str(index)}.{source.split(".")[-1]}")


time.sleep(2)
driver.quit()
```

```python
## HTML 크롤링을 위한 XPATH 기본 포맷 (*형태)
- //tagname[@Attribut="value"]
- / : 절대경로를 의미 (ex : \html\body\dev\a)
- // : 문서 내에서 바로 검색을 하라는 의미 (ex : \\h1 -> 문서에서 바로 h1 태그 찾아와라)
- //a[@href="https:www.google.com"] : 문서 내 a 태그를 찾아오는데, href 속성값으로 구글 닷컷에서 찾아라
- //a[@href] : 문서 내 href라는 속성을 가지고 있는 모든 태그를 다 찾아와라
- (//a)[3] : 문서 내 a 태그를 찾아오는데 3번째 a만 찾아와라
- (//tabe)[last()] : 문서 내 table 태그를찾아오는데, 마지막번째 table만 찾아와라
- (//a)[position() < 3] : 문서 내 a 태그를 찾아오는데, 앞에서부터 2개만 찾아와라
- //div[@*] : 문서 내 div 태그를 찾아오는데, 어떤 속성이든 상관없이 속성이 하나라도 있으면 찾아와라
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.keys import Keys
import time

service = Service(ChromeDriverManager().install())

options = Options()
driver = webdriver.Chrome(service = service, options = options)
driver.get("https://divjason.github.io/sellenium-test/")

search_input1 = driver.find_element(By.XPATH, "//input[@id='username']")
search_input1.clear()
search_input1.send_keys("error@error.com")
search_input2 = driver.find_element(By.XPATH, "//input[@id='password']")
search_input2.clear()
search_input2.send_keys("1234")
search_input2.send_keys(Keys.RETURN)

search_input3 = driver.find_element(By.XPATH, "//div[@class='message']")
print(search_input3.text)

time.sleep(3)

driver.quit()
```
