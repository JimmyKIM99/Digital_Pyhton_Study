```python
# -*- coding: utf-8 -*-
"""
도시락 밴드 1/2 형태소 분석 (밴드 +-1500 포함 + 핵심구간 가중 그래프)
- Windows 한글 폰트 자동 탐색( Malgun → Hancom/HCR → Gulim/Dotum/Batang/Gungsuh → Nanum )
- KoNLPy(Okt 우선, Mecab 대체), 미설치 시 정규식 폴백
- 명사/형용사만 추출, 확장 불용어 적용
- 밴드별: 빈도/바이그램/트라이그램/공동등장/워드클라우드/TF-IDF
- 추가: 확장구간 분석 + 핵심구간(원래 밴드) 토큰에 더 높은 가중을 준 '가중 빈도' 그래프
- 모든 CSV: utf-8-sig 저장
"""

# ========= 0) 옵션 =========
USE_OKT = True               # konlpy 설치되어 있으면 True 권장
OUTDIR  = "./out"            # 결과 저장 폴더
FONT_PATH = None             # 특정 폰트 경로를 강제 사용하려면 지정 (예: r"C:\Windows\Fonts\malgun.ttf")
PRICE_MARGIN = 1500          # 밴드 범위 확장 폭 (±원)
CORE_WEIGHT  = 1.0           # 밴드 '핵심 구간'(원래 밴드)에 부여할 가중치
MARGIN_WEIGHT= 0.6           # 확장 영역(밴드 ±PRICE_MARGIN 중 핵심 외)에 부여할 가중치
EXTRA_STOPWORDS = []         # 사용자 추가 불용어

# ========= 1) 공통 유틸 & 데이터 로드 =========
import os, re, math, itertools
from collections import Counter
from typing import List, Tuple, Dict

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

os.makedirs(OUTDIR, exist_ok=True)

def pick_first_exist(paths):
    for p in paths:
        if p and os.path.exists(p):
            return p
    return None

summary_path = pick_first_exist([
    "./스윗스팟_분석_요약_통합_최저4000제외.csv",
    "./스윗스팟_분석_요약_통합.csv",
])
data_path = pick_first_exist([
    "./분석사용데이터_통합_최저4000제외.csv",
    "./분석사용데이터_통합.csv",
])
assert summary_path and data_path, "요약/데이터 CSV를 찾을 수 없습니다."

summary_df = pd.read_csv(summary_path, encoding="utf-8-sig")
data_df    = pd.read_csv(data_path,    encoding="utf-8-sig")

def parse_band_range(s: str):
    m = re.search(r"\[(\d+)\s*~\s*(\d+)\)", str(s))
    if not m: return None, None
    return int(m.group(1)), int(m.group(2))

bands: Dict[str, Tuple[int,int]] = {}
for i in (1, 2):
    row = summary_df[summary_df["밴드"] == f"Band {i}"]
    if len(row):
        L, R = parse_band_range(row.iloc[0]["구간(원)"])
        if L is not None:
            bands[f"Band {i}"] = (L, R)
assert bands, "밴드 정보가 없습니다. '밴드'와 '구간(원)' 컬럼을 확인하세요."

# ========= 2) Windows 한글 폰트 자동 탐색 & 적용 =========
import matplotlib
from matplotlib import font_manager
matplotlib.rcParams["axes.unicode_minus"] = False

def _candidate_paths_windows():
    win_dir = r"C:\Windows\Fonts"
    names_like = [
        "malgun",                 # 맑은 고딕
        "HCR", "Hancom", "hcr",   # 함초롬
        "gulim", "dotum", "batang", "gungsuh",
        "NanumGothic", "NanumSquare",
        "YuGoth", "msgothic"      # Yu Gothic, MS Gothic (있을 수도 있음)
    ]
    cands = []
    if os.path.isdir(win_dir):
        for fn in os.listdir(win_dir):
            lower = fn.lower()
            if any(key.lower() in lower for key in names_like):
                if lower.endswith((".ttf",".ttc",".otf")):
                    cands.append(os.path.join(win_dir, fn))
    def sort_key(p):
        pl = os.path.basename(p).lower()
        if "malgun" in pl: return (0, pl)
        if "hcr" in pl or "hancom" in pl: return (1, pl)
        if any(x in pl for x in ["gulim","dotum","batang","gungsuh"]): return (2, pl)
        if any(x in pl for x in ["nanum"]): return (3, pl)
        return (4, pl)
    cands.sort(key=sort_key)
    return cands

def pick_korean_font(font_path_hint=None):
    # 1) 직접 지정한 경로
    if font_path_hint and os.path.exists(font_path_hint):
        return font_path_hint
    # 2) Windows 후보 탐색
    for fp in _candidate_paths_windows():
        try:
            font_manager.fontManager.addfont(fp)
            return fp
        except Exception:
            continue
    # 3) 패밀리명으로 재탐색 (스크린샷 내 패밀리 반영)
    for fam in ["Malgun Gothic","Hancom Gothic","MS Gothic","Yu Gothic",
                "Gulim","Dotum","Batang","Gungsuh","NanumGothic","NanumSquare"]:
        try:
            p = font_manager.findfont(fam, fallback_to_default=False)
            if os.path.exists(p):
                font_manager.fontManager.addfont(p)
                return p
        except Exception:
            pass
    return None

font_used = pick_korean_font(FONT_PATH)
if font_used and os.path.exists(font_used):
    matplotlib.rcParams["font.family"] = [os.path.basename(font_used)]
else:
    # 경로를 못 잡아도 패밀리명으로 최대한 시도
    matplotlib.rcParams["font.family"] = ["Malgun Gothic","Hancom Gothic","MS Gothic","Yu Gothic",
                                          "Gulim","Dotum","Batang","Gungsuh","DejaVu Sans"]

def wordcloud_font_path():
    if font_used and os.path.exists(font_used):
        return font_used
    try:
        p = font_manager.findfont(matplotlib.rcParams["font.family"][0], fallback_to_default=False)
        if os.path.exists(p): return p
    except Exception:
        pass
    return None  # 없으면 워드클라우드 한글이 깨질 수 있음

print("사용 폰트 파일:", font_used)

# ========= 3) 형태소 분석기 / 불용어 / 토큰화 =========
analyzer_name = "Fallback(정규식)"
if USE_OKT:
    try:
        from konlpy.tag import Okt
        okt = Okt()
        analyzer_name = "KoNLPy/Okt"
        def extract_tokens(text: str):
            return [w for (w,t) in okt.pos(text, norm=True, stem=True) if t in ("Noun","Adjective")]
    except Exception:
        try:
            from konlpy.tag import Mecab
            mecab = Mecab()
            analyzer_name = "KoNLPy/Mecab"
            def extract_tokens(text: str):
                return [w for (w,t) in mecab.pos(text) if t.startswith("NN") or t.startswith("VA")]
        except Exception:
            analyzer_name = "Fallback(정규식)"
            def extract_tokens(text: str):
                if not isinstance(text, str): return []
                text = re.sub(r"[\(\[\{].*?[\)\]\}]", " ", text)
                text = re.sub(r"[^가-힣\s]", " ", text)
                text = re.sub(r"\s+", " ", text).strip()
                return [t for t in text.split() if len(t)>=2]
else:
    def extract_tokens(text: str):
        if not isinstance(text, str): return []
        text = re.sub(r"[\(\[\{].*?[\)\]\}]", " ", text)
        text = re.sub(r"[^가-힣\s]", " ", text)
        text = re.sub(r"\s+", " ", text).strip()
        return [t for t in text.split() if len(t)>=2]

BASE_STOP = set("""
도시락 식단 정기 구독 구독형 정기배송 세트 구성 옵션 특가 행사 증정 사은품 무료 배송 추천 인기 베스트
국내산 수입산 냉동 냉장 간편 간편식 건강 프리미엄 오리지널 플러스 기획 정품
맛 맛있 영양 칼로리 kcal 칼
ml g kg L l 팩 개 개입 입 봉 포 세트상품 박스 box BOX 세트형 구성품
혼합 모음 랜덤 선택 선택형 꾸러미 기본 일반
""".split())
BRAND_STOP = set("""
샐러디 노브랜드 오뚜기 CJ제일제당 CJ 풀무원 동원 대림 비비고 피코크 컬리 마켓컬리 코스트코 홈플러스 이마트 트레이더스 쿠팡
""".split())
UNIT_STOP = set("""
팩입 봉입 개입 인분 인용 미니 대용량 소용량 기본형 특대형 대형 소형 중형
""".split())

STOP = BASE_STOP | BRAND_STOP | UNIT_STOP | set(EXTRA_STOPWORDS)

def clean_tokens(tokens):
    out=[]
    for t in tokens:
        if t in STOP: continue
        if any(ch.isdigit() for ch in t): continue
        if len(t)<=1: continue
        out.append(t)
    return out

# ========= 4) 분석 함수 =========
def make_ngrams(tokens, n):
    return ["".join(tokens[i:i+n]) for i in range(len(tokens)-n+1)]

def build_cooccurrence(tokens_docs, min_freq=2):
    node, edge = Counter(), Counter()
    for toks in tokens_docs:
        uniq = sorted(set(toks))
        node.update(uniq)
        for a,b in itertools.combinations(uniq,2):
            edge[(a,b)] += 1
    node_df = pd.DataFrame(node.items(), columns=["토큰","빈도"]).sort_values("빈도", ascending=False)
    edge_df = pd.DataFrame([(a,b,c) for (a,b),c in edge.items() if c>=min_freq],
                           columns=["소스","타겟","빈도"]).sort_values("빈도", ascending=False)
    return node_df, edge_df

def tfidf_two_docs(tokens_docs_A, tokens_docs_B):
    def flat(docs): return " ".join([" ".join(t) for t in docs]).split()
    A, B = flat(tokens_docs_A), flat(tokens_docs_B)
    terms = sorted(set(A)|set(B))
    def vec(tokens):
        cnt = Counter(tokens); tot = sum(cnt.values()) if sum(cnt.values())>0 else 1
        vals = []
        for t in terms:
            tf = cnt.get(t,0)/tot
            df = int(t in set(A)) + int(t in set(B))
            idf = math.log((2+1)/(df+1)) + 1.0
            vals.append(tf*idf)
        return np.array(vals)
    return terms, vec(A), vec(B)

# ========= 5) 밴드별: 확장구간 포함 분석 + 가중 빈도 =========
def tokens_and_weights_for_band(df, L, R, margin=1500, core_w=1.0, margin_w=0.6):
    """
    확장구간 [L-margin, R+margin) 안의 상품만 사용.
    원래 밴드 [L, R) 는 core_w, 나머지(확장부)는 margin_w 가중치.
    """
    L_ext = max(0, L - margin)
    R_ext = R + margin
    sub = df[(df["1회 가격"] >= L_ext) & (df["1회 가격"] < R_ext)].copy()

    docs, weights = [], []
    for _, row in sub.iterrows():
        name = str(row["상품명"])
        price = float(row["1회 가격"])
        toks = clean_tokens(extract_tokens(name))
        docs.append(toks)
        if (price >= L) and (price < R):
            weights.append(core_w)
        else:
            weights.append(margin_w)
    return docs, weights, (L_ext, R_ext)

def top_freq_unweighted(tokens_docs, topn=100):
    cnt = Counter()
    for toks in tokens_docs: cnt.update(toks)
    return pd.DataFrame(cnt.most_common(topn), columns=["토큰","빈도"])

def top_freq_weighted(tokens_docs, weights, topn=100):
    acc = Counter()
    for toks, w in zip(tokens_docs, weights):
        for t in toks:
            acc[t] += float(w)
    items = sorted(acc.items(), key=lambda x: x[1], reverse=True)[:topn]
    return pd.DataFrame(items, columns=["토큰","가중빈도"])

def draw_bar(df, col, title, path_png, top=15):
    topdf = df.head(top)
    plt.figure(figsize=(8,6))
    plt.barh(topdf["토큰"][::-1], topdf[col][::-1])
    plt.xlabel(col)
    plt.title(title)
    plt.tight_layout()
    plt.savefig(path_png, dpi=200, bbox_inches="tight")
    plt.show()

# 실행
band_docs = {}
for band,(L,R) in bands.items():
    # 확장구간 포함 토큰/가중치
    docs, weights, (L_ext, R_ext) = tokens_and_weights_for_band(
        data_df, L, R, margin=PRICE_MARGIN, core_w=CORE_WEIGHT, margin_w=MARGIN_WEIGHT
    )
    band_docs[band] = docs

    # 1) (확장구간 기준) 일반 빈도 & 가중 빈도
    freq_df = top_freq_unweighted(docs, topn=100)
    wfreq_df = top_freq_weighted(docs, weights, topn=100)

    freq_df.to_csv(f"{OUTDIR}/{band}_토큰빈도(확장±{PRICE_MARGIN}).csv", index=False, encoding="utf-8-sig")
    wfreq_df.to_csv(f"{OUTDIR}/{band}_가중토큰빈도(핵심{CORE_WEIGHT}_확장{MARGIN_WEIGHT}).csv", index=False, encoding="utf-8-sig")

    print(f"[{band}] 확장구간: [{L_ext} ~ {R_ext}) / 핵심구간: [{L} ~ {R})  (CORE={CORE_WEIGHT}, MARGIN={MARGIN_WEIGHT})")
    display(freq_df.head(20))
    display(wfreq_df.head(20))

    # 2) 시각화 : 일반 빈도 / 가중 빈도
    draw_bar(freq_df, "빈도",
             f"{band} 토큰 상위 빈도 (확장 ±{PRICE_MARGIN}원)",
             f"{OUTDIR}/{band}_토큰빈도_막대(확장).png")

    draw_bar(wfreq_df, "가중빈도",
             f"{band} 토큰 상위 '가중' 빈도 (핵심>확장)",
             f"{OUTDIR}/{band}_가중토큰빈도_막대.png")

    # 3) 엔그램 (확장구간)
    bi, tri = Counter(), Counter()
    for toks in docs:
        bi.update(make_ngrams(toks, 2))
        tri.update(make_ngrams(toks, 3))
    pd.DataFrame(bi.most_common(100),  columns=["바이그램","빈도"]).to_csv(f"{OUTDIR}/{band}_바이그램(확장).csv",  index=False, encoding="utf-8-sig")
    pd.DataFrame(tri.most_common(100), columns=["트라이그램","빈도"]).to_csv(f"{OUTDIR}/{band}_트라이그램(확장).csv", index=False, encoding="utf-8-sig")

    # 4) 공동등장 네트워크 (확장구간)
    nodes, edges = build_cooccurrence(docs, min_freq=2)
    nodes.to_csv(f"{OUTDIR}/{band}_co_nodes(확장).csv", index=False, encoding="utf-8-sig")
    edges.to_csv(f"{OUTDIR}/{band}_co_edges(확장).csv", index=False, encoding="utf-8-sig")

    keep = list(nodes.head(30)["토큰"])
    if keep:
        angles = np.linspace(0, 2*np.pi, len(keep), endpoint=False)
        pos = {k:(np.cos(a), np.sin(a)) for k,a in zip(keep, angles)}
        plt.figure(figsize=(9,7))
        for k,(x,y) in pos.items():
            size = 250 + 45*int(nodes.set_index("토큰").loc[k,"빈도"])
            plt.scatter([x],[y], s=size); plt.text(x,y,k,ha="center",va="center")
        sub_e = edges[edges["소스"].isin(keep) & edges["타겟"].isin(keep)]
        for _,r in sub_e.iterrows():
            x1,y1 = pos[r["소스"]]; x2,y2 = pos[r["타겟"]]
            plt.plot([x1,x2],[y1,y2], alpha=0.25)
        plt.title(f"{band} 공동등장 네트워크 (확장구간)")
        plt.axis("off"); plt.tight_layout(); plt.show()
        plt.savefig(f"{OUTDIR}/{band}_공동등장_네트워크(확장).png", dpi=200, bbox_inches="tight")

    # 5) 워드클라우드(가능 시)
    try:
        from wordcloud import WordCloud
        wc_font = wordcloud_font_path()
        if wc_font:
            wc = WordCloud(width=1200, height=800, background_color="white", font_path=wc_font)
            wc_img = wc.generate_from_frequencies({r["토큰"]: int(r["빈도"]) for _, r in freq_df.iterrows()})
            plt.figure(figsize=(10,6)); plt.imshow(wc_img, interpolation="bilinear"); plt.axis("off")
            plt.title(f"{band} 워드클라우드 (확장구간)"); plt.tight_layout(); plt.show()
            wc.to_file(f"{OUTDIR}/{band}_워드클라우드(확장).png")
    except Exception as e:
        print("워드클라우드 생성 건너뜀:", e)

# ========= 6) 밴드 간 TF-IDF 상대 비교 (확장구간 기준) =========
labels = list(band_docs.keys())
if len(labels) >= 2:
    terms, v1, v2 = tfidf_two_docs(band_docs[labels[0]], band_docs[labels[1]])
    tfidf_df = pd.DataFrame({"토큰": terms, f"{labels[0]}": v1, f"{labels[1]}": v2, "차이(B1-B2)": v1-v2}) \
                .sort_values("차이(B1-B2)", ascending=False)
    display(tfidf_df.head(30))
    tfidf_df.to_csv(f"{OUTDIR}/TFIDF_밴드비교(확장).csv", index=False, encoding="utf-8-sig")

    top = tfidf_df.head(15)
    plt.figure(figsize=(8,6))
    plt.barh(top["토큰"][::-1], top["차이(B1-B2)"][::-1])
    plt.xlabel("TF-IDF 차이 (Band1 - Band2)"); plt.title("밴드 간 TF-IDF 상대 차이 Top 15 (확장)")
    plt.tight_layout(); plt.show()
    plt.savefig(f"{OUTDIR}/TFIDF_밴드비교_그래프(확장).png", dpi=200, bbox_inches="tight")

# ========= 7) 메타정보 저장 =========
pd.DataFrame([{
    "형태소 분석기": analyzer_name,
    "요약파일": os.path.basename(summary_path),
    "데이터파일": os.path.basename(data_path),
    "폰트": (font_used if font_used else "(패밀리 지정/폴백)"),
    "밴드 확장(±원)": PRICE_MARGIN,
    "가중치(핵심/확장)": f"{CORE_WEIGHT}/{MARGIN_WEIGHT}",
    "출력 폴더": os.path.abspath(OUTDIR)
}]).to_csv(f"{OUTDIR}/메타정보.csv", index=False, encoding="utf-8-sig")

print("✅ 완료! 결과 폴더:", os.path.abspath(OUTDIR))
```

```python
# -*- coding: utf-8 -*-
"""
도시락 밴드 1/2 형태소 분석 (폰트 Fix: 파일경로→패밀리명 설정, 밴드±1500/가중 그래프 포함)
"""

# ========= 0) 옵션 =========
USE_OKT = True
OUTDIR  = "./out"
FONT_PATH = r"C:\Windows\Fonts\malgun.ttf"   # << 여기 경로를 본인 PC에 맞게 지정!
PRICE_MARGIN = 1500
CORE_WEIGHT  = 1.0
MARGIN_WEIGHT= 0.6
EXTRA_STOPWORDS = []

# ========= 1) 공통 유틸 & 데이터 로드 =========
import os, re, math, itertools
from collections import Counter
from typing import List, Tuple, Dict

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

os.makedirs(OUTDIR, exist_ok=True)

def pick_first_exist(paths):
    for p in paths:
        if p and os.path.exists(p):
            return p
    return None

summary_path = pick_first_exist([
    "./스윗스팟_분석_요약_통합_최저4000제외.csv",
    "./스윗스팟_분석_요약_통합.csv",
])
data_path = pick_first_exist([
    "./분석사용데이터_통합_최저4000제외.csv",
    "./분석사용데이터_통합.csv",
])
assert summary_path and data_path, "요약/데이터 CSV를 찾을 수 없습니다."

summary_df = pd.read_csv(summary_path, encoding="utf-8-sig")
data_df    = pd.read_csv(data_path,    encoding="utf-8-sig")

def parse_band_range(s: str):
    m = re.search(r"\[(\d+)\s*~\s*(\d+)\)", str(s))
    if not m: return None, None
    return int(m.group(1)), int(m.group(2))

bands: Dict[str, Tuple[int,int]] = {}
for i in (1, 2):
    row = summary_df[summary_df["밴드"] == f"Band {i}"]
    if len(row):
        L, R = parse_band_range(row.iloc[0]["구간(원)"])
        if L is not None:
            bands[f"Band {i}"] = (L, R)
assert bands, "밴드 정보가 없습니다. '밴드'와 '구간(원)' 컬럼을 확인하세요."

# ========= 2) 폰트: 파일경로 → 패밀리명으로 지정 (핵심 수정) =========
import matplotlib
from matplotlib import font_manager as fm
matplotlib.rcParams["axes.unicode_minus"] = False

def set_korean_font(font_path_hint=None):
    # 1) 파일 경로가 있으면: addfont → 패밀리명 추출 → rcParams 설정
    if font_path_hint and os.path.exists(font_path_hint):
        fm.fontManager.addfont(font_path_hint)
        family = fm.FontProperties(fname=font_path_hint).get_name()
        matplotlib.rcParams["font.family"] = [family]
        # 워드클라우드용 경로는 그대로 파일 경로 사용
        return font_path_hint, family

    # 2) 패밀리명 폴백 (스크린샷 기반 우선순위)
    fallback_families = ["Malgun Gothic","Hancom Gothic","MS Gothic","Yu Gothic",
                         "Gulim","Dotum","Batang","Gungsuh","NanumGothic","DejaVu Sans"]
    matplotlib.rcParams["font.family"] = fallback_families
    return None, fallback_families[0]

WORDCLOUD_FONT_PATH, KOREAN_FAMILY = set_korean_font(FONT_PATH)
print("폰트 패밀리 설정:", matplotlib.rcParams["font.family"])
print("워드클라우드 폰트 경로:", WORDCLOUD_FONT_PATH or "(패밀리 기반)")

# ========= 3) 형태소 분석기 / 불용어 / 토큰화 =========
analyzer_name = "Fallback(정규식)"
if USE_OKT:
    try:
        from konlpy.tag import Okt
        okt = Okt()
        analyzer_name = "KoNLPy/Okt"
        def extract_tokens(text: str):
            return [w for (w,t) in okt.pos(text, norm=True, stem=True) if t in ("Noun","Adjective")]
    except Exception:
        try:
            from konlpy.tag import Mecab
            mecab = Mecab()
            analyzer_name = "KoNLPy/Mecab"
            def extract_tokens(text: str):
                return [w for (w,t) in mecab.pos(text) if t.startswith("NN") or t.startswith("VA")]
        except Exception:
            analyzer_name = "Fallback(정규식)"
            def extract_tokens(text: str):
                if not isinstance(text, str): return []
                text = re.sub(r"[\(\[\{].*?[\)\]\}]", " ", text)
                text = re.sub(r"[^가-힣\s]", " ", text)
                text = re.sub(r"\s+", " ", text).strip()
                return [t for t in text.split() if len(t)>=2]
else:
    def extract_tokens(text: str):
        if not isinstance(text, str): return []
        text = re.sub(r"[\(\[\{].*?[\)\]\}]", " ", text)
        text = re.sub(r"[^가-힣\s]", " ", text)
        text = re.sub(r"\s+", " ", text).strip()
        return [t for t in text.split() if len(t)>=2]

BASE_STOP = set("""
도시락 식단 정기 구독 구독형 정기배송 세트 구성 옵션 특가 행사 증정 사은품 무료 배송 추천 인기 베스트
국내산 수입산 냉동 냉장 간편 간편식 건강 프리미엄 오리지널 플러스 기획 정품
맛 맛있 영양 칼로리 kcal 칼
ml g kg L l 팩 개 개입 입 봉 포 세트상품 박스 box BOX 세트형 구성품
혼합 모음 랜덤 선택 선택형 꾸러미 기본 일반
""".split())
BRAND_STOP = set("""
샐러디 노브랜드 오뚜기 CJ제일제당 CJ 풀무원 동원 대림 비비고 피코크 컬리 마켓컬리 코스트코 홈플러스 이마트 트레이더스 쿠팡
""".split())
UNIT_STOP = set("""
팩입 봉입 개입 인분 인용 미니 대용량 소용량 기본형 특대형 대형 소형 중형
""".split())

STOP = BASE_STOP | BRAND_STOP | UNIT_STOP | set(EXTRA_STOPWORDS)

def clean_tokens(tokens):
    out=[]
    for t in tokens:
        if t in STOP: continue
        if any(ch.isdigit() for ch in t): continue
        if len(t)<=1: continue
        out.append(t)
    return out

# ========= 4) 분석 함수 =========
def make_ngrams(tokens, n):
    return ["".join(tokens[i:i+n]) for i in range(len(tokens)-n+1)]

def build_cooccurrence(tokens_docs, min_freq=2):
    node, edge = Counter(), Counter()
    for toks in tokens_docs:
        uniq = sorted(set(toks))
        node.update(uniq)
        for a,b in itertools.combinations(uniq,2):
            edge[(a,b)] += 1
    node_df = pd.DataFrame(node.items(), columns=["토큰","빈도"]).sort_values("빈도", ascending=False)
    edge_df = pd.DataFrame([(a,b,c) for (a,b),c in edge.items() if c>=min_freq],
                           columns=["소스","타겟","빈도"]).sort_values("빈도", ascending=False)
    return node_df, edge_df

def tfidf_two_docs(tokens_docs_A, tokens_docs_B):
    def flat(docs): return " ".join([" ".join(t) for t in docs]).split()
    A, B = flat(tokens_docs_A), flat(tokens_docs_B)
    terms = sorted(set(A)|set(B))
    def vec(tokens):
        cnt = Counter(tokens); tot = sum(cnt.values()) if sum(cnt.values())>0 else 1
        vals = []
        for t in terms:
            tf = cnt.get(t,0)/tot
            df = int(t in set(A)) + int(t in set(B))
            idf = math.log((2+1)/(df+1)) + 1.0
            vals.append(tf*idf)
        return np.array(vals)
    return terms, vec(A), vec(B)

# ========= 5) 밴드별: ±1500 포함 + 가중 빈도 =========
def tokens_and_weights_for_band(df, L, R, margin=1500, core_w=1.0, margin_w=0.6):
    L_ext = max(0, L - margin)
    R_ext = R + margin
    sub = df[(df["1회 가격"] >= L_ext) & (df["1회 가격"] < R_ext)].copy()

    docs, weights = [], []
    for _, row in sub.iterrows():
        name = str(row["상품명"])
        price = float(row["1회 가격"])
        toks = clean_tokens(extract_tokens(name))
        docs.append(toks)
        weights.append(core_w if (L <= price < R) else margin_w)
    return docs, weights, (L_ext, R_ext)

def top_freq_unweighted(tokens_docs, topn=100):
    cnt = Counter()
    for toks in tokens_docs: cnt.update(toks)
    return pd.DataFrame(cnt.most_common(topn), columns=["토큰","빈도"])

def top_freq_weighted(tokens_docs, weights, topn=100):
    acc = Counter()
    for toks, w in zip(tokens_docs, weights):
        for t in toks:
            acc[t] += float(w)
    items = sorted(acc.items(), key=lambda x: x[1], reverse=True)[:topn]
    return pd.DataFrame(items, columns=["토큰","가중빈도"])

def draw_bar(df, col, title, path_png, top=15):
    topdf = df.head(top)
    plt.figure(figsize=(8,6))
    plt.barh(topdf["토큰"][::-1], topdf[col][::-1])
    plt.xlabel(col)
    plt.title(title)
    plt.tight_layout()
    plt.savefig(path_png, dpi=200, bbox_inches="tight")
    plt.show()

# 실행
band_docs = {}
for band,(L,R) in bands.items():
    docs, weights, (L_ext, R_ext) = tokens_and_weights_for_band(
        data_df, L, R, margin=PRICE_MARGIN, core_w=CORE_WEIGHT, margin_w=MARGIN_WEIGHT
    )
    band_docs[band] = docs

    # 빈도 & 가중 빈도
    freq_df  = top_freq_unweighted(docs, topn=100)
    wfreq_df = top_freq_weighted(docs, weights, topn=100)

    freq_df.to_csv(f"{OUTDIR}/{band}_토큰빈도(확장±{PRICE_MARGIN}).csv", index=False, encoding="utf-8-sig")
    wfreq_df.to_csv(f"{OUTDIR}/{band}_가중토큰빈도(핵심{CORE_WEIGHT}_확장{MARGIN_WEIGHT}).csv", index=False, encoding="utf-8-sig")

    print(f"[{band}] 확장구간: [{L_ext} ~ {R_ext}) / 핵심구간: [{L} ~ {R})  (CORE={CORE_WEIGHT}, MARGIN={MARGIN_WEIGHT})")
    display(freq_df.head(20)); display(wfreq_df.head(20))

    draw_bar(freq_df,  "빈도",
             f"{band} 토큰 상위 빈도 (확장 ±{PRICE_MARGIN}원)",
             f"{OUTDIR}/{band}_토큰빈도_막대(확장).png")
    draw_bar(wfreq_df, "가중빈도",
             f"{band} 토큰 상위 '가중' 빈도 (핵심>확장)",
             f"{OUTDIR}/{band}_가중토큰빈도_막대.png")

    # 엔그램
    bi, tri = Counter(), Counter()
    for toks in docs:
        bi.update(["".join(toks[i:i+2]) for i in range(len(toks)-1)])
        tri.update(["".join(toks[i:i+3]) for i in range(len(toks)-2)])
    pd.DataFrame(bi.most_common(100),  columns=["바이그램","빈도"]).to_csv(f"{OUTDIR}/{band}_바이그램(확장).csv",  index=False, encoding="utf-8-sig")
    pd.DataFrame(tri.most_common(100), columns=["트라이그램","빈도"]).to_csv(f"{OUTDIR}/{band}_트라이그램(확장).csv", index=False, encoding="utf-8-sig")

    # 공동등장 네트워크
    nodes, edges = build_cooccurrence(docs, min_freq=2)
    nodes.to_csv(f"{OUTDIR}/{band}_co_nodes(확장).csv", index=False, encoding="utf-8-sig")
    edges.to_csv(f"{OUTDIR}/{band}_co_edges(확장).csv", index=False, encoding="utf-8-sig")

    keep = list(nodes.head(30)["토큰"])
    if keep:
        angles = np.linspace(0, 2*np.pi, len(keep), endpoint=False)
        pos = {k:(np.cos(a), np.sin(a)) for k,a in zip(keep, angles)}
        plt.figure(figsize=(9,7))
        for k,(x,y) in pos.items():
            size = 250 + 45*int(nodes.set_index("토큰").loc[k,"빈도"])
            plt.scatter([x],[y], s=size)
            plt.text(x,y,k,ha="center",va="center")
        sub_e = edges[edges["소스"].isin(keep) & edges["타겟"].isin(keep)]
        for _,r in sub_e.iterrows():
            x1,y1 = pos[r["소스"]]; x2,y2 = pos[r["타겟"]]
            plt.plot([x1,x2],[y1,y2], alpha=0.25)
        plt.title(f"{band} 공동등장 네트워크 (확장구간)")
        plt.axis("off"); plt.tight_layout(); plt.show()
        plt.savefig(f"{OUTDIR}/{band}_공동등장_네트워크(확장).png", dpi=200, bbox_inches="tight")

    # 워드클라우드
    try:
        from wordcloud import WordCloud
        wc = WordCloud(width=1200, height=800, background_color="white",
                       font_path=(WORDCLOUD_FONT_PATH or FONT_PATH))
        freq_map = {r["토큰"]: int(r["빈도"]) for _, r in freq_df.iterrows()}
        wc_img = wc.generate_from_frequencies(freq_map)
        plt.figure(figsize=(10,6)); plt.imshow(wc_img, interpolation="bilinear"); plt.axis("off")
        plt.title(f"{band} 워드클라우드 (확장구간)"); plt.tight_layout(); plt.show()
        wc.to_file(f"{OUTDIR}/{band}_워드클라우드(확장).png")
    except Exception as e:
        print("워드클라우드 생성 건너뜀:", e)

# ========= 6) 밴드 간 TF-IDF (확장구간) =========
def tfidf_two_docs(tokens_docs_A, tokens_docs_B):
    def flat(docs): return " ".join([" ".join(t) for t in docs]).split()
    A, B = flat(tokens_docs_A), flat(tokens_docs_B)
    terms = sorted(set(A)|set(B))
    def vec(tokens):
        cnt = Counter(tokens); tot = sum(cnt.values()) if sum(cnt.values())>0 else 1
        vals = []
        for t in terms:
            tf = cnt.get(t,0)/tot
            df = int(t in set(A)) + int(t in set(B))
            idf = math.log((2+1)/(df+1)) + 1.0
            vals.append(tf*idf)
        return np.array(vals)
    return terms, vec(A), vec(B)

labels = list(band_docs.keys())
if len(labels) >= 2:
    terms, v1, v2 = tfidf_two_docs(band_docs[labels[0]], band_docs[labels[1]])
    tfidf_df = pd.DataFrame({"토큰": terms, f"{labels[0]}": v1, f"{labels[1]}": v2, "차이(B1-B2)": v1-v2}) \
                .sort_values("차이(B1-B2)", ascending=False)
    display(tfidf_df.head(30))
    tfidf_df.to_csv(f"{OUTDIR}/TFIDF_밴드비교(확장).csv", index=False, encoding="utf-8-sig")

    top = tfidf_df.head(15)
    plt.figure(figsize=(8,6))
    plt.barh(top["토큰"][::-1], top["차이(B1-B2)"][::-1])
    plt.xlabel("TF-IDF 차이 (Band1 - Band2)"); plt.title("밴드 간 TF-IDF 상대 차이 Top 15 (확장)")
    plt.tight_layout(); plt.show()
    plt.savefig(f"{OUTDIR}/TFIDF_밴드비교_그래프(확장).png", dpi=200, bbox_inches="tight")

# ========= 7) 메타 =========
pd.DataFrame([{
    "형태소 분석기": analyzer_name,
    "요약파일": os.path.basename(summary_path),
    "데이터파일": os.path.basename(data_path),
    "폰트패밀리": matplotlib.rcParams["font.family"],
    "워드클라우드폰트": WORDCLOUD_FONT_PATH or "(패밀리 기반)",
    "밴드 확장(±원)": PRICE_MARGIN,
    "가중치(핵심/확장)": f"{CORE_WEIGHT}/{MARGIN_WEIGHT}",
    "출력 폴더": os.path.abspath(OUTDIR)
}]).to_csv(f"{OUTDIR}/메타정보.csv", index=False, encoding="utf-8-sig")

print("✅ 완료! 결과 폴더:", os.path.abspath(OUTDIR))
```

```python
# -*- coding: utf-8 -*-
# 밴드1/밴드2 토큰빈도 CSV → 밴드 내 비율(%)로 비교 · 시각화 3종
# - 파일 인코딩: utf-8-sig 가정
# - 그래프 폰트: Windows/맥/나눔/Noto/유고딕 자동 탐색(경로 직접 지정 가능)

import os, glob
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import font_manager as fm, rcParams

# ------------------------------------------------------------
# 0) 경로 · 옵션
# ------------------------------------------------------------
BASE_DIR = "."                 # CSV가 있는 폴더
OUT_DIR  = "./out_band_compare_ratio"
os.makedirs(OUT_DIR, exist_ok=True)

TOP_COMBINED = 20   # 결합(=Band1+Band2) 빈도 상위 N개 토큰을 시각화 대상으로 선정

# (필요시) 시스템 폰트 경로 직접 지정 (예: r"C:\Windows\Fonts\malgun.ttf")
FONT_PATH_HINT = r"C:\Windows\Fonts\malgun.ttf"

# ------------------------------------------------------------
# 1) 한글 폰트 설정
# ------------------------------------------------------------
def set_korean_font(font_path_hint=FONT_PATH_HINT):
    try:
        family = None
        if font_path_hint and os.path.exists(font_path_hint):
            fm.fontManager.addfont(font_path_hint)
            family = fm.FontProperties(fname=font_path_hint).get_name()
        if family:
            rcParams["font.family"] = [family]
        else:
            # 시스템에 존재 가능성이 높은 패밀리 후보
            rcParams["font.family"] = [
                "Malgun Gothic", "AppleGothic", "NanumGothic",
                "Noto Sans CJK KR", "Hancom Gothic", "Yu Gothic",
                "DejaVu Sans"
            ]
        rcParams["axes.unicode_minus"] = False
        # 폰트 캐시 리로드
        fm._load_fontmanager(try_read_cache=False)
    except Exception as e:
        print("⚠ 폰트 설정 실패:", e)

set_korean_font()

# ------------------------------------------------------------
# 2) 데이터 로드 (가장 최신 파일 자동 선택)
#    예상 파일명:
#    - Band 1_토큰빈도(확장±1500).csv
#    - Band 2_토큰빈도(확장±1500).csv
# ------------------------------------------------------------
def find_latest(pattern, base=BASE_DIR):
    paths = glob.glob(os.path.join(base, pattern))
    if not paths:
        return None
    paths.sort(key=os.path.getmtime, reverse=True)
    return paths[0]

b1_path = find_latest("Band 1_토큰빈도*.csv")
b2_path = find_latest("Band 2_토큰빈도*.csv")
if not (b1_path and b2_path):
    raise FileNotFoundError("CSV 파일을 찾지 못했습니다. 'Band 1_토큰빈도*.csv', 'Band 2_토큰빈도*.csv' 확인")

b1 = pd.read_csv(b1_path, encoding="utf-8-sig")
b2 = pd.read_csv(b2_path, encoding="utf-8-sig")

# 첫 두 컬럼을 표준화
b1 = b1.rename(columns={b1.columns[0]: "토큰", b1.columns[1]: "빈도"})
b2 = b2.rename(columns={b2.columns[0]: "토큰", b2.columns[1]: "빈도"})

# ------------------------------------------------------------
# 3) 밴드별 비율(%) 계산
#    B1_% = (빈도_B1 / Σ빈도_B1)×100, B2_% = (빈도_B2 / Σ빈도_B2)×100
# ------------------------------------------------------------
tot1 = max(float(b1["빈도"].sum()), 1.0)
tot2 = max(float(b2["빈도"].sum()), 1.0)

df = pd.merge(b1, b2, on="토큰", how="outer", suffixes=("_B1","_B2")).fillna(0)
df["빈도_B1"] = df["빈도_B1"].astype(float)
df["빈도_B2"] = df["빈도_B2"].astype(float)
df["B1_%"] = (df["빈도_B1"] / tot1 * 100).round(3)
df["B2_%"] = (df["빈도_B2"] / tot2 * 100).round(3)
df["combined"] = df["빈도_B1"] + df["빈도_B2"]

# 결과 저장(표준화 테이블)
cmp_csv = os.path.join(OUT_DIR, "밴드1_vs_밴드2_토큰_사용비율_비교(정규화).csv")
df[["토큰","빈도_B1","빈도_B2","B1_%","B2_%","combined"]].sort_values(
    "combined", ascending=False
).to_csv(cmp_csv, index=False, encoding="utf-8-sig")

# ------------------------------------------------------------
# 4) 시각화 (절댓값/차이 기반 X —> 오직 비율 기반)
#    (A) 나란히 막대: B1_% vs B2_% (결합빈도 Top N)
#    (B) 100% 스택: 각 토큰 내 Band1/2 분할 비율
#    (C) 산점도: B1_% vs B2_% (점 크기=결합빈도)
# ------------------------------------------------------------
top = df.sort_values("combined", ascending=False).head(TOP_COMBINED).copy()

# (A) 나란히 막대
x = np.arange(len(top))
width = 0.4
plt.figure(figsize=(12,7))
plt.bar(x - width/2, top["B1_%"], width, label="Band 1")
plt.bar(x + width/2, top["B2_%"], width, label="Band 2")
plt.xticks(x, top["토큰"], rotation=45, ha="right")
plt.ylabel("사용 비율(%)  — 밴드 내 토큰 점유율")
plt.title(f"밴드별 토큰 사용 비율 비교 (결합 빈도 Top {TOP_COMBINED})")
plt.legend()
plt.tight_layout()
plt.savefig(os.path.join(OUT_DIR, f"키워드_비율비교_나란히_Top{TOP_COMBINED}.png"), dpi=200, bbox_inches="tight")
plt.show()

# (B) 100% 스택
top["token_share_B1_%"] = (top["빈도_B1"] / top["combined"].replace(0, np.nan) * 100).round(2)
top["token_share_B2_%"] = 100 - top["token_share_B1_%"]
plt.figure(figsize=(12,7))
bottom = np.zeros(len(top))
plt.bar(np.arange(len(top)), top["token_share_B1_%"], label="Band 1")
bottom = top["token_share_B1_%"]
plt.bar(np.arange(len(top)), top["token_share_B2_%"], bottom=bottom, label="Band 2")
plt.xticks(np.arange(len(top)), top["토큰"], rotation=45, ha="right")
plt.ylabel("해당 토큰 내 밴드 비율(%) — 100% Stacked")
plt.title(f"토큰 기준 밴드 분할 비율 (결합 빈도 Top {TOP_COMBINED})")
plt.legend()
plt.tight_layout()
plt.savefig(os.path.join(OUT_DIR, f"키워드_토큰기준_분할비율_Stacked_Top{TOP_COMBINED}.png"), dpi=200, bbox_inches="tight")
plt.show()

# (C) 산점도
plt.figure(figsize=(8,7))
sizes = (top["combined"] / top["combined"].max() * 1200).fillna(100)
plt.scatter(top["B1_%"], top["B2_%"], s=sizes)
for _, r in top.iterrows():
    plt.text(r["B1_%"], r["B2_%"], str(r["토큰"]), ha="center", va="center")
lim = max(top["B1_%"].max(), top["B2_%"].max()) * 1.1
plt.plot([0, lim], [0, lim])  # 대각선
plt.xlim(0, lim); plt.ylim(0, lim)
plt.xlabel("Band 1 사용 비율(%)")
plt.ylabel("Band 2 사용 비율(%)")
plt.title(f"B1% vs B2% 산점도 (점 크기=결합 빈도, Top {TOP_COMBINED})")
plt.tight_layout()
plt.savefig(os.path.join(OUT_DIR, f"키워드_B1vsB2_산점도_Top{TOP_COMBINED}.png"), dpi=200, bbox_inches="tight")
plt.show()

print("완료! 폴더:", os.path.abspath(OUT_DIR))
```

```python
# -*- coding: utf-8 -*-
# 밴드별(1/2) 주요 키워드 중심 "키워드-키워드 상관관계" 분석/시각화
# - 입력: Band 1/2_co_nodes(확장).csv, Band 1/2_co_edges(확장).csv
#   (컬럼명: nodes -> ["토큰","빈도"], edges -> ["소스","타겟","빈도"])
# - 선택 입력(있으면 정확도 ↑): 분석사용데이터_통합.csv, 스윗스팟_분석_요약_통합*.csv
#   -> 밴드 가격 구간을 읽어 문서수(N) 산출 (없으면 보수적 추정치 사용)
# - 출력: out_band_corr 폴더에 CSV/PNG 저장

import os, glob, re, math
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import font_manager as fm, rcParams

# ------------------------------ 옵션 ------------------------------
BASE_DIR = "."                     # 데이터가 있는 폴더
OUT_DIR  = "./out_band_corr"
os.makedirs(OUT_DIR, exist_ok=True)

# 상위 주요 키워드 선택 개수(밴드별 node 빈도 기준, 두 밴드의 합집합으로 선정)
TOP_MAIN_TOKENS = 25
# 네트워크/히트맵에 반영할 엣지 최소 공동등장(문서) 수
EDGE_MIN_CO = 2
# 히트맵/네트워크 가중치로 사용할 지표("npmi" 또는 "cosine" 또는 "jaccard")
EDGE_SCORE = "npmi"

# 폰트(Windows 권장 경로) — 환경에 맞게 자동/수동 지정
FONT_PATH_HINT = r"C:\Windows\Fonts\malgun.ttf"

# 밴드 확장 폭(스윗스팟 가격 경계 ±)
BAND_PAD = 1500

# ------------------------------ 한글 폰트 ------------------------------
def set_korean_font(font_path_hint=FONT_PATH_HINT):
    try:
        family = None
        if font_path_hint and os.path.exists(font_path_hint):
            fm.fontManager.addfont(font_path_hint)
            family = fm.FontProperties(fname=font_path_hint).get_name()
        if family:
            rcParams["font.family"] = [family]
        else:
            rcParams["font.family"] = ["Malgun Gothic","AppleGothic",
                                       "NanumGothic","Noto Sans CJK KR",
                                       "Hancom Gothic","Yu Gothic","DejaVu Sans"]
        rcParams["axes.unicode_minus"] = False
        fm._load_fontmanager(try_read_cache=False)
        print("✓ 폰트:", rcParams["font.family"])
    except Exception as e:
        print("⚠ 폰트 설정 실패:", e)

set_korean_font()

# ------------------------------ 유틸 ------------------------------
def find_latest(pattern, base=BASE_DIR):
    paths = glob.glob(os.path.join(base, pattern))
    if not paths:
        return None
    paths.sort(key=os.path.getmtime, reverse=True)
    return paths[0]

def parse_band_range(s: str):
    # 예: "[4500 ~ 5900)"
    m = re.search(r"\[(\d+)\s*~\s*(\d+)\)", str(s))
    if not m: return None, None
    return int(m.group(1)), int(m.group(2))

def load_nodes_edges(prefix):  # prefix: "Band 1" / "Band 2"
    n_path = find_latest(f"{prefix}_co_nodes*.csv")
    e_path = find_latest(f"{prefix}_co_edges*.csv")
    if not (n_path and e_path):
        raise FileNotFoundError(f"{prefix} nodes/edges CSV를 찾을 수 없습니다.")
    nodes = pd.read_csv(n_path, encoding="utf-8-sig")
    edges = pd.read_csv(e_path, encoding="utf-8-sig")
    nodes = nodes.rename(columns={nodes.columns[0]:"토큰", nodes.columns[1]:"빈도"})
    edges = edges.rename(columns={edges.columns[0]:"소스", edges.columns[1]:"타겟", edges.columns[2]:"빈도"})
    return nodes, edges, n_path, e_path

# ------------------------------ 밴드 문서수(N) 산출(선택) ------------------------------
def estimate_doc_count_from_price(prefix):
    """분석사용데이터_통합.csv + 스윗스팟 요약이 있으면
       가격 밴드(±BAND_PAD)로 문서 수(N)을 추정. 실패 시 None 반환."""
    data_path = find_latest("분석사용데이터_통합*.csv")
    sum_path  = find_latest("스윗스팟_분석_요약_통합*.csv")
    if not (data_path and sum_path):
        return None

    df_data = pd.read_csv(data_path, encoding="utf-8-sig")
    df_sum  = pd.read_csv(sum_path,  encoding="utf-8-sig")
    # 컬럼명 표준화 가정: "밴드", "구간(원)" / "1회 가격"
    row = df_sum[df_sum["밴드"]==prefix.replace("_"," ")].copy()
    if row.empty:  # "Band 1" 형태 대응
        row = df_sum[df_sum["밴드"]==prefix]
    if row.empty: 
        return None
    L, R = parse_band_range(row.iloc[0]["구간(원)"])
    if L is None: 
        return None
    Lp, Rp = max(L-BAND_PAD, 0), R+BAND_PAD

    # 숫자화
    p = pd.to_numeric(df_data["1회 가격"], errors="coerce")
    N = int(((p>=Lp)&(p<Rp)).sum())
    return N if N>0 else None

# ------------------------------ 상관/연관 지표 ------------------------------
def compute_assoc_table(nodes, edges, main_tokens, N_est=None):
    """nodes: (토큰,빈도=df), edges:(소스,타겟,빈도=co), main_tokens: 관심 키워드 리스트
       N_est: 문서 수(없으면 보수적 추정)"""
    nodes = nodes.copy(); edges = edges.copy()
    nodes["df"] = nodes["빈도"].astype(float)
    edges["co"] = edges["빈도"].astype(float)

    # 문서 수 추정 (없으면 df의 최대값으로 보수적 추정)
    if not N_est:
        N_est = int(nodes["df"].max()) if len(nodes) else 1

    df_map = nodes.set_index("토큰")["df"].to_dict()
    # (i,j) → co
    co_map = {(r["소스"], r["타겟"]): r["co"] for _,r in edges.iterrows()}
    co_map.update({(r["타겟"], r["소스"]): r["co"] for _,r in edges.iterrows()})

    recs = []
    toks = [t for t in main_tokens if t in df_map]
    for i in range(len(toks)):
        for j in range(i+1, len(toks)):
            a, b = toks[i], toks[j]
            co = co_map.get((a,b), 0.0)
            if co < EDGE_MIN_CO: 
                continue
            dfa, dfb = df_map.get(a, 0.0), df_map.get(b, 0.0)

            # 확률
            p_a  = dfa / N_est if N_est>0 else 0
            p_b  = dfb / N_est if N_est>0 else 0
            p_ab = co  / N_est if N_est>0 else 0

            # Cosine(문서공간 근사), Jaccard
            cosine  = co / (math.sqrt(dfa*dfb) + 1e-9)
            jaccard = co / (dfa + dfb - co + 1e-9)

            # PMI / NPMI (p_ab==0이면 제외)
            if p_ab > 0 and p_a>0 and p_b>0:
                pmi  = math.log2(p_ab / (p_a*p_b))
                npmi = pmi / (-math.log2(p_ab))
            else:
                pmi, npmi = float("-inf"), 0.0

            recs.append([a,b,int(co),int(dfa),int(dfb),cosine,jaccard,pmi,npmi])

    cols = ["토큰A","토큰B","co","dfA","dfB","cosine","jaccard","pmi","npmi"]
    out  = pd.DataFrame(recs, columns=cols)
    return out.sort_values(EDGE_SCORE, ascending=False), N_est

def choose_main_tokens(nodes1, nodes2, k=TOP_MAIN_TOKENS):
    n1 = nodes1.sort_values("빈도", ascending=False).head(k)["토큰"]
    n2 = nodes2.sort_values("빈도", ascending=False).head(k)["토큰"]
    return list(pd.unique(pd.concat([n1, n2], ignore_index=True)))

# ------------------------------ 데이터 로드 ------------------------------
n1, e1, n1_path, e1_path = load_nodes_edges("Band 1")
n2, e2, n2_path, e2_path = load_nodes_edges("Band 2")
print("Band1:", os.path.basename(n1_path), "|", os.path.basename(e1_path))
print("Band2:", os.path.basename(n2_path), "|", os.path.basename(e2_path))

main_tokens = choose_main_tokens(n1, n2, k=TOP_MAIN_TOKENS)
pd.Series(main_tokens, name="주요키워드").to_csv(
    os.path.join(OUT_DIR, "주요키워드_선정.csv"), index=False, encoding="utf-8-sig"
)

# 밴드별 문서 수(있으면 정확히 산출)
N1 = estimate_doc_count_from_price("Band 1")
N2 = estimate_doc_count_from_price("Band 2")

assoc1, N1 = compute_assoc_table(n1, e1, main_tokens, N_est=N1)
assoc2, N2 = compute_assoc_table(n2, e2, main_tokens, N_est=N2)

assoc1.to_csv(os.path.join(OUT_DIR, f"Band1_상관표_{EDGE_SCORE}.csv"), index=False, encoding="utf-8-sig")
assoc2.to_csv(os.path.join(OUT_DIR, f"Band2_상관표_{EDGE_SCORE}.csv"), index=False, encoding="utf-8-sig")
print(f"N1={N1}, N2={N2} (문서 수 추정/계산)")

# ------------------------------ 히트맵 ------------------------------
def heatmap_from_assoc(df_assoc, tokens, title, score=EDGE_SCORE, fname="heatmap.png"):
    toks = tokens
    idx = {t:i for i,t in enumerate(toks)}
    M = np.zeros((len(toks), len(toks)), dtype=float)
    for _,r in df_assoc.iterrows():
        a,b,val = r["토큰A"], r["토큰B"], r[score]
        if a in idx and b in idx:
            i,j = idx[a], idx[b]
            M[i,j] = M[j,i] = val

    plt.figure(figsize=(1.0+0.35*len(toks), 1.0+0.35*len(toks)))
    im = plt.imshow(M, cmap="coolwarm", vmin=np.min(M), vmax=np.max(M))
    plt.xticks(range(len(toks)), toks, rotation=90)
    plt.yticks(range(len(toks)), toks)
    plt.colorbar(im, fraction=0.046, pad=0.04, label=score.upper())
    plt.title(title)
    plt.tight_layout()
    path = os.path.join(OUT_DIR, fname)
    plt.savefig(path, dpi=220, bbox_inches="tight"); plt.show()
    print("✓ 저장:", path)

# 상위 토큰(히트맵에 너무 많으면 읽기 어려우므로 15~20 권장)
HM_TOP = min(20, len(main_tokens))
hm_tokens = main_tokens[:HM_TOP]
heatmap_from_assoc(assoc1, hm_tokens, f"Band 1 키워드 상관 히트맵 ({EDGE_SCORE})",
                   fname=f"Band1_heatmap_{EDGE_SCORE}.png")
heatmap_from_assoc(assoc2, hm_tokens, f"Band 2 키워드 상관 히트맵 ({EDGE_SCORE})",
                   fname=f"Band2_heatmap_{EDGE_SCORE}.png")

# ------------------------------ 네트워크(간단) ------------------------------
def network_plot(df_assoc, nodes_df, tokens, title, score=EDGE_SCORE, fname="network.png",
                 keep_top_edges=150, min_score=None):
    # 점수 높은 엣지 상위 N 또는 임계값으로 필터
    df = df_assoc.copy()
    if min_score is not None:
        df = df[df[score] >= min_score]
    df = df.sort_values(score, ascending=False)
    if keep_top_edges:
        df = df.head(keep_top_edges)

    toks = list(tokens)
    # 좌표: 원형
    theta = np.linspace(0, 2*np.pi, len(toks), endpoint=False)
    pos = {t:(np.cos(a), np.sin(a)) for t,a in zip(toks, theta)}
    size_map = nodes_df.set_index("토큰")["빈도"].to_dict()

    plt.figure(figsize=(10,10))
    # 엣지
    if not df.empty:
        smax = df[score].max()
        for _,r in df.iterrows():
            a,b,w = r["토큰A"], r["토큰B"], r[score]
            if a in pos and b in pos:
                x1,y1 = pos[a]; x2,y2 = pos[b]
                lw = 0.5 + 3.5*(w / (smax if smax>0 else 1))
                plt.plot([x1,x2],[y1,y2], alpha=0.25, linewidth=lw)

    # 노드
    for t,(x,y) in pos.items():
        s = 250 + 25*size_map.get(t, 1)
        plt.scatter([x],[y], s=s)
        plt.text(x,y,t, ha="center", va="center")

    plt.title(title)
    plt.axis("off"); plt.tight_layout()
    path = os.path.join(OUT_DIR, fname)
    plt.savefig(path, dpi=220, bbox_inches="tight"); plt.show()
    print("✓ 저장:", path)

network_plot(assoc1, n1, hm_tokens, f"Band 1 키워드 네트워크 ({EDGE_SCORE})",
             fname=f"Band1_network_{EDGE_SCORE}.png", keep_top_edges=120)
network_plot(assoc2, n2, hm_tokens, f"Band 2 키워드 네트워크 ({EDGE_SCORE})",
             fname=f"Band2_network_{EDGE_SCORE}.png", keep_top_edges=120)

# ------------------------------ 밴드 간 차이 비교 ------------------------------
# 동일 쌍에 대해 점수 차이(예: NPMI_B1 - NPMI_B2)
key = ["토큰A","토큰B"]
merged = pd.merge(assoc1[key+[EDGE_SCORE]].rename(columns={EDGE_SCORE:"B1"}),
                  assoc2[key+[EDGE_SCORE]].rename(columns={EDGE_SCORE:"B2"}),
                  on=key, how="outer").fillna(0.0)
merged["차이(B1-B2)"] = merged["B1"] - merged["B2"]
merged_sorted = merged.sort_values("차이(B1-B2)", ascending=False)

merged_sorted.to_csv(os.path.join(OUT_DIR, f"밴드간_{EDGE_SCORE}_차이.csv"),
                     index=False, encoding="utf-8-sig")

# 상위/하위 15개 바차트
TOP_DIFF = 15
def bar_top_diff(df, title, fname):
    top_pos = df.head(TOP_DIFF)
    top_neg = df.tail(TOP_DIFF)
    fig, axes = plt.subplots(1,2, figsize=(16,7), sharex=False)
    axes[0].barh(top_pos.apply(lambda r: f"{r['토큰A']}-{r['토큰B']}", axis=1)[::-1],
                 top_pos["차이(B1-B2)"][::-1])
    axes[0].set_title("Band1 > Band2 (Top)")
    axes[0].set_xlabel(f"{EDGE_SCORE.upper()} 차이")
    axes[1].barh(top_neg.apply(lambda r: f"{r['토큰A']}-{r['토큰B']}", axis=1),
                 top_neg["차이(B1-B2)"])
    axes[1].set_title("Band2 > Band1 (Top)")
    axes[1].set_xlabel(f"{EDGE_SCORE.upper()} 차이")
    plt.suptitle(title)
    plt.tight_layout()
    path = os.path.join(OUT_DIR, fname)
    plt.savefig(path, dpi=220, bbox_inches="tight"); plt.show()
    print("✓ 저장:", path)

bar_top_diff(merged_sorted, f"밴드 간 {EDGE_SCORE.upper()} 차이 상하위", 
             f"밴드간_{EDGE_SCORE}_차이_상하위.png")

print("완료! 결과 폴더:", os.path.abspath(OUT_DIR))
```

```python
# -*- coding: utf-8 -*-
# ◼︎ 도시락 스윗스팟 분석 (Band 1/2) — Jupyter에서 바로 실행
# - 세 CSV를 통합해 표본을 늘리고(중복 제거), 가중점수(평점×리뷰가중)로
#   가격대별 가치지수(가중점수합/가격합)를 계산해 최적 2개 밴드 도출/시각화
# - 이상치: 1) 1회 가격 ≤ 4000 원, 2) IQR 상단 초과값 제거
# - 모든 저장 파일은 utf-8-sig

import os, math, glob, textwrap
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import font_manager as fm, rcParams
from IPython.display import display

# =========================
# 0) 한글 폰트 안정 설정
# =========================
def set_korean_font():
    """
    Windows에서 한글 깨짐 방지:
    - 'C:/Windows/Fonts/malgun.ttf'를 우선 시도(가장 안정적)
    - 실패 시 시스템에 등록된 한글 폰트 후보를 순서대로 시도
    """
    candidates_path = [
        r"C:\Windows\Fonts\malgun.ttf",         # 맑은 고딕
        r"C:\Windows\Fonts\H2GTRM.TTF",         # 한컴 굴림(예시)
        r"C:\Windows\Fonts\H2HDRM.TTF",         # 한컴 돋움(예시)
    ]
    candidates_family = [
        "Malgun Gothic", "Hancom Gothic", "HY Gulim", "HY Dotum",
        "AppleGothic", "NanumGothic", "Noto Sans CJK KR", "Yu Gothic",
        "MS Gothic", "DejaVu Sans"
    ]

    family_set = None
    # 1) 경로로 로딩
    for p in candidates_path:
        if os.path.exists(p):
            try:
                fm.fontManager.addfont(p)
                fam = fm.FontProperties(fname=p).get_name()
                rcParams["font.family"] = [fam]
                family_set = fam
                break
            except Exception:
                pass
    # 2) 패밀리명으로 설정 시도
    if family_set is None:
        rcParams["font.family"] = candidates_family

    rcParams["axes.unicode_minus"] = False
    # 캐시 리로드
    try:
        fm._load_fontmanager(try_read_cache=False)
    except Exception:
        pass
    print("폰트 사용:", rcParams["font.family"])

set_korean_font()

# =========================
# 1) 파일 로드 & 통합
# =========================
def read_csv_safely(path):
    try:
        return pd.read_csv(path, encoding="utf-8-sig")
    except Exception:
        return pd.read_csv(path)

def to_num(s):
    return pd.to_numeric(s, errors="coerce")

def find_col(df, candidates):
    for c in candidates:
        if c in df.columns:
            return c
    return None

# 사용할 파일 후보(경로 수정 가능)
file_candidates = [
    "dosirak_scored.csv",
    "dosirak_from_text.csv",
    "도시락_정기구독_상품_정리_from_txt_enriched.csv",
    "dosirak_unit_scored.csv",    # 있으면 자동 포함
]

frames, loaded = [], []
for name in file_candidates:
    for p in glob.glob(name) + glob.glob(os.path.join(".", name)) + glob.glob(os.path.join("..", name)):
        try:
            df = read_csv_safely(p)
            df["__src__"] = os.path.basename(p)
            frames.append(df)
            loaded.append((os.path.basename(p), len(df)))
        except Exception:
            pass

assert frames, "CSV를 찾지 못했습니다. 파일 경로를 확인하세요."
raw = pd.concat(frames, ignore_index=True)
print("로드 파일:", loaded)

# 통합 스키마 매핑
name_col   = find_col(raw, ["상품명","상품","제품명","상품 이름","상품명/옵션","title","name"])
unit_col   = find_col(raw, ["1회 가격","1회가격","단가(원)_사용자","단가","단가(원)=총가격/수량",
                            "1개당 가격 = 1개 가격 + 배송비","unit_price"])
rating_col = find_col(raw, ["평점","별점(5점만점)","rating","별점"])
review_col = find_col(raw, ["리뷰 수","리뷰수","리뷰수(개)","리뷰 개수","review_count","리뷰"])

# 총가격/수량으로 1회 가격 계산 컬럼이 있으면 보조 계산
if unit_col is None and ("총가격(원)=상품가+배송비" in raw.columns and "수량(추정)" in raw.columns):
    raw["__unit_calc__"] = to_num(raw["총가격(원)=상품가+배송비"]) / to_num(raw["수량(추정)"])
    unit_col = "__unit_calc__"

assert all([name_col, unit_col, rating_col, review_col]), \
    f"필수 컬럼 누락: name={name_col}, unit={unit_col}, rating={rating_col}, review={review_col}"

df = pd.DataFrame({
    "상품명": raw[name_col].astype(str),
    "1회 가격": to_num(raw[unit_col]),
    "평점": to_num(raw[rating_col]),
    "리뷰 수": to_num(raw[review_col]).fillna(0),
    "__src__": raw["__src__"]
})

# 기본 정리
df = df[(df["1회 가격"] > 0) & (df["평점"] > 0) & (df["평점"] <= 5)].copy()
df = df.drop_duplicates(subset=["상품명","1회 가격"])

# =========================
# 2) 이상치 제거
# =========================
# 하한: 4000 이하 모두 제외
low_out = df["1회 가격"] <= 4000

# 상한: IQR 상단(= Q3 + 2.5×IQR) 초과
q1, q3 = df["1회 가격"].quantile(0.25), df["1회 가격"].quantile(0.75)
iqr = q3 - q1
upper = q3 + 2.5*iqr
high_out = df["1회 가격"] > upper

outliers = df[low_out | high_out].copy()
data = df[~(low_out | high_out)].copy()

print(f"전체 {len(df):,}건 → 사용 {len(data):,}건, 이상치 {len(outliers):,}건 (상한={int(upper):,}원)")

# =========================
# 3) 가중치 설계
# =========================
# 표본을 최대한 살리기 위해 리뷰수 가중 상한을 완화(3000)
CAP = 3000.0
data["rating_norm"]   = (data["평점"] / 5.0).clip(0, 1)
data["review_weight"] = np.minimum(data["리뷰 수"], CAP) / CAP
data["가중 점수"]       = data["rating_norm"] * data["review_weight"]

# 확인용 상위 20
explain = data.sort_values("가중 점수", ascending=False)[
    ["상품명","1회 가격","평점","리뷰 수","rating_norm","review_weight","가중 점수","__src__"]
].head(20).copy()
explain["상품명"] = explain["상품명"].apply(lambda x: textwrap.shorten(str(x), width=40, placeholder="…"))
display(explain.style.format({"1회 가격":"{:,.0f}","리뷰 수":"{:,.0f}","가중 점수":"{:.4f}"}))

# =========================
# 4) 스윗스팟 탐색 (Band 1/2)
# =========================
def sweetspot_bands(df, bin_step=500, min_n=5):
    # 가격 구간을 bin_step 간격으로 나누고, 연속 2-bin(≈1,000원 폭)마다
    # 가치지수 = (가중점수합) / (가격합) 계산 → 상위 2개 겹치지 않게 선택
    pmin = math.floor(df["1회 가격"].min()/bin_step)*bin_step
    pmax = math.ceil(df["1회 가격"].max()/bin_step)*bin_step
    edges = np.arange(pmin, pmax+bin_step, bin_step)
    tmp = df.copy()
    tmp["_bin"] = pd.cut(tmp["1회 가격"], bins=edges, right=False, include_lowest=True)
    tmp["L"] = tmp["_bin"].apply(lambda x: x.left if pd.notna(x) else np.nan).astype(float)
    tmp["R"] = tmp["_bin"].apply(lambda x: x.right if pd.notna(x) else np.nan).astype(float)

    gb = tmp.groupby(["L","R"], observed=True).agg(
        sum_w=("가중 점수","sum"),
        sum_p=("1회 가격","sum"),
        n=("1회 가격","count"),
        avg_price=("1회 가격","mean"),
        avg_rating=("평점","mean")
    ).reset_index().sort_values("L").reset_index(drop=True)

    gb["center"]      = (gb["L"] + gb["R"]) / 2
    gb["value_index"] = (gb["sum_w"] / gb["sum_p"]).replace([np.inf,-np.inf], np.nan).fillna(0)

    # 2-bin 이동창
    gb["sum_w2"] = gb["sum_w"].rolling(2, min_periods=1).sum()
    gb["sum_p2"] = gb["sum_p"].rolling(2, min_periods=1).sum()
    gb["n2"]     = gb["n"].rolling(2, min_periods=1).sum()
    gb["val2"]   = (gb["sum_w2"] / gb["sum_p2"]).replace([np.inf,-np.inf], np.nan).fillna(0)
    gb["L2"]     = gb["L"].rolling(2, min_periods=1).min()
    gb["R2"]     = gb["R"].rolling(2, min_periods=1).max()

    # 표본수 조건 완화(5→4→3→2)
    cand = gb[gb["n2"] >= min_n].copy()
    if len(cand) < 2:
        for mn in [4,3,2]:
            cand = gb[gb["n2"] >= mn].copy()
            if len(cand) >= 2:
                min_n = mn
                break

    # 가치지수 높은 구간부터 비겹치게 2개 선택
    cand = cand.sort_values("val2", ascending=False)
    selected = []
    for _, r in cand.iterrows():
        L, R = float(r["L2"]), float(r["R2"])
        if not any(not (R <= s["L"] or L >= s["R"]) for s in selected):
            selected.append({"L":L, "R":R, "value":float(r["val2"]), "N":int(r["n2"])})
        if len(selected) == 2:
            break
    return selected, gb, min_n, bin_step

# 우선 500원 bin 시도, 부족 시 1000원으로 재시도
selected, gb, min_n_used, bin_step_used = sweetspot_bands(data, bin_step=500, min_n=5)
if len(selected) < 2:
    selected, gb, min_n_used, bin_step_used = sweetspot_bands(data, bin_step=1000, min_n=5)

# 결과 요약
rows, band_tables = [], []
for i, b in enumerate(selected[:2], start=1):
    inside = data[(data["1회 가격"] >= b["L"]) & (data["1회 가격"] < b["R"])].copy()
    inside = inside.sort_values("가중 점수", ascending=False)
    rows.append({
        "밴드": f"Band {i}",
        "구간(원)": f"[{int(b['L'])} ~ {int(b['R'])})",
        "가치지수(value)": round(b["value"], 6),
        "표본수": len(inside),
        "평균 1회가격": round(inside["1회 가격"].mean(), 0) if len(inside) else np.nan,
        "평균 평점": round(inside["평점"].mean(), 2) if len(inside) else np.nan,
        "bin_step(원)": bin_step_used,
        "최소 표본수 기준": min_n_used
    })
    band_tables.append((f"Band {i} 상위 10개",
                        inside[["상품명","1회 가격","평점","리뷰 수","가중 점수","__src__"]].head(10)))

band_summary = pd.DataFrame(rows)
display(band_summary.style.format({"가치지수(value)":"{:.6f}","평균 1회가격":"{:,.0f}"}))

# =========================
# 5) 시각화
# =========================
# (1) 산점도: x=1회 가격, y=가중 점수, 점 크기=리뷰 수
plt.figure(figsize=(12,6))
sizes = 25 + 75*(np.minimum(data["리뷰 수"], 3000)/3000.0)
plt.scatter(data["1회 가격"], data["가중 점수"], s=sizes, alpha=0.65)
for i, b in enumerate(selected[:2], start=1):
    plt.axvline(b["L"], color="grey", linestyle="--", alpha=0.7)
    plt.axvline(b["R"], color="grey", linestyle="--", alpha=0.7)
    ymax = float(data["가중 점수"].max()) if len(data) else 1.0
    plt.text((b["L"]+b["R"])/2, ymax*0.97, f"Band {i}", ha="center", va="top")
plt.title("도시락 전체: 1회 가격 vs 가중 점수 (≤4000원/상한IQR 제외)")
plt.xlabel("1회 가격(원)")
plt.ylabel("가중 점수 (평점×리뷰가중)")
plt.tight_layout()
plt.show()

# (2) 가격대별 가치지수 라인 그래프
plt.figure(figsize=(12,6))
plt.plot(gb["center"], gb["value_index"], marker="o", linewidth=1.8)
for i, b in enumerate(selected[:2], start=1):
    plt.axvline(b["L"], color="grey", linestyle="--", alpha=0.7)
    plt.axvline(b["R"], color="grey", linestyle="--", alpha=0.7)
    ymax2 = float(gb["value_index"].max()) if len(gb) else 1.0
    plt.text((b["L"]+b["R"])/2, ymax2*0.92, f"Band {i}", ha="center", va="top")
plt.title(f"가격대별 가치지수 ({bin_step_used}원 bin)")
plt.xlabel("가격대(중심)")
plt.ylabel("가치지수(가중점수합 / 가격합)")
plt.tight_layout()
plt.show()

# (3) 밴드별 상위 10개 테이블
for title, sub in band_tables:
    print("\n", title)
    display(sub.style.format({"1회 가격":"{:,.0f}","리뷰 수":"{:,.0f}","가중 점수":"{:.4f}"}))

# =========================
# 6) 결과 저장 (utf-8-sig)
# =========================
summary_out = "스윗스팟_분석_요약_통합_최저4000제외.csv"
explain_out = "가중치_계산_샘플_통합_최저4000제외.csv"
used_out    = "분석사용데이터_통합_최저4000제외.csv"
outlier_out = "제외된_이상치목록_통합_최저4000제외.csv"

band_summary.to_csv(summary_out, index=False, encoding="utf-8-sig")
explain.to_csv(explain_out, index=False, encoding="utf-8-sig")
data.to_csv(used_out, index=False, encoding="utf-8-sig")
outliers.to_csv(outlier_out, index=False, encoding="utf-8-sig")

print("\n저장 완료:")
for p in [summary_out, explain_out, used_out, outlier_out]:
    print(" -", os.path.abspath(p))
```

```python
# -*- coding: utf-8 -*-
# ◼︎ 도시락 스윗스팟 분석 (Band 1/2) — 그래프 PNG 저장 포함 (Jupyter에서 바로 실행)

import os, math, glob, textwrap
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import font_manager as fm, rcParams
from IPython.display import display

# ===== 0) 출력 폴더 =====
OUTDIR = "out_figs"
os.makedirs(OUTDIR, exist_ok=True)

# ===== 1) 한글 폰트 설정 (Windows 우선) =====
def set_korean_font():
    candidates_path = [
        r"C:\Windows\Fonts\malgun.ttf",   # 맑은 고딕(권장)
        r"C:\Windows\Fonts\H2GTRM.TTF",   # 한컴계열 예시
        r"C:\Windows\Fonts\H2HDRM.TTF",
    ]
    candidates_family = [
        "Malgun Gothic", "Hancom Gothic", "HY Gulim", "HY Dotum",
        "AppleGothic", "NanumGothic", "Noto Sans CJK KR", "Yu Gothic",
        "MS Gothic", "DejaVu Sans"
    ]
    family_set = None
    for p in candidates_path:
        if os.path.exists(p):
            try:
                fm.fontManager.addfont(p)
                fam = fm.FontProperties(fname=p).get_name()
                rcParams["font.family"] = [fam]
                family_set = fam
                break
            except Exception:
                pass
    if family_set is None:
        rcParams["font.family"] = candidates_family
    rcParams["axes.unicode_minus"] = False
    try:
        fm._load_fontmanager(try_read_cache=False)
    except Exception:
        pass
    print("폰트 사용:", rcParams["font.family"])
set_korean_font()

# ===== 2) 파일 로드 & 통합 =====
def read_csv_safely(path):
    try:
        return pd.read_csv(path, encoding="utf-8-sig")
    except Exception:
        return pd.read_csv(path)

def to_num(s):
    return pd.to_numeric(s, errors="coerce")

def find_col(df, candidates):
    for c in candidates:
        if c in df.columns:
            return c
    return None

file_candidates = [
    "dosirak_scored.csv",
    "dosirak_from_text.csv",
    "도시락_정기구독_상품_정리_from_txt_enriched.csv",
    "dosirak_unit_scored.csv",
]

frames, loaded = [], []
for name in file_candidates:
    for p in glob.glob(name) + glob.glob(os.path.join(".", name)) + glob.glob(os.path.join("..", name)):
        try:
            df = read_csv_safely(p)
            df["__src__"] = os.path.basename(p)
            frames.append(df)
            loaded.append((os.path.basename(p), len(df)))
        except Exception:
            pass

assert frames, "CSV를 찾지 못했습니다. 파일 경로를 확인하세요."
raw = pd.concat(frames, ignore_index=True)
print("로드 파일:", loaded)

name_col   = find_col(raw, ["상품명","상품","제품명","상품 이름","상품명/옵션","title","name"])
unit_col   = find_col(raw, ["1회 가격","1회가격","단가(원)_사용자","단가","단가(원)=총가격/수량",
                            "1개당 가격 = 1개 가격 + 배송비","unit_price"])
rating_col = find_col(raw, ["평점","별점(5점만점)","rating","별점"])
review_col = find_col(raw, ["리뷰 수","리뷰수","리뷰수(개)","리뷰 개수","review_count","리뷰"])

if unit_col is None and ("총가격(원)=상품가+배송비" in raw.columns and "수량(추정)" in raw.columns):
    raw["__unit_calc__"] = to_num(raw["총가격(원)=상품가+배송비"]) / to_num(raw["수량(추정)"])
    unit_col = "__unit_calc__"

assert all([name_col, unit_col, rating_col, review_col]), \
    f"필수 컬럼 누락: name={name_col}, unit={unit_col}, rating={rating_col}, review={review_col}"

df = pd.DataFrame({
    "상품명": raw[name_col].astype(str),
    "1회 가격": to_num(raw[unit_col]),
    "평점": to_num(raw[rating_col]),
    "리뷰 수": to_num(raw[review_col]).fillna(0),
    "__src__": raw["__src__"]
})

df = df[(df["1회 가격"] > 0) & (df["평점"] > 0) & (df["평점"] <= 5)].copy()
df = df.drop_duplicates(subset=["상품명","1회 가격"])

# ===== 3) 이상치 제거 =====
low_out = df["1회 가격"] <= 4000
q1, q3 = df["1회 가격"].quantile(0.25), df["1회 가격"].quantile(0.75)
iqr = q3 - q1
upper = q3 + 2.5*iqr
high_out = df["1회 가격"] > upper

outliers = df[low_out | high_out].copy()
data = df[~(low_out | high_out)].copy()
print(f"전체 {len(df):,}건 → 사용 {len(data):,}건, 이상치 {len(outliers):,}건 (상한={int(upper):,}원)")

# ===== 4) 가중치 =====
CAP = 3000.0
data["rating_norm"]   = (data["평점"] / 5.0).clip(0, 1)
data["review_weight"] = np.minimum(data["리뷰 수"], CAP) / CAP
data["가중 점수"]       = data["rating_norm"] * data["review_weight"]

explain = data.sort_values("가중 점수", ascending=False)[
    ["상품명","1회 가격","평점","리뷰 수","rating_norm","review_weight","가중 점수","__src__"]
].head(20).copy()
explain["상품명"] = explain["상품명"].apply(lambda x: textwrap.shorten(str(x), width=40, placeholder="…"))
display(explain.style.format({"1회 가격":"{:,.0f}","리뷰 수":"{:,.0f}","가중 점수":"{:.4f}"}))

# ===== 5) 스윗스팟 탐색 (Band 1/2) =====
def sweetspot_bands(df, bin_step=500, min_n=5):
    pmin = math.floor(df["1회 가격"].min()/bin_step)*bin_step
    pmax = math.ceil(df["1회 가격"].max()/bin_step)*bin_step
    edges = np.arange(pmin, pmax+bin_step, bin_step)

    tmp = df.copy()
    tmp["_bin"] = pd.cut(tmp["1회 가격"], bins=edges, right=False, include_lowest=True)
    tmp["L"] = tmp["_bin"].apply(lambda x: x.left if pd.notna(x) else np.nan).astype(float)
    tmp["R"] = tmp["_bin"].apply(lambda x: x.right if pd.notna(x) else np.nan).astype(float)

    gb = tmp.groupby(["L","R"], observed=True).agg(
        sum_w=("가중 점수","sum"),
        sum_p=("1회 가격","sum"),
        n=("1회 가격","count"),
        avg_price=("1회 가격","mean"),
        avg_rating=("평점","mean")
    ).reset_index().sort_values("L").reset_index(drop=True)

    gb["center"]      = (gb["L"] + gb["R"]) / 2
    gb["value_index"] = (gb["sum_w"] / gb["sum_p"]).replace([np.inf,-np.inf], np.nan).fillna(0)

    gb["sum_w2"] = gb["sum_w"].rolling(2, min_periods=1).sum()
    gb["sum_p2"] = gb["sum_p"].rolling(2, min_periods=1).sum()
    gb["n2"]     = gb["n"].rolling(2, min_periods=1).sum()
    gb["val2"]   = (gb["sum_w2"] / gb["sum_p2"]).replace([np.inf,-np.inf], np.nan).fillna(0)
    gb["L2"]     = gb["L"].rolling(2, min_periods=1).min()
    gb["R2"]     = gb["R"].rolling(2, min_periods=1).max()

    cand = gb[gb["n2"] >= min_n].copy()
    if len(cand) < 2:
        for mn in [4,3,2]:
            cand = gb[gb["n2"] >= mn].copy()
            if len(cand) >= 2:
                min_n = mn
                break

    cand = cand.sort_values("val2", ascending=False)
    selected = []
    for _, r in cand.iterrows():
        L, R = float(r["L2"]), float(r["R2"])
        if not any(not (R <= s["L"] or L >= s["R"]) for s in selected):
            selected.append({"L":L, "R":R, "value":float(r["val2"]), "N":int(r["n2"])})
        if len(selected) == 2:
            break
    return selected, gb, min_n, bin_step

selected, gb, min_n_used, bin_step_used = sweetspot_bands(data, bin_step=500, min_n=5)
if len(selected) < 2:
    selected, gb, min_n_used, bin_step_used = sweetspot_bands(data, bin_step=1000, min_n=5)

rows, band_tables = [], []
for i, b in enumerate(selected[:2], start=1):
    inside = data[(data["1회 가격"] >= b["L"]) & (data["1회 가격"] < b["R"])].copy()
    inside = inside.sort_values("가중 점수", ascending=False)
    rows.append({
        "밴드": f"Band {i}",
        "구간(원)": f"[{int(b['L'])} ~ {int(b['R'])})",
        "가치지수(value)": round(b["value"], 6),
        "표본수": len(inside),
        "평균 1회가격": round(inside["1회 가격"].mean(), 0) if len(inside) else np.nan,
        "평균 평점": round(inside["평점"].mean(), 2) if len(inside) else np.nan,
        "bin_step(원)": bin_step_used,
        "최소 표본수 기준": min_n_used
    })
    band_tables.append((f"Band {i} 상위 10개",
                        inside[["상품명","1회 가격","평점","리뷰 수","가중 점수","__src__"]].head(10)))

band_summary = pd.DataFrame(rows)
display(band_summary.style.format({"가치지수(value)":"{:.6f}","평균 1회가격":"{:,.0f}"}))

# ===== 6) 시각화 + PNG 저장 =====
# (1) 산점도
fig1 = plt.figure(figsize=(12,6))
sizes = 25 + 75*(np.minimum(data["리뷰 수"], 3000)/3000.0)
plt.scatter(data["1회 가격"], data["가중 점수"], s=sizes, alpha=0.65)
for i, b in enumerate(selected[:2], start=1):
    plt.axvline(b["L"], color="grey", linestyle="--", alpha=0.7)
    plt.axvline(b["R"], color="grey", linestyle="--", alpha=0.7)
    ymax = float(data["가중 점수"].max()) if len(data) else 1.0
    plt.text((b["L"]+b["R"])/2, ymax*0.97, f"Band {i}", ha="center", va="top")
plt.title("도시락 전체: 1회 가격 vs 가중 점수 (≤4000원/상한IQR 제외)")
plt.xlabel("1회 가격(원)")
plt.ylabel("가중 점수 (평점×리뷰가중)")
plt.tight_layout()
plt.show()
fig1.savefig(os.path.join(OUTDIR, "스윗스팟_산점도.png"), dpi=300, bbox_inches="tight")

# (2) 가격대별 가치지수 라인
fig2 = plt.figure(figsize=(12,6))
plt.plot(gb["center"], gb["value_index"], marker="o", linewidth=1.8)
for i, b in enumerate(selected[:2], start=1):
    plt.axvline(b["L"], color="grey", linestyle="--", alpha=0.7)
    plt.axvline(b["R"], color="grey", linestyle="--", alpha=0.7)
    ymax2 = float(gb["value_index"].max()) if len(gb) else 1.0
    plt.text((b["L"]+b["R"])/2, ymax2*0.92, f"Band {i}", ha="center", va="top")
plt.title(f"가격대별 가치지수 ({bin_step_used}원 bin)")
plt.xlabel("가격대(중심)")
plt.ylabel("가치지수(가중점수합 / 가격합)")
plt.tight_layout()
plt.show()
fig2.savefig(os.path.join(OUTDIR, "스윗스팟_가치지수_라인.png"), dpi=300, bbox_inches="tight")

# (3) 선택: 밴드별 상위 10개 막대 차트(가중 점수 기준) PNG 저장
for i, (_, sub) in enumerate(band_tables, start=1):
    sub_plot = sub.sort_values("가중 점수", ascending=True).copy()
    fig = plt.figure(figsize=(10,6))
    plt.barh(sub_plot["상품명"], sub_plot["가중 점수"])
    plt.title(f"Band {i} 상위 10개 (가중 점수)")
    plt.xlabel("가중 점수")
    plt.tight_layout()
    plt.show()
    fig.savefig(os.path.join(OUTDIR, f"Band{i}_상위10_가중점수.png"), dpi=300, bbox_inches="tight")

# ===== 7) 결과 저장 (utf-8-sig) =====
summary_out = os.path.join(OUTDIR, "스윗스팟_분석_요약_통합_최저4000제외.csv")
explain_out = os.path.join(OUTDIR, "가중치_계산_샘플_통합_최저4000제외.csv")
used_out    = os.path.join(OUTDIR, "분석사용데이터_통합_최저4000제외.csv")
outlier_out = os.path.join(OUTDIR, "제외된_이상치목록_통합_최저4000제외.csv")

band_summary.to_csv(summary_out, index=False, encoding="utf-8-sig")
explain.to_csv(explain_out, index=False, encoding="utf-8-sig")
data.to_csv(used_out, index=False, encoding="utf-8-sig")
outliers.to_csv(outlier_out, index=False, encoding="utf-8-sig")

print("\n저장 완료 (CSV & PNG):")
for p in [summary_out, explain_out, used_out, outlier_out,
          os.path.join(OUTDIR, "스윗스팟_산점도.png"),
          os.path.join(OUTDIR, "스윗스팟_가치지수_라인.png"),
          os.path.join(OUTDIR, "Band1_상위10_가중점수.png"),
          os.path.join(OUTDIR, "Band2_상위10_가중점수.png")]:
    if os.path.exists(p):
        print(" -", os.path.abspath(p))
```
