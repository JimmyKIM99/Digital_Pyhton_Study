```python
# -*- coding: utf-8 -*-
"""
Kurly: 도시락·밥류(카테고리 912003) — 예시 버전
- 상품 10개만 크롤링
- 상품당 최대 50개 리뷰
- 결과 CSV: category_name, product_name, brand, product_url, price, review_count, review_date, review_text
"""

import re, time, json
from datetime import datetime
from typing import List, Dict, Any, Optional, Tuple

import pandas as pd
from bs4 import BeautifulSoup

import undetected_chromedriver as uc
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

# ========= 설정 =========
CATEGORY_ID = "912003"                  # 도시락·밥류
CATEGORY_NAME = "도시락·밥류"            # 소카테고리명 고정
HEADLESS = False
WINDOW_SIZE = "1920,1080"
SHORT, MID, LONG = 3, 7, 12

MAX_PRODUCTS = 10                       # ✅ 예시: 10개만
PAGES_SCAN_LIMIT = 5                    # 앞 페이지부터 훑다가 10개 모이면 중단
MAX_REVIEWS_PER_PRODUCT = 50            # 상품당 리뷰 최대
MAX_LOAD_MORE_CLICKS = 40

DIGIT_ONLY = re.compile(r"^[0-9,]+$")
DATE_PAT = re.compile(r"\d{4}\.\d{1,2}\.\d{1,2}")

# 스샷 기반 '더보기' 버튼 우선 CSS + 보강
MORE_BTN_CSS = [
    "button.css-1orpr5k.e5b97xx9",
    "button.css-l0prsx.e5b97xx9",
    "button[class*='css-'][class*='e5b97']",
]
MORE_BTN_XPATH = ["//button[contains(.,'더보기')]", "//a[contains(.,'더보기')]"]
REVIEW_TAB_XPATHS = [
    "//nav//li[.//span[contains(.,'후기')]]",
    "//nav//*[self::a or self::button][contains(.,'후기')]",
    "//button[contains(.,'후기')]", "//a[contains(.,'후기')]",
]

# ========= 유틸 =========
def make_driver():
    opts = uc.ChromeOptions()
    if HEADLESS: opts.add_argument("--headless=new")
    opts.add_argument("--disable-dev-shm-usage"); opts.add_argument("--no-sandbox")
    opts.add_argument("--lang=ko_KR"); opts.add_argument(f"--window-size={WINDOW_SIZE}")
    opts.add_argument("--start-maximized"); opts.add_argument("--disable-blink-features=AutomationControlled")
    service = Service(ChromeDriverManager().install())
    d = uc.Chrome(service=service, options=opts); d.set_page_load_timeout(60); return d

def wait_css(d, css, timeout=MID):  return WebDriverWait(d, timeout).until(EC.presence_of_element_located((By.CSS_SELECTOR, css)))
def wait_xpath(d, xp, timeout=MID): return WebDriverWait(d, timeout).until(EC.presence_of_element_located((By.XPATH, xp)))
def safe_click(d, by, val, timeout=MID):
    try:
        el = WebDriverWait(d, timeout).until(EC.element_to_be_clickable((by, val)))
        d.execute_script("arguments[0].scrollIntoView({block:'center'});", el); time.sleep(0.2); el.click(); return True
    except: return False
def slow_scroll(d, loops=20, step=1400, sl=0.6):
    last = 0
    for _ in range(loops):
        d.execute_script(f"window.scrollBy(0,{step});"); time.sleep(sl)
        h = d.execute_script("return document.body.scrollHeight;")
        if h == last: break
        last = h

def list_page_url(page): return f"https://www.kurly.com/categories/{CATEGORY_ID}?filters=&page={page}"

def collect_links_up_to(d, max_n=MAX_PRODUCTS) -> List[str]:
    links, seen = [], set()
    for p in range(1, PAGES_SCAN_LIMIT + 1):
        if len(links) >= max_n: break
        d.get(list_page_url(p)); wait_css(d, "div#container"); time.sleep(SHORT); slow_scroll(d, loops=25)
        anchors = d.find_elements(By.CSS_SELECTOR, "a[href*='/goods/']")
        for a in anchors:
            href = a.get_attribute("href") or ""
            if "/goods/" in href and href not in seen:
                seen.add(href); links.append(href)
                if len(links) >= max_n: break
    return links[:max_n]

def parse_json_ld(html: str) -> Dict[str, Any]:
    soup = BeautifulSoup(html, "lxml"); data={}
    for tag in soup.select('script[type="application/ld+json"]'):
        try: j=json.loads(tag.string)
        except: continue
        for obj in (j if isinstance(j,list) else [j]):
            if isinstance(obj,dict) and obj.get("@type")=="Product":
                off=obj.get("offers"); price=None
                if isinstance(off,dict): price=off.get("price")
                elif isinstance(off,list):
                    for o in off:
                        if isinstance(o,dict) and o.get("price"): price=o["price"]; break
                if price:
                    try: data["price"]=int(str(price).replace(",",""))
                    except: pass
                ar=obj.get("aggregateRating")
                if isinstance(ar,dict) and ar.get("reviewCount") is not None:
                    try: data["review_count"]=int(ar["reviewCount"])
                    except: pass
                return data
    return data

def extract_price_dom(d)->Optional[int]:
    try:
        h1=wait_xpath(d,"//h1",timeout=LONG)
        box=h1.find_element(By.XPATH,"./ancestor::section[1] | ./ancestor::div[2]")
        spans=box.find_elements(By.XPATH,".//span[normalize-space()]")
        cand=[]
        for s in spans:
            t=s.text.strip()
            if t and "%" not in t and DIGIT_ONLY.match(t): cand.append(int(t.replace(",","")))
        if cand: return cand[0]
    except: pass
    return None

def extract_reviewcount_dom(d)->Optional[int]:
    try:
        node=d.find_element(By.XPATH,"//nav//*[contains(.,'후기')]/ancestor::li")
        m=re.search(r"후기\s*\(?\s*(\d+)", (node.text or "").replace(",",""))
        if m: return int(m.group(1))
    except: pass
    return None

def open_review_tab(d):
    for xp in REVIEW_TAB_XPATHS:
        if safe_click(d, By.XPATH, xp, timeout=SHORT): time.sleep(1.0); return

def click_more_once(d)->bool:
    for css in MORE_BTN_CSS:
        try:
            for b in d.find_elements(By.CSS_SELECTOR, css):
                if b.is_displayed() and b.is_enabled():
                    d.execute_script("arguments[0].scrollIntoView({block:'center'});", b)
                    time.sleep(0.2); b.click(); time.sleep(1.0); return True
        except: pass
    for xp in MORE_BTN_XPATH:
        if safe_click(d, By.XPATH, xp, timeout=SHORT): time.sleep(1.0); return True
    return False

def expand_reviews(d, max_reviews=MAX_REVIEWS_PER_PRODUCT, max_clicks=MAX_LOAD_MORE_CLICKS):
    clicks=0
    while clicks<max_clicks:
        if not click_more_once(d): break
        clicks+=1
        # 본문 더보기도 한 번씩 눌러주기
        for _ in range(3):
            if not safe_click(d, By.XPATH, "//article//button[contains(.,'더보기')]", timeout=SHORT): break
        slow_scroll(d, loops=2)

        soup=BeautifulSoup(d.page_source,"lxml")
        items=soup.select("section[id*='review'] article") or soup.select("article")
        if len(items)>=max_reviews: break

def parse_reviews(d, product_url:str)->List[Dict[str,Any]]:
    soup=BeautifulSoup(d.page_source,"lxml")
    root=soup.select_one("section[id*='review'], section[class*='review'], div[id*='review'], div[class*='review']") or soup
    cards=root.select("article") or root.select("li")
    rows=[]
    for tag in cards:
        ps=[p.get_text(" ",strip=True) for p in tag.select("p") if p.get_text(strip=True)]
        body=max(ps, key=len) if ps else tag.get_text(" ",strip=True)
        if not body or len(body)<10: continue
        m=DATE_PAT.search(tag.get_text(" ",strip=True))
        date_str=m.group(0) if m else None
        rows.append({"product_url":product_url,"review_date":date_str,"review_text":body})
        if len(rows)>=MAX_REVIEWS_PER_PRODUCT: break
    return rows

def split_brand_from_title(title:str)->Tuple[Optional[str],str]:
    if not title: return None,""
    m=re.match(r"\s*\[([^\]]+)\]\s*(.*)", title)
    if m:
        return m.group(1).strip(), (m.group(2).strip() or title.strip())
    return None, title.strip()

def crawl_product(d, url:str)->List[Dict[str,Any]]:
    d.get(url)
    try: title=wait_xpath(d,"//h1",timeout=LONG).text.strip()
    except:
        try: title=wait_css(d,"h1, h2",timeout=MID).text.strip()
        except: title=""
    brand, product_name = split_brand_from_title(title)

    j=parse_json_ld(d.page_source)
    price=j.get("price") or extract_price_dom(d)
    review_count=j.get("review_count") or extract_reviewcount_dom(d)

    open_review_tab(d); expand_reviews(d)
    reviews=parse_reviews(d, url)
    if not reviews: reviews=[{"product_url":url,"review_date":None,"review_text":None}]

    return [{
        "category_name": CATEGORY_NAME,
        "product_name": product_name,
        "brand": brand,
        "product_url": url,
        "price": price,
        "review_count": review_count,
        "review_date": r["review_date"],
        "review_text": r["review_text"],
    } for r in reviews]

def main():
    d=make_driver()
    rows=[]
    try:
        links=collect_links_up_to(d, MAX_PRODUCTS)
        print(f"[INFO] 대상 상품 {len(links)}개")
        for i, lk in enumerate(links, 1):
            print(f"[{i}/{len(links)}] {lk}")
            try:
                rows.extend(crawl_product(d, lk))
            except Exception as e:
                print("  └ 에러:", e)
    finally:
        d.quit()

    out=f"kurly_meal_rice_example10_{datetime.now().strftime('%Y%m%d_%H%M')}.csv"
    cols=["category_name","product_name","brand","product_url","price","review_count","review_date","review_text"]
    pd.DataFrame(rows, columns=cols).to_csv(out, index=False, encoding="utf-8-sig")
    print(f"[DONE] 저장: {out} (행 {len(rows)})")

if __name__=="__main__":
    main()
```

```python
# ===== 추가 import =====
from selenium.common.exceptions import TimeoutException, StaleElementReferenceException, JavascriptException

# ===== 리뷰 섹션 / 프레임 헬퍼 =====
def _switch_into_review_frame_if_any(d):
    """리뷰 섹션 내부에 iframe이 있으면 진입. 진입했는지 여부 반환."""
    try:
        frs = d.find_elements(By.XPATH, "//section[@aria-label='상품 후기']//iframe | //div[@id='review']//iframe")
        if frs:
            d.switch_to.frame(frs[0])
            return True
    except Exception:
        pass
    return False

def _back_to_default_content(d):
    try:
        d.switch_to.default_content()
    except Exception:
        pass

def _find_next_button(d, timeout=5):
    """
    리뷰 섹션의 '다음 >' 버튼을 정확히 하나 찾는다.
    1) '다음' 텍스트 기반
    2) 버튼 컨테이너의 2번째 버튼(우측 화살표)
    3) '>' 텍스트 기반(아이콘 렌더링 케이스)
    """
    xpaths = [
        # 텍스트 '다음' 또는 '다음 >'
        "(//section[@aria-label='상품 후기'] | //div[@id='review'] | //div[contains(@class,'review')])[1]"
        "//button[not(@disabled)][.//div[normalize-space()='다음' or normalize-space()='다음 >']]",

        # 컨테이너 내 2번째 버튼(Prev, Next 구조) → Next
        "(//section[@aria-label='상품 후기'] | //div[@id='review'] | //div[contains(@class,'review')])[1]"
        "//div[./button][./button[2]]/button[2][not(@disabled)]",

        # 단순히 '>' 포함(아이콘과 함께 랜더링될 때)
        "(//section[@aria-label='상품 후기'] | //div[@id='review'] | //div[contains(@class,'review')])[1]"
        "//button[not(@disabled)][.//div[contains(text(), '>')] or contains(normalize-space(.), '>')]",
    ]
    last_err = None
    for xp in xpaths:
        try:
            btn = WebDriverWait(d, timeout).until(EC.element_to_be_clickable((By.XPATH, xp)))
            if btn:
                return btn
        except Exception as e:
            last_err = e
            continue
    return None

def _scroll_small_until_button_or_limit(d, max_tries=8, step=650, sleep=0.35):
    """
    버튼이 안 보이면 작은 스텝으로만 아래로 스크롤하며 재탐색.
    버튼을 찾는 즉시 스크롤 중단.
    """
    for _ in range(max_tries):
        btn = _find_next_button(d, timeout=1)
        if btn:
            return btn
        d.execute_script(f"window.scrollBy(0, {step});")
        time.sleep(sleep)
    return None

def _click_element_stably(d, el):
    try:
        d.execute_script("arguments[0].scrollIntoView({block:'center'});", el)
        time.sleep(0.2)
        el.click()
        return True
    except Exception:
        try:
            d.execute_script("arguments[0].click();", el)
            return True
        except JavascriptException:
            return False

def open_review_tab(d):
    """기존 함수 유지 가능. 여기선 리뷰 섹션까지 스크롤만 보강."""
    for xp in ["//nav//li[.//span[contains(.,'후기')]]",
               "//nav//*[self::a or self::button][contains(.,'후기')]",
               "//button[contains(.,'후기')]", "//a[contains(.,'후기')]"]:
        if safe_click(d, By.XPATH, xp, timeout=SHORT):
            time.sleep(0.8)
            break
    # 리뷰 섹션으로 스크롤
    try:
        target = WebDriverWait(d, MID).until(
            EC.presence_of_element_located((By.XPATH, "//section[@aria-label='상품 후기'] | //div[@id='review']"))
        )
        d.execute_script("arguments[0].scrollIntoView({block:'start'});", target)
        time.sleep(0.4)
    except TimeoutException:
        pass

def parse_reviews_on_current_view(d, product_url: str, max_take=None, already=None):
    """
    현재 화면에 보이는 리뷰 카드들만 파싱.
    already: (date,text) 튜플 set 로 중복 방지
    """
    soup = BeautifulSoup(d.page_source, "lxml")
    root = soup.select_one("section[aria-label='상품 후기']") or soup.select_one("#review") or soup
    cards = root.select("article")
    if not cards:
        cards = root.select("li")

    out, seen = [], (already or set())
    for tag in cards:
        # 본문
        ps = [p.get_text(" ", strip=True) for p in tag.select("p") if p.get_text(strip=True)]
        body = max(ps, key=len) if ps else tag.get_text(" ", strip=True)
        if not body or len(body) < 6:
            continue
        # 날짜
        m = DATE_PAT.search(tag.get_text(" ", strip=True))
        date_str = m.group(0) if m else None

        key = (date_str, body)
        if key in seen:
            continue
        seen.add(key)
        out.append({"product_url": product_url, "review_date": date_str, "review_text": body})
        if max_take and len(out) >= max_take:
            break
    return out, seen

def paginate_reviews(d, product_url: str, want_count=50, max_next_clicks=200):
    """
    리뷰 더보기(다음) 버튼만 눌러서 리뷰를 모은다.
    - 버튼을 찾으면 스크롤을 멈추고 눌러 진행
    - 없을 때에만 소량 스크롤로 버튼 탐색
    """
    collected, seen = [], set()

    # (옵션) iframe 진입
    entered_frame = _switch_into_review_frame_if_any(d)

    # 첫 화면 리뷰 수집
    chunk, seen = parse_reviews_on_current_view(d, product_url, max_take=None, already=seen)
    collected.extend(chunk)
    if len(collected) >= want_count:
        if entered_frame: _back_to_default_content(d)
        return collected[:want_count]

    clicks = 0
    while clicks < max_next_clicks and len(collected) < want_count:
        # 버튼 시도: 먼저 바로 찾기 → 없으면 소량 스크롤 탐색
        btn = _find_next_button(d, timeout=2)
        if not btn:
            btn = _scroll_small_until_button_or_limit(d, max_tries=6)
        if not btn:
            # 더 이상 페이지가 없다고 판단
            break

        # 버튼 클릭
        ok = _click_element_stably(d, btn)
        if not ok:
            break
        clicks += 1
        time.sleep(1.0)

        # 새 리뷰 수집
        chunk, seen = parse_reviews_on_current_view(d, product_url, max_take=None, already=seen)
        if chunk:
            collected.extend(chunk)
        else:
            # 새 내용이 안 늘어나면 종료
            break

        if len(collected) >= want_count:
            break

    if entered_frame:
        _back_to_default_content(d)

    return collected[:want_count]
```

```python
# -*- coding: utf-8 -*-
"""
Kurly 도시락·밥류(912003) → 각 상품 상세 → 리뷰 최대 50개
리뷰 '다음(>)' 버튼을 정확히 1개만 잡아 클릭 (iframe 대응, 무한 스크롤 방지)
저장 컬럼: category_name, product_name, brand, product_url, price, review_count, review_date, review_text
"""

import re, json, time
from datetime import datetime
from typing import List, Dict, Any, Optional, Tuple

import pandas as pd
from bs4 import BeautifulSoup

import undetected_chromedriver as uc
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.common.exceptions import TimeoutException

# =========================== 설정 ===========================
CATEGORY_ID = "912003"               # 도시락·밥류
CATEGORY_NAME = "도시락·밥류"
PAGES_TO_CRAWL = 1                   # 리스트 페이지 수 (예시: 1페이지만)
MAX_PRODUCTS = None             # None = 전부, 정수면 상한
MAX_REVIEWS_PER_PRODUCT = 50
HEADLESS = False
WIN_SIZE = "1920,1080"
SHORT, MID, LONG = 3, 7, 12
DIGIT_ONLY = re.compile(r"^[0-9,]+$")
DATE_PAT = re.compile(r"\d{4}\.\d{1,2}\.\d{1,2}")

# =========================== 드라이버 =========================
def make_driver() -> uc.Chrome:
    opts = uc.ChromeOptions()
    if HEADLESS:
        opts.add_argument("--headless=new")
    opts.add_argument("--disable-dev-shm-usage")
    opts.add_argument("--no-sandbox")
    opts.add_argument("--lang=ko_KR")
    opts.add_argument(f"--window-size={WIN_SIZE}")
    opts.add_argument("--start-maximized")
    opts.add_argument("--disable-blink-features=AutomationControlled")
    service = Service(ChromeDriverManager().install())
    d = uc.Chrome(service=service, options=opts)
    d.set_page_load_timeout(60)
    return d

def wait_css(d, css, timeout=MID):
    return WebDriverWait(d, timeout).until(EC.presence_of_element_located((By.CSS_SELECTOR, css)))

def wait_xpath(d, xp, timeout=MID):
    return WebDriverWait(d, timeout).until(EC.presence_of_element_located((By.XPATH, xp)))

def js_click(d, el):
    d.execute_script("arguments[0].scrollIntoView({block:'center'});", el)
    time.sleep(0.15)
    d.execute_script("arguments[0].click();", el)

# =========================== 리스트 → 링크 ====================
def list_page_url(page:int)->str:
    return f"https://www.kurly.com/categories/{CATEGORY_ID}?filters=&page={page}"

def collect_links_from_list_page(d, page:int)->List[str]:
    d.get(list_page_url(page))
    wait_css(d, "div#container")
    # 카드가 lazy-load → 적당히만 스크롤
    for _ in range(8):
        d.execute_script("window.scrollBy(0, 1400);")
        time.sleep(0.5)
    anchors = d.find_elements(By.CSS_SELECTOR, "a[href*='/goods/']")
    seen, links = set(), []
    for a in anchors:
        href = a.get_attribute("href") or ""
        if "/goods/" in href and href not in seen:
            seen.add(href); links.append(href)
    return links

# =========================== 가격/리뷰수 ======================
def parse_json_ld(html:str)->Dict[str,Any]:
    soup = BeautifulSoup(html, "lxml")
    out = {}
    for tag in soup.select('script[type="application/ld+json"]'):
        try:
            j = json.loads(tag.string)
        except Exception:
            continue
        objs = j if isinstance(j, list) else [j]
        for o in objs:
            if isinstance(o, dict) and o.get("@type")=="Product":
                # price
                off = o.get("offers")
                price = None
                if isinstance(off, dict):
                    price = off.get("price")
                elif isinstance(off, list):
                    for oo in off:
                        if isinstance(oo, dict) and oo.get("price"):
                            price = oo["price"]; break
                if price:
                    try: out["price"] = int(str(price).replace(",", ""))
                    except: pass
                # reviewCount
                ag = o.get("aggregateRating")
                if isinstance(ag, dict) and ag.get("reviewCount") is not None:
                    try: out["review_count"] = int(str(ag["reviewCount"]).replace(",", ""))
                    except: pass
                return out
    return out

def extract_price_dom(d)->Optional[int]:
    try:
        h1 = wait_xpath(d, "//h1", timeout=LONG)
        box = h1.find_element(By.XPATH, "./ancestor::section[1] | ./ancestor::div[2]")
        spans = box.find_elements(By.XPATH, ".//span[normalize-space()]")
        cands = []
        for s in spans:
            t = s.text.strip()
            if t and "%" not in t and DIGIT_ONLY.match(t):
                cands.append(int(t.replace(",", "")))
        if cands: return cands[0]
    except Exception:
        pass
    return None

def extract_reviewcount_dom(d)->Optional[int]:
    try:
        node = d.find_element(By.XPATH, "//nav//*[contains(.,'후기')]/ancestor::li")
        txt = node.text.replace(",", "")
        m = re.search(r"후기\s*\(?\s*(\d+)", txt)
        if m: return int(m.group(1))
    except Exception:
        pass
    return None

# =========================== 제목 → 브랜드/상품명 ===============
def split_brand_from_title(title:str)->Tuple[Optional[str], str]:
    if not title: return None, ""
    m = re.match(r"\s*\[([^\]]+)\]\s*(.*)", title)
    if m:
        brand = m.group(1).strip()
        name = m.group(2).strip() or title.strip()
        return brand, name
    return None, title.strip()

# =========================== 리뷰 섹션 조작 =====================
def get_review_section(d):
    """'상품 후기' 섹션 루트 반환(없으면 None). 이 섹션이 보이면 글로벌 스크롤을 멈춘다."""
    xps = [
        "//section[@aria-label='상품 후기' or contains(@aria-label,'상품 후기')]",
        "//*[@id='review' and (self::section or self::div)]",
        "//section[contains(@class,'review')]",
        "//div[@id='review']"
    ]
    for xp in xps:
        try:
            el = d.find_element(By.XPATH, xp)
            if el.is_displayed():
                d.execute_script("arguments[0].scrollIntoView({block:'center'});", el)
                time.sleep(0.4)
                return el
        except Exception:
            continue
    return None

def switch_into_review_iframe(d)->bool:
    """리뷰 근처 iframe이 있으면 들어가고 True 반환. 없으면 False."""
    try:
        frames = d.find_elements(By.CSS_SELECTOR, "#review iframe, section[aria-label] iframe, section[id*='review'] iframe, section[class*='review'] iframe")
        for fr in frames:
            if fr.is_displayed():
                d.switch_to.frame(fr)
                return True
    except Exception:
        pass
    return False

def leave_iframe(d):
    try:
        d.switch_to.default_content()
    except Exception:
        pass

def _find_next_button_in(node):
    """리뷰 섹션 안에서 '다음' 버튼 후보를 순차 시도(Union 제거)."""
    candidates = []
    # 1) '다음' 고정 텍스트(div)
    for xp in [
        ".//button[not(@disabled)][.//div[normalize-space()='다음' or normalize-space()='다음 >']]",
        ".//button[not(@disabled)][.//span[normalize-space()='다음' or normalize-space()='다음 >']]",
        # 2) Prev/Next 두 개 버튼 컨테이너의 2번째 버튼
        ".//div[./button][./button[2]]/button[2][not(@disabled)]",
        # 3) '>' 포함(아이콘 + 텍스트 케이스)
        ".//button[not(@disabled)][contains(normalize-space(.), '>')]",
    ]:
        try:
            els = node.find_elements(By.XPATH, xp)
            vis = [e for e in els if e.is_displayed()]
            if vis:
                candidates.extend(vis)
        except Exception:
            continue
    # '이전' 제외, 여러 개면 우측(마지막) 우선
    filtered = []
    for b in candidates:
        try:
            t = (b.text or "").strip()
            if "이전" not in t:
                filtered.append(b)
        except Exception:
            continue
    if not filtered:
        return None
    if len(filtered) == 1:
        return filtered[0]
    # '다음' 포함 우선, 아니면 마지막
    pri = [b for b in filtered if "다음" in (b.text or "")]
    return pri[0] if len(pri) == 1 else filtered[-1]

def click_review_next(d)->bool:
    """리뷰 섹션의 '다음' 버튼을 정확히 1개 찾아 클릭. iframe 안/밖 모두 시도."""
    # 1) 기본 컨텍스트
    sec = get_review_section(d)
    if sec:
        btn = _find_next_button_in(sec)
        if btn:
            js_click(d, btn)
            return True

    # 2) iframe 내부
    entered = switch_into_review_iframe(d)
    try:
        if entered:
            # 프레임 안에서는 전역에서 버튼 후보를 순차 검색(Union 제거)
            for xp in [
                "//button[not(@disabled)][.//*[normalize-space()='다음' or normalize-space()='다음 >']]", 
                "//div[./button][./button[2]]/button[2][not(@disabled)]",
                "//button[not(@disabled)][contains(normalize-space(.), '>') and not(contains(.,'이전'))]"
            ]:
                try:
                    els = d.find_elements(By.XPATH, xp)
                    vis = [e for e in els if e.is_displayed()]
                    if vis:
                        js_click(d, vis[-1] if len(vis)>1 else vis[0])
                        return True
                except Exception:
                    continue
    finally:
        if entered:
            leave_iframe(d)

    # 3) 최후: 기본 컨텍스트에서 텍스트 기반
    try:
        btn = d.find_element(By.XPATH, "//button[not(@disabled)][.//*[contains(.,'다음') or contains(.,'>')]][last()]")
        js_click(d, btn)
        return True
    except Exception:
        return False

def find_first_review_card(d):
    xps = [
        "//section[@aria-label='상품 후기']//article[1]",
        "//*[@id='review']//article[1]",
        "//article[1]"
    ]
    for xp in xps:
        try:
            el = d.find_element(By.XPATH, xp)
            if el.is_displayed():
                return el
        except Exception:
            continue
    return None

def parse_reviews_now(d, product_url:str)->List[Dict[str,Any]]:
    soup = BeautifulSoup(d.page_source, "lxml")
    root = soup.select_one("section[aria-label='상품 후기']") or \
           soup.select_one("#review") or \
           soup.select_one("section[class*='review']") or \
           soup
    cards = root.select("article")
    if not cards: cards = root.select("li")
    rows=[]
    for tag in cards:
        ps = [p.get_text(" ", strip=True) for p in tag.select("p") if p.get_text(strip=True)]
        body = max(ps, key=len) if ps else tag.get_text(" ", strip=True)
        if not body or len(body)<10: continue
        m = DATE_PAT.search(tag.get_text(" ", strip=True))
        date_str = m.group(0) if m else None
        rows.append({"product_url":product_url, "review_date":date_str, "review_text":body})
        if len(rows) >= MAX_REVIEWS_PER_PRODUCT:
            break
    return rows

def paginate_and_collect_reviews(d, product_url:str, max_reviews:int=MAX_REVIEWS_PER_PRODUCT)->List[Dict[str,Any]]:
    """'후기' 탭 열고, 다음 버튼만 사용해 최대 max_reviews까지 수집."""
    # 후기 탭 열기
    for xp in [
        "//nav//li[.//span[contains(.,'후기')]]",
        "//nav//*[self::a or self::button][contains(.,'후기')]",
        "//button[contains(.,'후기')]",
        "//a[contains(.,'후기')]",
    ]:
        try:
            el = d.find_element(By.XPATH, xp)
            js_click(d, el); time.sleep(0.6); break
        except Exception:
            continue

    # 첫 페이지 수집
    rows = parse_reviews_now(d, product_url)
    seen = set((r["review_date"], r["review_text"]) for r in rows)

    # 다음 버튼 클릭 반복 (무한 스크롤 없이 버튼만)
    tries, pages = 0, 0
    while len(rows) < max_reviews and tries < 120:
        base = find_first_review_card(d)  # 페이지 전/후 비교용

        # 버튼이 안 보이면 소량 스크롤 탐색(찾자마자 스크롤 중단)
        btn = None
        sec = get_review_section(d)
        if sec:
            btn = _find_next_button_in(sec)
        if not btn:
            for _ in range(6):
                btn = _find_next_button_in(get_review_section(d) or sec) if get_review_section(d) else None
                if btn: break
                d.execute_script("window.scrollBy(0, 650);")
                time.sleep(0.3)

        if not btn and not click_review_next(d):  # 최후 전역/iframe 시도
            break

        if btn:
            js_click(d, btn)

        # 페이지 변경 대기
        changed = False
        try:
            if base:
                WebDriverWait(d, LONG).until(EC.staleness_of(base))
                changed = True
            else:
                time.sleep(1.0)
                changed = True
        except TimeoutException:
            pass

        # 새 리뷰 수집
        new_rows = parse_reviews_now(d, product_url)
        added = 0
        for r in new_rows:
            key = (r["review_date"], r["review_text"])
            if key not in seen:
                rows.append(r); seen.add(key); added += 1
                if len(rows) >= max_reviews:
                    break
        pages += 1
        tries += 1
        if added == 0 and not changed:
            break

    return rows[:max_reviews] if max_reviews else rows

# =========================== 상품 상세 크롤 ====================
def crawl_product(d, url:str)->List[Dict[str,Any]]:
    d.get(url)
    time.sleep(1.0)
    # 제목/브랜드
    try:
        title = wait_xpath(d, "//h1", timeout=LONG).text.strip()
    except Exception:
        try: title = wait_css(d, "h1, h2", timeout=MID).text.strip()
        except Exception: title = ""
    brand, product_name = split_brand_from_title(title)

    # 가격/리뷰수
    j = parse_json_ld(d.page_source)
    price = j.get("price") or extract_price_dom(d)
    review_count = j.get("review_count") or extract_reviewcount_dom(d)

    # 리뷰 페이지네이션 수집
    reviews = paginate_and_collect_reviews(d, url, MAX_REVIEWS_PER_PRODUCT)
    if not reviews:
        reviews = [{"product_url":url, "review_date":None, "review_text":None}]

    rows=[]
    for r in reviews:
        rows.append({
            "category_name": CATEGORY_NAME,
            "product_name": product_name,
            "brand": brand,
            "product_url": url,
            "price": price,
            "review_count": review_count,
            "review_date": r["review_date"],
            "review_text": r["review_text"],
        })
    return rows

# =========================== 메인 =============================
def main():
    d = make_driver()
    all_rows: List[Dict[str,Any]] = []
    try:
        # 리스트에서 링크 모으기 (예시: 1페이지만)
        links, seen = [], set()
        for p in range(1, PAGES_TO_CRAWL+1):
            page_links = collect_links_from_list_page(d, p)
            for lk in page_links:
                if lk not in seen:
                    seen.add(lk); links.append(lk)
        if MAX_PRODUCTS:
            links = links[:MAX_PRODUCTS]
        print(f"[INFO] 대상 상품 {len(links)}개")

        for i, url in enumerate(links, 1):
            print(f"[{i}/{len(links)}] {url}")
            try:
                rows = crawl_product(d, url)
                all_rows.extend(rows)
            except Exception as e:
                print(f"  └─ 에러: {e}")

    finally:
        d.quit()

    out = f"kurly_meal_rice_reviews_{datetime.now().strftime('%Y%m%d_%H%M')}.csv"
    cols = ["category_name","product_name","brand","product_url","price","review_count","review_date","review_text"]
    pd.DataFrame(all_rows, columns=cols).to_csv(out, index=False, encoding="utf-8-sig")
    print(f"\n[DONE] 저장: {out} (행 {len(all_rows)})")

if __name__ == "__main__":
    main()
```

```python
# -*- coding: utf-8 -*-
"""
Kurly 도시락·밥류(912003) → 각 상품 상세 → 리뷰 최대 500개
- 대상: 상품명에 ["도시락","김밥","컵밥","죽"] 중 하나라도 포함된 상품만
- 리뷰 '다음(>)' 버튼을 정확히 1개만 잡아 클릭 (iframe 대응, 무한 스크롤 방지)
저장 컬럼: category_name, product_name, brand, product_url, price, review_count, review_date, review_text
"""

import re, json, time
from datetime import datetime
from typing import List, Dict, Any, Optional, Tuple

import pandas as pd
from bs4 import BeautifulSoup

import undetected_chromedriver as uc
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.common.exceptions import TimeoutException

# =========================== 설정 ===========================
CATEGORY_ID = "912003"               # 도시락·밥류
CATEGORY_NAME = "도시락·밥류"
PAGES_TO_CRAWL = 1                   # 리스트 페이지 수 (예시: 1페이지만)
MAX_PRODUCTS = 5                  # None = 전부, 정수면 상한
MAX_REVIEWS_PER_PRODUCT = 10       # ★ 상품당 최대 리뷰 수
PAGINATION_TRIES_LIMIT = 1000        # 페이지 넘김 상한
HEADLESS = False
WIN_SIZE = "1920,1080"
SHORT, MID, LONG = 3, 7, 12
DIGIT_ONLY = re.compile(r"^[0-9,]+$")
DATE_PAT = re.compile(r"\d{4}\.\d{1,2}\.\d{1,2}")

# 키워드 필터 (상품명 기준)
TARGET_KEYWORDS = ["도시락", "김밥", "컵밥", "죽"]

def is_target_product(name: str) -> bool:
    if not name: return False
    return any(k in name for k in TARGET_KEYWORDS)

# =========================== 드라이버 =========================
def make_driver() -> uc.Chrome:
    opts = uc.ChromeOptions()
    if HEADLESS:
        opts.add_argument("--headless=new")
    opts.add_argument("--disable-dev-shm-usage")
    opts.add_argument("--no-sandbox")
    opts.add_argument("--lang=ko_KR")
    opts.add_argument(f"--window-size={WIN_SIZE}")
    opts.add_argument("--start-maximized")
    opts.add_argument("--disable-blink-features=AutomationControlled")
    service = Service(ChromeDriverManager().install())
    d = uc.Chrome(service=service, options=opts)
    d.set_page_load_timeout(60)
    return d

def wait_css(d, css, timeout=MID):
    return WebDriverWait(d, timeout).until(EC.presence_of_element_located((By.CSS_SELECTOR, css)))

def wait_xpath(d, xp, timeout=MID):
    return WebDriverWait(d, timeout).until(EC.presence_of_element_located((By.XPATH, xp)))

def js_click(d, el):
    d.execute_script("arguments[0].scrollIntoView({block:'center'});", el)
    time.sleep(0.15)
    d.execute_script("arguments[0].click();", el)

# =========================== 리스트 → 링크 ====================
def list_page_url(page:int)->str:
    return f"https://www.kurly.com/categories/{CATEGORY_ID}?filters=&page={page}"

def collect_links_from_list_page(d, page:int)->List[str]:
    d.get(list_page_url(page))
    wait_css(d, "div#container")
    # 카드가 lazy-load → 적당히만 스크롤
    for _ in range(8):
        d.execute_script("window.scrollBy(0, 1400);")
        time.sleep(0.5)
    anchors = d.find_elements(By.CSS_SELECTOR, "a[href*='/goods/']")
    seen, links = set(), []
    for a in anchors:
        href = a.get_attribute("href") or ""
        if "/goods/" in href and href not in seen:
            seen.add(href); links.append(href)
    return links

# =========================== 가격/리뷰수 ======================
def parse_json_ld(html:str)->Dict[str,Any]:
    soup = BeautifulSoup(html, "lxml")
    out = {}
    for tag in soup.select('script[type="application/ld+json"]'):
        try:
            j = json.loads(tag.string)
        except Exception:
            continue
        objs = j if isinstance(j, list) else [j]
        for o in objs:
            if isinstance(o, dict) and o.get("@type")=="Product":
                # price
                off = o.get("offers")
                price = None
                if isinstance(off, dict):
                    price = off.get("price")
                elif isinstance(off, list):
                    for oo in off:
                        if isinstance(oo, dict) and oo.get("price"):
                            price = oo["price"]; break
                if price:
                    try: out["price"] = int(str(price).replace(",", ""))
                    except: pass
                # reviewCount
                ag = o.get("aggregateRating")
                if isinstance(ag, dict) and ag.get("reviewCount") is not None:
                    try: out["review_count"] = int(str(ag["reviewCount"]).replace(",", ""))
                    except: pass
                return out
    return out

def extract_price_dom(d)->Optional[int]:
    try:
        h1 = wait_xpath(d, "//h1", timeout=LONG)
        box = h1.find_element(By.XPATH, "./ancestor::section[1] | ./ancestor::div[2]")
        spans = box.find_elements(By.XPATH, ".//span[normalize-space()]")
        cands = []
        for s in spans:
            t = s.text.strip()
            if t and "%" not in t and DIGIT_ONLY.match(t):
                cands.append(int(t.replace(",", "")))
        if cands: return cands[0]
    except Exception:
        pass
    return None

def extract_reviewcount_dom(d)->Optional[int]:
    try:
        node = d.find_element(By.XPATH, "//nav//*[contains(.,'후기')]/ancestor::li")
        txt = node.text.replace(",", "")
        m = re.search(r"후기\s*\(?\s*(\d+)", txt)
        if m: return int(m.group(1))
    except Exception:
        pass
    return None

# =========================== 제목 → 브랜드/상품명 ===============
def split_brand_from_title(title:str)->Tuple[Optional[str], str]:
    if not title: return None, ""
    m = re.match(r"\s*\[([^\]]+)\]\s*(.*)", title)
    if m:
        brand = m.group(1).strip()
        name = m.group(2).strip() or title.strip()
        return brand, name
    return None, title.strip()

# =========================== 리뷰 섹션 조작 =====================
def get_review_section(d):
    """'상품 후기' 섹션 루트 반환(없으면 None). 이 섹션이 보이면 글로벌 스크롤을 멈춘다."""
    xps = [
        "//section[@aria-label='상품 후기' or contains(@aria-label,'상품 후기')]",
        "//*[@id='review' and (self::section or self::div)]",
        "//section[contains(@class,'review')]",
        "//div[@id='review']"
    ]
    for xp in xps:
        try:
            el = d.find_element(By.XPATH, xp)
            if el.is_displayed():
                d.execute_script("arguments[0].scrollIntoView({block:'center'});", el)
                time.sleep(0.4)
                return el
        except Exception:
            continue
    return None

def switch_into_review_iframe(d)->bool:
    """리뷰 근처 iframe이 있으면 들어가고 True 반환. 없으면 False."""
    try:
        frames = d.find_elements(By.CSS_SELECTOR, "#review iframe, section[aria-label] iframe, section[id*='review'] iframe, section[class*='review'] iframe")
        for fr in frames:
            if fr.is_displayed():
                d.switch_to.frame(fr)
                return True
    except Exception:
        pass
    return False

def leave_iframe(d):
    try:
        d.switch_to.default_content()
    except Exception:
        pass

def _find_next_button_in(node):
    """리뷰 섹션 안에서 '다음' 버튼 후보를 순차 시도(Union 제거)."""
    candidates = []
    # 1) '다음' 고정 텍스트(div, span)
    for xp in [
        ".//button[not(@disabled)][.//div[normalize-space()='다음' or normalize-space()='다음 >']]",
        ".//button[not(@disabled)][.//span[normalize-space()='다음' or normalize-space()='다음 >']]",
        # 2) Prev/Next 두 개 버튼 컨테이너의 2번째 버튼
        ".//div[./button][./button[2]]/button[2][not(@disabled)]",
        # 3) '>' 포함(아이콘 + 텍스트 케이스)
        ".//button[not(@disabled)][contains(normalize-space(.), '>')]",
    ]:
        try:
            els = node.find_elements(By.XPATH, xp)
            vis = [e for e in els if e.is_displayed()]
            if vis:
                candidates.extend(vis)
        except Exception:
            continue
    # '이전' 제외, 여러 개면 우측(마지막) 우선
    filtered = []
    for b in candidates:
        try:
            t = (b.text or "").strip()
            if "이전" not in t:
                filtered.append(b)
        except Exception:
            continue
    if not filtered:
        return None
    if len(filtered) == 1:
        return filtered[0]
    # '다음' 포함 우선, 아니면 마지막
    pri = [b for b in filtered if "다음" in (b.text or "")]
    return pri[0] if len(pri) == 1 else filtered[-1]

def click_review_next(d)->bool:
    """리뷰 섹션의 '다음' 버튼을 정확히 1개 찾아 클릭. iframe 안/밖 모두 시도."""
    # 1) 기본 컨텍스트
    sec = get_review_section(d)
    if sec:
        btn = _find_next_button_in(sec)
        if btn:
            js_click(d, btn)
            return True

    # 2) iframe 내부
    entered = switch_into_review_iframe(d)
    try:
        if entered:
            for xp in [
                "//button[not(@disabled)][.//*[normalize-space()='다음' or normalize-space()='다음 >']]", 
                "//div[./button][./button[2]]/button[2][not(@disabled)]",
                "//button[not(@disabled)][contains(normalize-space(.), '>') and not(contains(.,'이전'))]"
            ]:
                try:
                    els = d.find_elements(By.XPATH, xp)
                    vis = [e for e in els if e.is_displayed()]
                    if vis:
                        js_click(d, vis[-1] if len(vis)>1 else vis[0])
                        return True
                except Exception:
                    continue
    finally:
        if entered:
            leave_iframe(d)

    # 3) 최후: 기본 컨텍스트에서 텍스트 기반
    try:
        btn = d.find_element(By.XPATH, "//button[not(@disabled)][.//*[contains(.,'다음') or contains(.,'>')]][last()]")
        js_click(d, btn)
        return True
    except Exception:
        return False

def find_first_review_card(d):
    xps = [
        "//section[@aria-label='상품 후기']//article[1]",
        "//*[@id='review']//article[1]",
        "//article[1]"
    ]
    for xp in xps:
        try:
            el = d.find_element(By.XPATH, xp)
            if el.is_displayed():
                return el
        except Exception:
            continue
    return None

def parse_reviews_now(d, product_url:str)->List[Dict[str,Any]]:
    soup = BeautifulSoup(d.page_source, "lxml")
    root = soup.select_one("section[aria-label='상품 후기']") or \
           soup.select_one("#review") or \
           soup.select_one("section[class*='review']") or \
           soup
    cards = root.select("article")
    if not cards: cards = root.select("li")
    rows=[]
    for tag in cards:
        ps = [p.get_text(" ", strip=True) for p in tag.select("p") if p.get_text(strip=True)]
        body = max(ps, key=len) if ps else tag.get_text(" ", strip=True)
        if not body or len(body)<10: continue
        m = DATE_PAT.search(tag.get_text(" ", strip=True))
        date_str = m.group(0) if m else None
        rows.append({"product_url":product_url, "review_date":date_str, "review_text":body})
        if len(rows) >= MAX_REVIEWS_PER_PRODUCT:
            break
    return rows

def paginate_and_collect_reviews(d, product_url:str, max_reviews:int=MAX_REVIEWS_PER_PRODUCT)->List[Dict[str,Any]]:
    """'후기' 탭 열고, 다음 버튼만 사용해 최대 max_reviews까지 수집."""
    # 후기 탭 열기
    for xp in [
        "//nav//li[.//span[contains(.,'후기')]]",
        "//nav//*[self::a or self::button][contains(.,'후기')]",
        "//button[contains(.,'후기')]",
        "//a[contains(.,'후기')]",
    ]:
        try:
            el = d.find_element(By.XPATH, xp)
            js_click(d, el); time.sleep(0.6); break
        except Exception:
            continue

    # 첫 페이지 수집
    rows = parse_reviews_now(d, product_url)
    seen = set((r["review_date"], r["review_text"]) for r in rows)

    # 다음 버튼 클릭 반복 (무한 스크롤 없이 버튼만)
    tries, pages = 0, 0
    while len(rows) < max_reviews and tries < PAGINATION_TRIES_LIMIT:
        base = find_first_review_card(d)  # 페이지 전/후 비교용

        # 버튼이 안 보이면 소량 스크롤 탐색(찾자마자 스크롤 중단)
        btn = None
        sec = get_review_section(d)
        if sec:
            btn = _find_next_button_in(sec)
        if not btn:
            for _ in range(6):
                btn = _find_next_button_in(get_review_section(d) or sec) if get_review_section(d) else None
                if btn: break
                d.execute_script("window.scrollBy(0, 650);")
                time.sleep(0.3)

        if not btn and not click_review_next(d):  # 최후 전역/iframe 시도
            break

        if btn:
            js_click(d, btn)

        # 페이지 변경 대기
        changed = False
        try:
            if base:
                WebDriverWait(d, LONG).until(EC.staleness_of(base))
                changed = True
            else:
                time.sleep(1.0)
                changed = True
        except TimeoutException:
            pass

        # 새 리뷰 수집
        new_rows = parse_reviews_now(d, product_url)
        added = 0
        for r in new_rows:
            key = (r["review_date"], r["review_text"])
            if key not in seen:
                rows.append(r); seen.add(key); added += 1
                if len(rows) >= max_reviews:
                    break
        pages += 1
        tries += 1
        if added == 0 and not changed:
            break

    return rows[:max_reviews] if max_reviews else rows

# =========================== 상품 상세 크롤 ====================
def crawl_product(d, url:str)->List[Dict[str,Any]]:
    d.get(url)
    time.sleep(1.0)
    # 제목/브랜드
    try:
        title = wait_xpath(d, "//h1", timeout=LONG).text.strip()
    except Exception:
        try: title = wait_css(d, "h1, h2", timeout=MID).text.strip()
        except Exception: title = ""
    brand, product_name = split_brand_from_title(title)

    # ★ 키워드 필터: 상품명이 대상이 아니면 스킵
    if not is_target_product(product_name):
        print(f"  └─ 스킵 (키워드 불일치): {product_name}")
        return []

    # 가격/리뷰수
    j = parse_json_ld(d.page_source)
    price = j.get("price") or extract_price_dom(d)
    review_count = j.get("review_count") or extract_reviewcount_dom(d)

    # 리뷰 페이지네이션 수집
    reviews = paginate_and_collect_reviews(d, url, MAX_REVIEWS_PER_PRODUCT)
    if not reviews:
        reviews = [{"product_url":url, "review_date":None, "review_text":None}]

    rows=[]
    for r in reviews:
        rows.append({
            "category_name": CATEGORY_NAME,
            "product_name": product_name,
            "brand": brand,
            "product_url": url,
            "price": price,
            "review_count": review_count,
            "review_date": r["review_date"],
            "review_text": r["review_text"],
        })
    return rows

# =========================== 메인 =============================
def main():
    d = make_driver()
    all_rows: List[Dict[str,Any]] = []
    try:
        # 리스트에서 링크 모으기 (예시: 1페이지만)
        links, seen = [], set()
        for p in range(1, PAGES_TO_CRAWL+1):
            page_links = collect_links_from_list_page(d, p)
            for lk in page_links:
                if lk not in seen:
                    seen.add(lk); links.append(lk)
        if MAX_PRODUCTS:
            links = links[:MAX_PRODUCTS]
        print(f"[INFO] 대상 상품(링크) {len(links)}개")

        for i, url in enumerate(links, 1):
            print(f"[{i}/{len(links)}] {url}")
            try:
                rows = crawl_product(d, url)
                all_rows.extend(rows)
            except Exception as e:
                print(f"  └─ 에러: {e}")

    finally:
        d.quit()

    out = f"kurly_meal_rice_reviews_{datetime.now().strftime('%Y%m%d_%H%M')}.csv"
    cols = ["category_name","product_name","brand","product_url","price","review_count","review_date","review_text"]
    pd.DataFrame(all_rows, columns=cols).to_csv(out, index=False, encoding="utf-8-sig")
    print(f"\n[DONE] 저장: {out} (행 {len(all_rows)})")

if __name__ == "__main__":
    main()
```

```python
# -*- coding: utf-8 -*-
"""
Kurly 도시락·밥류(912003)
- 리스트에서 product_name 먼저 추출 후 ["도시락","김밥","컵밥","죽"] 포함 상품만 상세 진입
- 상품당 최대 500개 리뷰 수집(다음 버튼 클릭, iframe 대응, 무한 스크롤 방지)
- 저장 컬럼: category_name, product_name, brand, product_url, price, review_count, review_date, review_text
"""

import re, json, time
from datetime import datetime
from typing import List, Dict, Any, Optional, Tuple

import pandas as pd
from bs4 import BeautifulSoup

import undetected_chromedriver as uc
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.common.exceptions import TimeoutException

# =========================== 설정 ===========================
CATEGORY_ID = "912003"
CATEGORY_NAME = "도시락·밥류"
PAGES_TO_CRAWL = 3
MAX_PRODUCTS = None                # None=모두, 정수면 상한
MAX_REVIEWS_PER_PRODUCT = 500
PAGINATION_TRIES_LIMIT = 1000
HEADLESS = False
WIN_SIZE = "1920,1080"
SHORT, MID, LONG = 3, 7, 12
DIGIT_ONLY = re.compile(r"^[0-9,]+$")
DATE_PAT = re.compile(r"\d{4}\.\d{1,2}\.\d{1,2}")
TARGET_KEYWORDS = ["도시락", "김밥", "컵밥", "죽"]

def is_target_product(name: str) -> bool:
    return bool(name) and any(k in name for k in TARGET_KEYWORDS)

# =========================== 드라이버 =========================
def make_driver() -> uc.Chrome:
    opts = uc.ChromeOptions()
    if HEADLESS: opts.add_argument("--headless=new")
    opts.add_argument("--disable-dev-shm-usage"); opts.add_argument("--no-sandbox")
    opts.add_argument("--lang=ko_KR"); opts.add_argument(f"--window-size={WIN_SIZE}")
    opts.add_argument("--start-maximized"); opts.add_argument("--disable-blink-features=AutomationControlled")
    service = Service(ChromeDriverManager().install())
    d = uc.Chrome(service=service, options=opts)
    d.set_page_load_timeout(60)
    return d

def wait_css(d, css, timeout=MID):  return WebDriverWait(d, timeout).until(EC.presence_of_element_located((By.CSS_SELECTOR, css)))
def wait_xpath(d, xp, timeout=MID): return WebDriverWait(d, timeout).until(EC.presence_of_element_located((By.XPATH, xp)))

def js_click(d, el):
    d.execute_script("arguments[0].scrollIntoView({block:'center'});", el)
    time.sleep(0.15)
    d.execute_script("arguments[0].click();", el)

# =========================== 리스트 → (이름,링크) 수집 ====================
def list_page_url(page:int)->str:
    return f"https://www.kurly.com/categories/{CATEGORY_ID}?filters=&page={page}"

def _guess_name_from_card(anchor) -> str:
    # 카드 컨테이너 추정
    try:
        card = anchor.find_element(By.XPATH, "./ancestor::li[1] | ./ancestor::div[contains(@class,'grid') or contains(@class,'item')][1]")
    except Exception:
        card = None
    texts = []
    nodes = []
    if card:
        nodes += card.find_elements(By.XPATH, ".//strong|.//span|.//em|.//div")
    else:
        nodes += anchor.find_elements(By.XPATH, ".//strong|.//span|.//em|.//div")
    for n in nodes:
        try:
            t = n.text.strip()
            if not t: continue
            if "담기" in t: continue
            if "원" in t and DIGIT_ONLY.match(re.sub(r"[원,\s]","",t)):  # 가격 제거
                continue
            if len(t) > 100: continue
            texts.append(t)
        except: pass
    if texts:
        def score(s): return len(s) + (20 if re.search("[가-힣]", s) else 0) - (10 if "원" in s else 0)
        return max(texts, key=score)
    return (anchor.get_attribute("aria-label") or anchor.get_attribute("title") or anchor.text or "").strip()

def collect_filtered_links_from_list_page(d, page:int)->List[Tuple[str,str]]:
    d.get(list_page_url(page))
    wait_css(d, "div#container")
    # lazy-load 대비 적당 스크롤 (버튼 찾으면 더 이상 스크롤 안 함)
    for _ in range(10):
        d.execute_script("window.scrollBy(0, 1400);"); time.sleep(0.45)
    anchors = d.find_elements(By.CSS_SELECTOR, "a[href*='/goods/']")
    seen, results = set(), []
    for a in anchors:
        href = a.get_attribute("href") or ""
        if "/goods/" not in href or href in seen: continue
        name = _guess_name_from_card(a)
        if is_target_product(name):
            seen.add(href)
            results.append((name, href))
    return results

# =========================== 가격/리뷰수 ======================
def parse_json_ld(html:str)->Dict[str,Any]:
    soup = BeautifulSoup(html, "lxml"); out = {}
    for tag in soup.select('script[type="application/ld+json"]'):
        try: j = json.loads(tag.string)
        except: continue
        for o in (j if isinstance(j, list) else [j]):
            if isinstance(o, dict) and o.get("@type")=="Product":
                off = o.get("offers"); price = None
                if isinstance(off, dict): price = off.get("price")
                elif isinstance(off, list):
                    for oo in off:
                        if isinstance(oo, dict) and oo.get("price"): price = oo["price"]; break
                if price:
                    try: out["price"] = int(str(price).replace(",", ""))
                    except: pass
                ag = o.get("aggregateRating")
                if isinstance(ag, dict) and ag.get("reviewCount") is not None:
                    try: out["review_count"] = int(str(ag["reviewCount"]).replace(",", ""))
                    except: pass
                return out
    return out

def extract_price_dom(d)->Optional[int]:
    try:
        h1 = wait_xpath(d, "//h1", timeout=LONG)
        box = h1.find_element(By.XPATH, "./ancestor::section[1] | ./ancestor::div[2]")
        spans = box.find_elements(By.XPATH, ".//span[normalize-space()]")
        cands = []
        for s in spans:
            t = s.text.strip()
            if t and "%" not in t and DIGIT_ONLY.match(t):
                cands.append(int(t.replace(",", "")))
        if cands: return cands[0]
    except: pass
    return None

def extract_reviewcount_dom(d)->Optional[int]:
    try:
        node = d.find_element(By.XPATH, "//nav//*[contains(.,'후기')]/ancestor::li")
        txt = node.text.replace(",", ""); m = re.search(r"후기\s*\(?\s*(\d+)", txt)
        if m: return int(m.group(1))
    except: pass
    return None

# =========================== 제목 → 브랜드/상품명 ===============
def split_brand_from_title(title:str)->Tuple[Optional[str], str]:
    if not title: return None, ""
    m = re.match(r"\s*\[([^\]]+)\]\s*(.*)", title)
    if m:
        brand = m.group(1).strip(); name = m.group(2).strip() or title.strip()
        return brand, name
    return None, title.strip()

# =========================== 리뷰 섹션/페이지네이션 =============
def get_review_section(d):
    xps = [
        "//section[@aria-label='상품 후기' or contains(@aria-label,'상품 후기')]",
        "//*[@id='review' and (self::section or self::div)]",
        "//section[contains(@class,'review')]", "//div[@id='review']",
    ]
    for xp in xps:
        try:
            el = d.find_element(By.XPATH, xp)
            if el.is_displayed():
                d.execute_script("arguments[0].scrollIntoView({block:'center'});", el)
                time.sleep(0.4); return el
        except: continue
    return None

def switch_into_review_iframe(d)->bool:
    try:
        frames = d.find_elements(By.CSS_SELECTOR, "#review iframe, section[aria-label] iframe, section[id*='review'] iframe, section[class*='review'] iframe")
        for fr in frames:
            if fr.is_displayed(): d.switch_to.frame(fr); return True
    except: pass
    return False

def leave_iframe(d):
    try: d.switch_to.default_content()
    except: pass

def _find_next_button_in(node):
    candidates = []
    for xp in [
        ".//button[not(@disabled)][.//div[normalize-space()='다음' or normalize-space()='다음 >']]",
        ".//button[not(@disabled)][.//span[normalize-space()='다음' or normalize-space()='다음 >']]",
        ".//div[./button][./button[2]]/button[2][not(@disabled)]",
        ".//button[not(@disabled)][contains(normalize-space(.), '>')]",
    ]:
        try:
            els = node.find_elements(By.XPATH, xp)
            vis = [e for e in els if e.is_displayed()]
            if vis: candidates.extend(vis)
        except: continue
    filtered = []
    for b in candidates:
        try:
            t = (b.text or "").strip()
            if "이전" not in t: filtered.append(b)
        except: pass
    if not filtered: return None
    if len(filtered)==1: return filtered[0]
    pri = [b for b in filtered if "다음" in (b.text or "")]
    return pri[0] if len(pri)==1 else filtered[-1]

def click_review_next(d)->bool:
    sec = get_review_section(d)
    if sec:
        btn = _find_next_button_in(sec)
        if btn: js_click(d, btn); return True
    entered = switch_into_review_iframe(d)
    try:
        if entered:
            for xp in [
                "//button[not(@disabled)][.//*[normalize-space()='다음' or normalize-space()='다음 >']]", 
                "//div[./button][./button[2]]/button[2][not(@disabled)]",
                "//button[not(@disabled)][contains(normalize-space(.), '>') and not(contains(.,'이전'))]"
            ]:
                try:
                    els = d.find_elements(By.XPATH, xp)
                    vis = [e for e in els if e.is_displayed()]
                    if vis: js_click(d, vis[-1] if len(vis)>1 else vis[0]); return True
                except: continue
    finally:
        if entered: leave_iframe(d)
    try:
        btn = d.find_element(By.XPATH, "//button[not(@disabled)][.//*[contains(.,'다음') or contains(.,'>')]][last()]")
        js_click(d, btn); return True
    except: return False

def find_first_review_card(d):
    for xp in ["//section[@aria-label='상품 후기']//article[1]", "//*[@id='review']//article[1]", "//article[1]"]:
        try:
            el = d.find_element(By.XPATH, xp)
            if el.is_displayed(): return el
        except: continue
    return None

def parse_reviews_now(d, product_url:str)->List[Dict[str,Any]]:
    soup = BeautifulSoup(d.page_source, "lxml")
    root = soup.select_one("section[aria-label='상품 후기']") or soup.select_one("#review") or soup.select_one("section[class*='review']") or soup
    cards = root.select("article") or root.select("li")
    rows=[]
    for tag in cards:
        ps = [p.get_text(" ", strip=True) for p in tag.select("p") if p.get_text(strip=True)]
        body = max(ps, key=len) if ps else tag.get_text(" ", strip=True)
        if not body or len(body)<10: continue
        m = DATE_PAT.search(tag.get_text(" ", strip=True))
        date_str = m.group(0) if m else None
        rows.append({"product_url":product_url, "review_date":date_str, "review_text":body})
        if len(rows) >= MAX_REVIEWS_PER_PRODUCT: break
    return rows

def paginate_and_collect_reviews(d, product_url:str, max_reviews:int=MAX_REVIEWS_PER_PRODUCT)->List[Dict[str,Any]]:
    for xp in [
        "//nav//li[.//span[contains(.,'후기')]]",
        "//nav//*[self::a or self::button][contains(.,'후기')]",
        "//button[contains(.,'후기')]", "//a[contains(.,'후기')]",
    ]:
        try:
            el = d.find_element(By.XPATH, xp); js_click(d, el); time.sleep(0.6); break
        except: continue

    rows = parse_reviews_now(d, product_url)
    seen = set((r["review_date"], r["review_text"]) for r in rows)

    tries = 0
    while len(rows) < max_reviews and tries < PAGINATION_TRIES_LIMIT:
        base = find_first_review_card(d)
        btn = None; sec = get_review_section(d)
        if sec: btn = _find_next_button_in(sec)
        if not btn:
            for _ in range(6):
                sec = get_review_section(d)
                btn = _find_next_button_in(sec) if sec else None
                if btn: break
                d.execute_script("window.scrollBy(0, 650);"); time.sleep(0.3)
        if not btn and not click_review_next(d): break
        if btn: js_click(d, btn)

        changed = False
        try:
            if base: WebDriverWait(d, LONG).until(EC.staleness_of(base)); changed = True
            else: time.sleep(1.0); changed = True
        except TimeoutException: pass

        new_rows = parse_reviews_now(d, product_url)
        added = 0
        for r in new_rows:
            key = (r["review_date"], r["review_text"])
            if key not in seen:
                rows.append(r); seen.add(key); added += 1
                if len(rows) >= max_reviews: break
        tries += 1
        if added == 0 and not changed: break
    return rows[:max_reviews] if max_reviews else rows

# =========================== 상세 크롤 =========================
def crawl_product(d, url:str)->List[Dict[str,Any]]:
    d.get(url); time.sleep(1.0)
    try: title = wait_xpath(d, "//h1", timeout=LONG).text.strip()
    except Exception:
        try: title = wait_css(d, "h1, h2", timeout=MID).text.strip()
        except: title = ""
    brand, product_name = split_brand_from_title(title)

    j = parse_json_ld(d.page_source)
    price = j.get("price") or extract_price_dom(d)
    review_count = j.get("review_count") or extract_reviewcount_dom(d)

    reviews = paginate_and_collect_reviews(d, url, MAX_REVIEWS_PER_PRODUCT)
    if not reviews: reviews = [{"product_url":url, "review_date":None, "review_text":None}]

    rows=[]
    for r in reviews:
        rows.append({
            "category_name": CATEGORY_NAME,
            "product_name": product_name,
            "brand": brand,
            "product_url": url,
            "price": price,
            "review_count": review_count,
            "review_date": r["review_date"],
            "review_text": r["review_text"],
        })
    return rows

# =========================== 메인 =============================
def main():
    d = make_driver()
    all_rows: List[Dict[str,Any]] = []
    try:
        # 1) 리스트에서 (이름,링크) 수집 후 키워드로 선필터
        pairs, seen = [], set()
        for p in range(1, PAGES_TO_CRAWL+1):
            pairs.extend(collect_filtered_links_from_list_page(d, p))
        if MAX_PRODUCTS: pairs = pairs[:MAX_PRODUCTS]
        print(f"[INFO] 필터 통과 상품 {len(pairs)}개")

        # 2) 필터링된 URL만 상세 진입
        for i, (name, url) in enumerate(pairs, 1):
            print(f"[{i}/{len(pairs)}] {name} -> {url}")
            try:
                rows = crawl_product(d, url)
                all_rows.extend(rows)
            except Exception as e:
                print(f"  └─ 에러: {e}")
    finally:
        d.quit()

    out = f"kurly_meal_rice_reviews_{datetime.now().strftime('%Y%m%d_%H%M')}.csv"
    cols = ["category_name","product_name","brand","product_url","price","review_count","review_date","review_text"]
    pd.DataFrame(all_rows, columns=cols).to_csv(out, index=False, encoding="utf-8-sig")
    print(f"\n[DONE] 저장: {out} (행 {len(all_rows)})")

if __name__ == "__main__":
    main()
```

```python
# -*- coding: utf-8 -*-
"""
Kurly 리뷰 형태소 분석 → 키워드 빈도 상위 100 저장
- 입력 CSV: review_text 컬럼 필수
- 토크나이저: KoNLPy Okt(normalize=True, stem=True)
- 품사 필터: Noun / Adjective / Verb
- 불용어, 숫자/영문-only, 1글자 토큰 제외
- 출력:
  1) <원본파일명>_top100_keywords_YYYYMMDD_HHMM.csv
  2) <원본파일명>_all_keywords_counts_YYYYMMDD_HHMM.csv
"""

import os
import re
from datetime import datetime
from collections import Counter

import pandas as pd
from konlpy.tag import Okt

# ======= 사용자 설정 =======
INPUT_CSV_PATH = r"/mnt/data/kurly_meal_키워드필터링_500_20250923_1242.csv"  # 파일 경로 수정 가능
TEXT_COLUMN = "review_text"
TOP_N = 100

# 자주 쓰는 한국어 불용어(필요에 맞게 추가/수정)
STOPWORDS = {
    "이","그","저","것","수","들","등","및","더","좀","거","때","건","듯","같","요","은","는","이요",
    "정말","진짜","너무","매우","아주","조금","그리고","하지만","그래도","또","또한","에서","으로","까지",
    "제품","상품","구매","사용","배송","도착","포장","가격","구입","주문","추천","만족","불만",
    "맛","향","양","상태","퀄리티","브랜드","마켓컬리","컬리","사진","리뷰","후기","지금","오늘","요즘",
    "한","두","세","네","번","개","정도","만","때문","거나","이라","라서","해서","인데",
}

# ======= 전처리/토큰화 =======
def clean_text(s: str) -> str:
    if not isinstance(s, str):
        return ""
    s = s.strip()
    s = re.sub(r"http[s]?://\S+|www\.\S+", " ", s)   # URL
    s = re.sub(r"<[^>]+>", " ", s)                   # HTML
    s = re.sub(r"[\u2600-\u27BF]", " ", s)           # 각종 기호
    s = re.sub(r"[\U00010000-\U0010FFFF]", " ", s)   # 이모지
    s = re.sub(r"[^0-9A-Za-z가-힣\s]", " ", s)         # 한/영/숫자/공백만
    s = re.sub(r"\s+", " ", s)
    return s.strip()

def tokenize_korean(texts):
    okt = Okt()
    keep_pos = {"Noun", "Adjective", "Verb"}
    tokens = []
    for t in texts:
        t = clean_text(t)
        if not t:
            continue
        for tok, pos in okt.pos(t, norm=True, stem=True):
            if pos not in keep_pos:
                continue
            if tok in STOPWORDS:
                continue
            if len(tok) < 2:
                continue
            if tok.isdigit():
                continue
            if re.fullmatch(r"[A-Za-z]+", tok):  # 영문-only 토큰 제거(원하면 주석 처리)
                continue
            tokens.append(tok)
    return tokens

# ======= 실행 =======
def main():
    # 입력
    df = pd.read_csv(INPUT_CSV_PATH, encoding="utf-8-sig")
    if TEXT_COLUMN not in df.columns:
        raise ValueError(f"입력 CSV에 '{TEXT_COLUMN}' 컬럼이 없습니다.")
    texts = df[TEXT_COLUMN].fillna("").astype(str).tolist()

    # 토큰화 & 빈도
    tokens = tokenize_korean(texts)
    counter = Counter(tokens)

    # 전체 빈도표
    all_items = counter.most_common()
    all_df = pd.DataFrame(all_items, columns=["keyword", "freq"])

    # 상위 TOP_N
    top_df = all_df.head(TOP_N)

    # 출력 파일명
    stamp = datetime.now().strftime("%Y%m%d_%H%M")
    base = os.path.splitext(os.path.basename(INPUT_CSV_PATH))[0]
    out_top = f"{base}_top{TOP_N}_keywords_{stamp}.csv"
    out_all = f"{base}_all_keywords_counts_{stamp}.csv"

    # 저장
    top_df.to_csv(out_top, index=False, encoding="utf-8-sig")
    all_df.to_csv(out_all, index=False, encoding="utf-8-sig")

    print(f"[DONE] 상위 {TOP_N} 저장: {out_top}")
    print(f"[DONE] 전체 빈도 저장: {out_all}")
    print("\n[PREVIEW - TOP 20]")
    print(top_df.head(20))

if __name__ == "__main__":
    main()
```

```python
# -*- coding: utf-8 -*-
"""
리뷰 형태소 분석 → 상위 100 키워드 & 전체 빈도 CSV 저장 (Windows 경로 대응)
- 기본 입력: 지정 폴더 내 '가장 최근' CSV 자동 선택 (원하면 FILENAME 지정)
- review_text가 없으면 '리뷰내용','리뷰','후기','review' 등 자동 탐색
"""

import os
import re
import glob
from datetime import datetime
from collections import Counter

import pandas as pd
from konlpy.tag import Okt

# ===== 사용자 설정 =====
DIR_PATH  = r"C:\Users\sagej\Digital_Pyhton_Study\Team_project_0919"  # <- 폴더 경로
FILENAME  = "kurly_meal_키워드필터링_500_20250923_1242.csv"
TEXT_CANDIDATES = ["review_text", "리뷰내용", "리뷰", "후기", "review", "content", "comment"]
TOP_N = 100

# 자주 쓰는 한국어 불용어(원하는 만큼 추가/삭제)
STOPWORDS = {
    "이","그","저","것","수","들","등","및","더","좀","거","때","건","듯","같","요","은","는","이요",
    "정말","진짜","너무","매우","아주","조금","그리고","하지만","그래도","또","또한","에서","으로","까지",
    "제품","상품","구매","사용","배송","도착","포장","가격","구입","주문","추천","만족","불만",
    "맛","향","양","상태","퀄리티","브랜드","마켓컬리","컬리","사진","리뷰","후기","지금","오늘","요즘",
    "한","두","세","네","번","개","정도","만","때문","거나","이라","라서","해서","인데", "하다", "먹다", "좋다"
}

def pick_latest_csv(dir_path: str) -> str:
    csvs = glob.glob(os.path.join(dir_path, "*.csv"))
    if not csvs:
        raise FileNotFoundError(f"폴더에 CSV가 없습니다: {dir_path}")
    csvs.sort(key=os.path.getmtime, reverse=True)
    return csvs[0]

def robust_read_csv(path: str) -> pd.DataFrame:
    errors = []
    for enc in ["utf-8-sig", "utf-8", "cp949", "euc-kr"]:
        try:
            return pd.read_csv(path, encoding=enc)
        except Exception as e:
            errors.append(f"{enc}: {e}")
    raise RuntimeError("CSV 인코딩 읽기 실패:\n" + "\n".join(errors))

def clean_text(s: str) -> str:
    if not isinstance(s, str):
        return ""
    s = s.strip()
    s = re.sub(r"http[s]?://\S+|www\.\S+", " ", s)   # URL
    s = re.sub(r"<[^>]+>", " ", s)                   # HTML
    s = re.sub(r"[\u2600-\u27BF]", " ", s)           # 각종 기호
    s = re.sub(r"[\U00010000-\U0010FFFF]", " ", s)   # 이모지
    s = re.sub(r"[^0-9A-Za-z가-힣\s]", " ", s)         # 한/영/숫자/공백만
    s = re.sub(r"\s+", " ", s)
    return s.strip()

def tokenize_korean(texts):
    okt = Okt()
    keep_pos = {"Noun", "Adjective", "Verb"}
    tokens = []
    for t in texts:
        t = clean_text(t)
        if not t:
            continue
        for tok, pos in okt.pos(t, norm=True, stem=True):
            if pos not in keep_pos:
                continue
            if tok in STOPWORDS:
                continue
            if len(tok) < 2:
                continue
            if tok.isdigit():
                continue
            if re.fullmatch(r"[A-Za-z]+", tok):  # 영문-only 제거(원하면 주석 처리)
                continue
            tokens.append(tok)
    return tokens

def choose_text_column(df: pd.DataFrame) -> str:
    # 1) 후보 컬럼 우선
    for c in TEXT_CANDIDATES:
        if c in df.columns:
            return c
    # 2) 'review'/'리뷰'/'후기'가 이름에 포함된 컬럼 탐색
    lower_map = {c: c.lower() for c in df.columns}
    for c in df.columns:
        lc = lower_map[c]
        if any(k in lc for k in ["review", "리뷰", "후기", "comment", "content", "text"]):
            return c
    # 3) 실패
    raise ValueError(f"리뷰 텍스트 컬럼을 찾지 못했습니다. 후보: {TEXT_CANDIDATES}")

def main():
    # 입력 파일 결정
    if FILENAME:
        input_path = os.path.join(DIR_PATH, FILENAME)
    else:
        input_path = pick_latest_csv(DIR_PATH)

    if not os.path.exists(input_path):
        raise FileNotFoundError(f"파일을 찾을 수 없습니다: {input_path}")

    print(f"[INFO] 입력 파일: {input_path}")

    # 읽기(인코딩 자동 시도)
    df = robust_read_csv(input_path)

    # 텍스트 컬럼 선택
    text_col = choose_text_column(df)
    print(f"[INFO] 리뷰 컬럼: {text_col}")

    texts = df[text_col].fillna("").astype(str).tolist()

    # 토큰화 & 빈도
    tokens = tokenize_korean(texts)
    counter = Counter(tokens)

    # 결과 데이터프레임
    all_items = counter.most_common()
    all_df = pd.DataFrame(all_items, columns=["keyword", "freq"])
    top_df = all_df.head(TOP_N)

    # 출력 경로
    stamp = datetime.now().strftime("%Y%m%d_%H%M")
    base = os.path.splitext(os.path.basename(input_path))[0]
    out_top = os.path.join(DIR_PATH, f"{base}_top{TOP_N}_keywords_{stamp}.csv")
    out_all = os.path.join(DIR_PATH, f"{base}_all_keywords_counts_{stamp}.csv")

    # 저장
    top_df.to_csv(out_top, index=False, encoding="utf-8-sig")
    all_df.to_csv(out_all, index=False, encoding="utf-8-sig")

    print(f"[DONE] 상위 {TOP_N} 저장: {out_top}")
    print(f"[DONE] 전체 빈도 저장: {out_all}")
    print("\n[PREVIEW - TOP 20]")
    print(top_df.head(20))

if __name__ == "__main__":
    main()
```

```python
# -*- coding: utf-8 -*-
"""
리뷰 형태소 분석 → 상위 100 키워드 & 전체 빈도 CSV 저장 (Windows 경로 대응)
- 기본 입력: 지정 폴더 내 '가장 최근' CSV 자동 선택 (원하면 FILENAME 지정)
- review_text가 없으면 '리뷰내용','리뷰','후기','review' 등 자동 탐색
"""

import os
import re
import glob
from datetime import datetime
from collections import Counter

import pandas as pd
from konlpy.tag import Okt

# ===== 사용자 설정 =====
DIR_PATH  = r"C:\Users\sagej\Digital_Pyhton_Study\Team_project_0919"  # <- 폴더 경로
FILENAME  = "kurly_meal_키워드필터링_500_20250923_1242.csv"
TEXT_CANDIDATES = ["review_text", "리뷰내용", "리뷰", "후기", "review", "content", "comment"]
TOP_N = 100

# 자주 쓰는 한국어 불용어(필요에 맞게 추가/수정)
STOPWORDS = {
    "이","그","저","것","수","들","등","및","더","좀","거","때","건","듯","같","요","은","는","이요",
    "정말","진짜","너무","매우","아주","조금","그리고","하지만","그래도","또","또한","에서","으로","까지",
    "제품","상품","구매","사용","배송","도착","포장","가격","구입","주문","추천","만족","불만",
    "맛","향","양","상태","퀄리티","브랜드","마켓컬리","컬리","사진","리뷰","후기","지금","오늘","요즘",
    "한","두","세","네","번","개","정도","만","때문","거나","이라","라서","해서","인데",
}

# ★ 추가 불용어(상위 0~19에서 '간편하다', '다이어트' 제외)
STOPWORDS |= {
    "하다","맛있다","먹다","좋다","있다","김밥","도시락","먹기","같다","자다","이다",
    "간단하다","않다","재다","괜찮다","보다","자주","들다"
}
# 주의: '간편하다', '다이어트'는 불용어에 넣지 않음

def pick_latest_csv(dir_path: str) -> str:
    csvs = glob.glob(os.path.join(dir_path, "*.csv"))
    if not csvs:
        raise FileNotFoundError(f"폴더에 CSV가 없습니다: {dir_path}")
    csvs.sort(key=os.path.getmtime, reverse=True)
    return csvs[0]

def robust_read_csv(path: str) -> pd.DataFrame:
    errors = []
    for enc in ["utf-8-sig", "utf-8", "cp949", "euc-kr"]:
        try:
            return pd.read_csv(path, encoding=enc)
        except Exception as e:
            errors.append(f"{enc}: {e}")
    raise RuntimeError("CSV 인코딩 읽기 실패:\n" + "\n".join(errors))

def clean_text(s: str) -> str:
    if not isinstance(s, str):
        return ""
    s = s.strip()
    s = re.sub(r"http[s]?://\S+|www\.\S+", " ", s)   # URL
    s = re.sub(r"<[^>]+>", " ", s)                   # HTML
    s = re.sub(r"[\u2600-\u27BF]", " ", s)           # 각종 기호
    s = re.sub(r"[\U00010000-\U0010FFFF]", " ", s)   # 이모지
    s = re.sub(r"[^0-9A-Za-z가-힣\s]", " ", s)         # 한/영/숫자/공백만
    s = re.sub(r"\s+", " ", s)
    return s.strip()

def tokenize_korean(texts):
    okt = Okt()
    keep_pos = {"Noun", "Adjective", "Verb"}
    tokens = []
    for t in texts:
        t = clean_text(t)
        if not t:
            continue
        for tok, pos in okt.pos(t, norm=True, stem=True):
            if pos not in keep_pos:
                continue
            if tok in STOPWORDS:
                continue
            if len(tok) < 2:
                continue
            if tok.isdigit():
                continue
            if re.fullmatch(r"[A-Za-z]+", tok):  # 영문-only 제거(원하면 주석 처리)
                continue
            tokens.append(tok)
    return tokens

def choose_text_column(df: pd.DataFrame) -> str:
    # 1) 후보 컬럼 우선
    for c in TEXT_CANDIDATES:
        if c in df.columns:
            return c
    # 2) 'review'/'리뷰'/'후기'가 이름에 포함된 컬럼 탐색
    lower_map = {c: c.lower() for c in df.columns}
    for c in df.columns:
        lc = lower_map[c]
        if any(k in lc for k in ["review", "리뷰", "후기", "comment", "content", "text"]):
            return c
    # 3) 실패
    raise ValueError(f"리뷰 텍스트 컬럼을 찾지 못했습니다. 후보: {TEXT_CANDIDATES}")

def main():
    # 입력 파일 결정
    if FILENAME:
        input_path = os.path.join(DIR_PATH, FILENAME)
    else:
        input_path = pick_latest_csv(DIR_PATH)

    if not os.path.exists(input_path):
        raise FileNotFoundError(f"파일을 찾을 수 없습니다: {input_path}")

    print(f"[INFO] 입력 파일: {input_path}")

    # 읽기(인코딩 자동 시도)
    df = robust_read_csv(input_path)

    # 텍스트 컬럼 선택
    text_col = choose_text_column(df)
    print(f"[INFO] 리뷰 컬럼: {text_col}")

    texts = df[text_col].fillna("").astype(str).tolist()

    # 토큰화 & 빈도
    tokens = tokenize_korean(texts)
    counter = Counter(tokens)

    # 결과 데이터프레임
    all_items = counter.most_common()
    all_df = pd.DataFrame(all_items, columns=["keyword", "freq"])
    top_df = all_df.head(TOP_N)

    # 출력 경로
    stamp = datetime.now().strftime("%Y%m%d_%H%M")
    base = os.path.splitext(os.path.basename(input_path))[0]
    out_top = os.path.join(DIR_PATH, f"{base}_top{TOP_N}_keywords_{stamp}.csv")
    out_all = os.path.join(DIR_PATH, f"{base}_all_keywords_counts_{stamp}.csv")

    # 저장 (형식 동일 유지)
    top_df.to_csv(out_top, index=False, encoding="utf-8-sig")
    all_df.to_csv(out_all, index=False, encoding="utf-8-sig")

    print(f"[DONE] 상위 {TOP_N} 저장: {out_top}")
    print(f"[DONE] 전체 빈도 저장: {out_all}")
    print("\n[PREVIEW - TOP 100]")
    print(top_df.head(100))  # ★ 콘솔에서 100위까지 미리보기

if __name__ == "__main__":
    main()
```

```python
# -*- coding: utf-8 -*-
"""
Kurly 리뷰 형태소 분석 → 단어 임베딩(Word2Vec) → KMeans 군집 분석
- 입력: 리뷰 CSV (review_text / 리뷰 / 후기 등 자동 탐색)
- 토큰화: KoNLPy Okt (norm=True, stem=True), 품사 {Noun, Adjective, Verb}
- 불용어: 이전 코드와 동일(상위 0~19에서 '간편하다', '다이어트' 제외)
- 임베딩: gensim Word2Vec (skip-gram), min_count & window 조절 가능
- 군집: KMeans (k 수 자동탐색 가능; 실루엣 스코어 최적 k 선택)
- 출력:
  1) <base>_word_clusters_k{K}_YYYYMMDD_HHMM.csv
  2) <base>_cluster_summary_k{K}_YYYYMMDD_HHMM.csv
  3) <base>_cluster_sample_reviews_k{K}_YYYYMMDD_HHMM.csv
"""

import os
import re
import glob
import math
from datetime import datetime
from collections import Counter, defaultdict

import pandas as pd
from konlpy.tag import Okt

# 임베딩/클러스터링
from gensim.models import Word2Vec
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# ======================= 사용자 설정 =======================
DIR_PATH  = r"C:\Users\sagej\Digital_Pyhton_Study\Team_project_0919"  # 리뷰 CSV 폴더
FILENAME  = "kurly_meal_키워드필터링_500_20250923_1242.csv"
TEXT_CANDIDATES = ["review_text", "리뷰내용", "리뷰", "후기", "review", "content", "comment"]

# 토큰/임베딩/클러스터 설정
KEEP_POS = {"Noun", "Adjective", "Verb"}
TOP_N_PREVIEW = 100             # 콘솔 미리보기 개수
MIN_TOKEN_LEN = 2               # 2글자 이상만
MIN_TOKEN_FREQ = 20             # 너무 희귀한 단어 제외(코퍼스 규모에 맞춰 조절)
REMOVE_ENG_ONLY = True          # 영문-only 토큰 제거
W2V_VECTOR_SIZE = 200
W2V_WINDOW = 5
W2V_MIN_COUNT = MIN_TOKEN_FREQ  # Word2Vec 자체의 min_count
W2V_EPOCHS = 10
W2V_SG = 1                      # 1=skipgram, 0=CBOW
W2V_WORKERS = 4

# K 자동 탐색 (None이면 자동; 정수 지정 시 고정)
N_CLUSTERS = None
K_SEARCH_RANGE = (5, 15)        # 자동 탐색 범위 (코퍼스에 맞게 조절)

# ======================= 불용어 (이전 코드와 동일) =======================
BASE_STOPWORDS = {
    "이","그","저","것","수","들","등","및","더","좀","거","때","건","듯","같","요","은","는","이요",
    "정말","진짜","너무","매우","아주","조금","그리고","하지만","그래도","또","또한","에서","으로","까지",
    "제품","상품","구매","사용","배송","도착","포장","가격","구입","주문","추천","만족","불만",
    "맛","향","양","상태","퀄리티","브랜드","마켓컬리","컬리","사진","리뷰","후기","지금","오늘","요즘",
    "한","두","세","네","번","개","정도","만","때문","거나","이라","라서","해서","인데",
}

# 상위 0~19 중 '간편하다', '다이어트' 제외하여 불용어에 추가
EXTRA_STOP = {"하다","맛있다","먹다","좋다","있다","김밥","도시락","먹기","같다","자다","이다",
              "간단하다","않다","재다","괜찮다","보다","자주","들다"}
STOPWORDS = set(BASE_STOPWORDS) | set(EXTRA_STOP)
# 주의: '간편하다', '다이어트'는 불용어에 포함하지 않음.

# ======================= 유틸 =======================
def pick_latest_csv(dir_path: str) -> str:
    csvs = glob.glob(os.path.join(dir_path, "*.csv"))
    if not csvs:
        raise FileNotFoundError(f"폴더에 CSV가 없습니다: {dir_path}")
    csvs.sort(key=os.path.getmtime, reverse=True)
    return csvs[0]

def robust_read_csv(path: str) -> pd.DataFrame:
    errors = []
    for enc in ["utf-8-sig", "utf-8", "cp949", "euc-kr"]:
        try:
            return pd.read_csv(path, encoding=enc)
        except Exception as e:
            errors.append(f"{enc}: {e}")
    raise RuntimeError("CSV 인코딩 읽기 실패:\n" + "\n".join(errors))

def choose_text_column(df: pd.DataFrame) -> str:
    for c in TEXT_CANDIDATES:
        if c in df.columns:
            return c
    low = {c: c.lower() for c in df.columns}
    for c in df.columns:
        lc = low[c]
        if any(k in lc for k in ["review", "리뷰", "후기", "comment", "content", "text"]):
            return c
    raise ValueError(f"리뷰 텍스트 컬럼을 찾지 못했습니다. 후보: {TEXT_CANDIDATES}")

def clean_text(s: str) -> str:
    if not isinstance(s, str): return ""
    s = s.strip()
    s = re.sub(r"http[s]?://\S+|www\.\S+", " ", s)
    s = re.sub(r"<[^>]+>", " ", s)
    s = re.sub(r"[\u2600-\u27BF]", " ", s)            # 다양한 기호
    s = re.sub(r"[\U00010000-\U0010FFFF]", " ", s)    # 이모지
    s = re.sub(r"[^0-9A-Za-z가-힣\s]", " ", s)
    s = re.sub(r"\s+", " ", s)
    return s.strip()

def tokenize_korean(texts):
    okt = Okt()
    tokens_per_doc = []
    freq = Counter()
    for t in texts:
        t = clean_text(t)
        if not t:
            tokens_per_doc.append([])
            continue
        toks = []
        for tok, pos in okt.pos(t, norm=True, stem=True):
            if pos not in KEEP_POS:        continue
            if tok in STOPWORDS:           continue
            if len(tok) < MIN_TOKEN_LEN:   continue
            if tok.isdigit():              continue
            if REMOVE_ENG_ONLY and re.fullmatch(r"[A-Za-z]+", tok):
                continue
            toks.append(tok)
        tokens_per_doc.append(toks)
        freq.update(toks)
    return tokens_per_doc, freq

# ======================= 메인 로직 =======================
def main():
    # 입력 파일
    input_path = os.path.join(DIR_PATH, FILENAME) if FILENAME else pick_latest_csv(DIR_PATH)
    if not os.path.exists(input_path):
        raise FileNotFoundError(f"파일을 찾을 수 없습니다: {input_path}")
    print(f"[INFO] 입력 파일: {input_path}")

    df = robust_read_csv(input_path)
    text_col = choose_text_column(df)
    print(f"[INFO] 리뷰 컬럼: {text_col}")

    texts = df[text_col].fillna("").astype(str).tolist()

    # 1) 형태소 분석
    token_seqs, token_freq = tokenize_korean(texts)
    print(f"[INFO] 토큰화 완료: 문서 {len(token_seqs)}개, 고유 토큰 {len(token_freq)}개")

    # 빈도 필터
    vocab = [w for w, c in token_freq.items() if c >= MIN_TOKEN_FREQ]
    if len(vocab) < 5:
        raise RuntimeError(f"유지 토큰이 너무 적습니다(>= {MIN_TOKEN_FREQ}): {len(vocab)}개")
    print(f"[INFO] 빈도≥{MIN_TOKEN_FREQ} 단어 수: {len(vocab)}개")

    # 2) Word2Vec 학습 (리뷰 문장 토큰 시퀀스 사용)
    print(f"[INFO] Word2Vec 학습 시작 (|V|≈{len(vocab)}, dim={W2V_VECTOR_SIZE})")
    model = Word2Vec(
        sentences=token_seqs,
        vector_size=W2V_VECTOR_SIZE,
        window=W2V_WINDOW,
        min_count=W2V_MIN_COUNT,
        sg=W2V_SG,
        negative=10,
        epochs=W2V_EPOCHS,
        workers=W2V_WORKERS,
        seed=42,
    )

    # 3) 단어 임베딩 행렬 생성 (학습된 vocab에 존재하는 단어만)
    words = [w for w in vocab if w in model.wv.key_to_index]
    if len(words) < 5:
        raise RuntimeError("임베딩이 생성된 단어가 너무 적습니다.")
    X = [model.wv[w] for w in words]

    # 4) k 선택 (실루엣 스코어로 자동)
    if N_CLUSTERS is None:
        k_min, k_max = K_SEARCH_RANGE
        k_min = max(2, min(k_min, len(words)-1))
        k_max = max(k_min, min(k_max, len(words)-1))
        best_k, best_score = None, -1
        print(f"[INFO] k 자동 탐색: {k_min}~{k_max}")
        for k in range(k_min, k_max+1):
            km = KMeans(n_clusters=k, random_state=42, n_init=10)
            labels = km.fit_predict(X)
            # 모든 클러스터에 최소 2개 포인트 필요 → 실패 시 continue
            try:
                score = silhouette_score(X, labels, metric="cosine")
            except Exception:
                continue
            print(f"  - k={k}: silhouette(cosine)={score:.4f}")
            if score > best_score:
                best_k, best_score = k, score
        K = best_k if best_k is not None else max(2, min(8, len(words)-1))
        print(f"[INFO] 선택된 k={K} (silhouette={best_score:.4f} 근사)")
    else:
        K = min(max(2, N_CLUSTERS), len(words)-1)
        print(f"[INFO] 고정 k={K}")

    # 5) 최종 KMeans
    kmeans = KMeans(n_clusters=K, random_state=42, n_init=25)
    labels = kmeans.fit_predict(X)

    # 6) 결과 구성: 단어→클러스터
    word_info = []
    for w, lab in zip(words, labels):
        word_info.append((w, int(token_freq[w]), int(lab)))
    word_df = pd.DataFrame(word_info, columns=["keyword", "freq", "cluster_id"])\
               .sort_values(["cluster_id", "freq"], ascending=[True, False])\
               .reset_index(drop=True)

    # 7) 군집 요약(상위 키워드)
    per_cluster = defaultdict(list)
    for w, f, cid in word_info:
        per_cluster[cid].append((w, f))
    cluster_rows = []
    for cid, items in sorted(per_cluster.items(), key=lambda x: x[0]):
        items.sort(key=lambda x: x[1], reverse=True)
        top_keywords = [w for w, _ in items[:15]]
        size = len(items)
        mean_freq = sum(f for _, f in items)/size if size > 0 else 0
        cluster_rows.append({
            "cluster_id": cid,
            "size": size,
            "mean_freq": round(mean_freq, 2),
            "top_keywords": " ".join(top_keywords)
        })
    summary_df = pd.DataFrame(cluster_rows)\
                  .sort_values("cluster_id").reset_index(drop=True)

    # 8) 군집별 대표 리뷰 샘플(각 군집 최대 5개)
    #    - 리뷰 토큰과 군집 단어의 교집합이 있는 리뷰를 수집
    cluster_words = {cid: set(w for w, _ in per_cluster[cid]) for cid in per_cluster}
    sample_rows = []
    for cid, cw in cluster_words.items():
        collected = 0
        for idx, toks in enumerate(token_seqs):
            hit = [t for t in toks if t in cw]
            if hit:
                sample_rows.append({
                    "cluster_id": cid,
                    "review_id": idx,
                    "matched_keywords": " ".join(sorted(set(hit), key=hit.index)[:8]),
                    "review_text": texts[idx][:800]  # 너무 길면 자름
                })
                collected += 1
                if collected >= 5:
                    break
    samples_df = pd.DataFrame(sample_rows).sort_values(["cluster_id","review_id"]).reset_index(drop=True)

    # 9) 저장
    stamp = datetime.now().strftime("%Y%m%d_%H%M")
    base = os.path.splitext(os.path.basename(input_path))[0]
    out1 = os.path.join(DIR_PATH, f"{base}_word_clusters_k{K}_{stamp}.csv")
    out2 = os.path.join(DIR_PATH, f"{base}_cluster_summary_k{K}_{stamp}.csv")
    out3 = os.path.join(DIR_PATH, f"{base}_cluster_sample_reviews_k{K}_{stamp}.csv")

    word_df.to_csv(out1, index=False, encoding="utf-8-sig")
    summary_df.to_csv(out2, index=False, encoding="utf-8-sig")
    samples_df.to_csv(out3, index=False, encoding="utf-8-sig")

    # 콘솔 미리보기
    print(f"\n[DONE] 저장:\n - {out1}\n - {out2}\n - {out3}")
    print("\n[PREVIEW] 단어 상위(빈도) TOP 100")
    print(word_df.sort_values("freq", ascending=False).head(TOP_N_PREVIEW))

if __name__ == "__main__":
    main()
```
