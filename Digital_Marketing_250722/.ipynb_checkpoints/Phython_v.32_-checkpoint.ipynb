```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from bs4 import BeautifulSoup
import time 
from openpyxl import Workbook
from konlpy.tag import Okt
# (konlpy) korean natural language python > 한국어라는 자연어 확인 후 특정 연산처리 결과를 파이썬으로 출력해주는 라이브러리
# (Okt) Open Korean Text => 구체적으로 한국어의 형태소를 분석 
from collections import Counter

options = Options()
options.add_argument("--no-sandbox")
# 구굴 보안 정책 이름 'sandbox' 적용 X
options.add_argument("--disable-dev-shm-usage")
# 구글 브라우저 가상 메모리 한계 5MB -> 가상 메모리 사용 끄기
options.add_argument("--headless")

driver = webdriver.Chrome(options=options)

url = "https://www.kurly.com/goods/1000078028"
driver.get(url)
time.sleep(5)

soup = BeautifulSoup(driver.page_source, "html.parser")
reviews = soup.select("p.css-y49dcn.e36z05c13")
review_lists = [review.get_text().strip().replace("\n", "") for review in reviews]

driver.quit()

wb = Workbook()
ws = wb.active
ws.title = "Kurly Reviews"
ws.append(["리뷰내용"])

for item in review_lists :
    ws.append([item])

okt = Okt()
# join 함수, list 안의 값 str 치환
all_text = " ".join(review_lists)
nouns = okt.nouns(all_text)
# okt 중 메서드 함수 nouns() => 명사 추출
filltered_nouns = [n for n in nouns if len(n)> 1]
count = Counter(filltered_nouns)

ws_keywords = wb.create_sheet("Keyword Frequency")
ws_keywords.append(["키워드", "빈도수"])
for word, freq in count.most_common(30) :
    ws_keywords.append([word, freq])

wb.save("kurly_riviews.xlsx")
print("엑셀 저장 완료 : kurly_riviews.xlsx😊👌")
```

```python
from openpyxl import load_workbook
from konlpy.tag import Okt
from collections import Counter

wb = load_workbook("연관키워드 20250722 1449.xlsx")
ws = wb.active
# value_only => 값만 가져옴 / if row[0] => True 값 반환 조건
texts = [row[0] for row in ws.iter_rows(min_row= 2, max_col= 1, values_only= True) if row[0]]
okt = Okt()
all_text = " ".join(texts)
nouns = okt.nouns(all_text)
filltered_nouns = [n for n in nouns if len(n) > 1]
count = Counter(filltered_nouns)

ws_keywords = wb.create_sheet("키워드 전체분석")
ws_keywords.append(["명사", "빈도수"])
for word, freq in sorted(count.items(), key=lambda x : x [1], reverse = True) :
    ws_keywords.append([word, freq])
wb.save("키워드 분석.xlsx")
print("저장완료")
```

```python
from openpyxl import load_workbook
from konlpy.tag import Okt
from collections import Counter
import re

wb = load_workbook("연관키워드 20250722 1449.xlsx")
ws = wb.active
# value_only => 값만 가져옴 / if row[0] => True 값 반환 조건
texts = [row[0] for row in ws.iter_rows(min_row= 2, max_col= 1, values_only= True) if row[0]]
okt = Okt()
all_text = " ".join(texts)

nouns_tokens = okt.nouns(all_text)
regex_tokens = re.findall(r'[A-Za-z]*\d+[A-Za-z]+|\d+[A-Za-z]' ,all_text)
# r'' 형식지정자 X ex) \d, \n
valid_one_letter = {"금", "은", "동"}
stopwords = {"사용", "정도", "조금", "때론", "생각", "만족", "다음"}

combinded_tokens = nouns_tokens + regex_tokens
filltered_tokens = [t for t in combinded_tokens if (len(t) > 1 or t in valid_one_letter) and t not in stopwords]

count = Counter(filltered_tokens)

ws_keywords = wb.create_sheet("키워드 전체분석")
ws_keywords.append(["명사", "빈도수"])
for word, freq in sorted(count.items(), key=lambda x : x [1], reverse = True) :
    ws_keywords.append([word, freq])
wb.save("키워드 분석_02!.xlsx")
print("저장완료")
```
