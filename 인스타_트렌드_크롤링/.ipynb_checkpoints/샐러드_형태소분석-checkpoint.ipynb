```python
# -*- coding: utf-8 -*-
# Instagram 캡션/해시태그 형태소 분석 (캡션/해시태그/둘다)
# - 입력: instagram_captions_샐러드_600.csv (현재 작업 폴더)
# - 일본어(히라가나/가타카나) 완전 제외 + 요청한 불용어 세트(BASE/EN/EXTRA) 적용
# - Okt 품사 필터 확장(Noun, Adjective, Verb, Alpha, Foreign)
# - 출력: 정제본/토큰 CSV + 캡션/해시태그/전체 빈도 CSV(utf-8-sig)

import os, re, ast, json, warnings
from datetime import datetime
from collections import Counter
import pandas as pd

# ===================== 설정 =====================
INPUT_PATH  = "instagram_captions_샐러드_600.csv"
OUTPUT_DIR  = "data"
MIN_COUNT   = 2
STAMP       = datetime.now().strftime("%Y%m%d_%H%M%S")
os.makedirs(OUTPUT_DIR, exist_ok=True)

# ===================== 안전 로더 =====================
def read_csv_safely(path):
    encodings = ["utf-8-sig", "utf-8", "cp949", "euc-kr"]
    last_err = None
    for enc in encodings:
        try:
            return pd.read_csv(path, encoding=enc)
        except Exception as e:
            last_err = e
    raise last_err

df = read_csv_safely(INPUT_PATH)

# ===================== 컬럼 자동 탐색 =====================
cap_candidates = ["caption", "text", "content", "contents", "내용", "캡션", "post", "body", "설명"]
tag_candidates = ["hashtags", "tags", "hash_tags", "해시태그", "태그"]

def find_col(cands):
    for c in df.columns:
        lc = str(c).strip().lower()
        if lc in [x.lower() for x in cands]:
            return c
    for c in df.columns:
        lc = str(c).strip().lower()
        if any(x.lower() in lc for x in cands):
            return c
    return None

CAP_COL = find_col(cap_candidates)
TAG_COL = find_col(tag_candidates)

if CAP_COL is None and TAG_COL is None:
    raise ValueError("캡션/해시태그 후보 컬럼을 찾을 수 없습니다. 컬럼명을 확인하세요.")

print({"rows": len(df), "columns": list(df.columns), "CAP_COL": CAP_COL, "TAG_COL": TAG_COL,
       "nonempty_caption": int(df[CAP_COL].notna().sum()) if CAP_COL else 0,
       "nonempty_hashtags": int(df[TAG_COL].notna().sum()) if TAG_COL else 0})

# ===================== 정규식 & 전처리 유틸 =====================
url_re      = re.compile(r"(https?://\S+|www\.\S+)")
mention_re  = re.compile(r"@[A-Za-z0-9_\.]+")
email_re    = re.compile(r"[A-Za-z0-9\._%+\-]+@[A-Za-z0-9\.\-]+\.[A-Za-z]{2,}")
num_re      = re.compile(r"\d+")
emoji_re    = re.compile(
    "[" 
    "\U0001F300-\U0001F64F"
    "\U0001F680-\U0001F6FF"
    "\U0001F900-\U0001F9FF"
    "\U0001FA70-\U0001FAFF"
    "\U00002702-\U000027B0"
    "\U000024C2-\U0001F251"
    "]+", flags=re.UNICODE
)
# 일본어(히라가나/가타카나/halfwidth katakana) → 완전 제외
jp_re       = re.compile(r"[\u3040-\u309F\u30A0-\u30FF\uFF65-\uFF9F]")

# 허용 문자(한글/영문/숫자/공백/#) 외 제거
keep_chars_re = re.compile(r"[^0-9A-Za-z\u3131-\u318E\uAC00-\uD7A3\s#]+")

def basic_clean(s: str) -> str:
    if not isinstance(s, str):
        s = "" if pd.isna(s) else str(s)
    s = url_re.sub(" ", s)
    s = email_re.sub(" ", s)
    s = mention_re.sub(" ", s)
    s = emoji_re.sub(" ", s)
    s = keep_chars_re.sub(" ", s)
    s = re.sub(r"\s+", " ", s).strip()
    return s

# ===================== 요청한 불용어 세트 =====================
BASE_STOP = set("""
그리고 그러나 그런데 또한 또는 그래서 때문에 등의 즉 및 으로 로 은 는 이 가 을 를 과 와 하고 보다 에서 에게 에 에도 에는 에다가 으니까 면 도 만 까지 뿐 처럼 같은 듯 듯이 것 거 데 수 들 등 더 가장 제일 아주 매우 너무 정말 진짜 그냥 혹시 거의 대부분 여러 각각 모든 아무 이런 그런 저런 어떤 무슨 있다 없다 이다 아니다 하다 되다 같다
제품 상품 구성 구입 구매 배송 포장 가격 행사 세트 옵션 용량 맛 향 느낌 사용 효과 후기 리뷰 평가 별점 평점 추천 만족 불만 개선 재구매 성분 브랜드
수량 개 수 박스 병 캡슐 정 분 알 가루 분말 ml mg g kg 개입 세일 이벤트 증정 사은품 먹다 좋다 선물 주문 챙기다 맛있다 건강 젤리 편하다 멀티 편하
꾸준하다 하루 들다 사다 알약 섭취 받다 요즘 복 자주 보다 같다 이다 않다 재다 괜찮다 가장 제일 최고 최애
오늘 내일 지금 문의 링크 프로필 DM 디엠 쿠폰 코드 할인 무료 증정 한정 리그램 리포스트 공유 참여 당첨 축하 체험단 협찬 광고 스폰서 서포터즈 리뷰단 포스팅 제휴 이벤트중 마감 임박 판매중 재고 품절 공구 공동구매 라이브 라이브중 라이브커머스
컬리 마켓컬리 kurly innerkurly kurlypick 이너컬리 컬리픽 스토어 스마트스토어 쿠팡 네이버쇼핑 스토어찜 톡톡 문의 톡톡친구 스샷 스토리 리일즈 릴스 reels 쇼츠 shorts 유튜브 인스타그램 인스타그램샵 instagram shop link bio 링크인바이오 탭 링크트리
맛집 외식 배달 포장 예약 주문 결제 카드 현금 적립 포인트 리뷰이벤트 후기이벤트
샐러드 도시락 한끼 밀키트 다이어트 건강식 저당 고단백 단백질 칼로리 영양 영양정보 식단 메뉴 점심 저녁 간식 야식
""".split())

EN_STOP = set("""
the a an and or but if while for of to in on at from with by as is are was were be been being than too very really just almost nearly most more less many much every any some none only
shop buy sale event free limited coupon code link bio click dm collab sponsor ad spon
salad lunchbox lunch boxes diet healthy wellness meal kit protein kcal cal nutrition menu
""".split())

EXTRA_STOP = set("""
문의주세요 문의환영 프로필링크 프로필링크클릭 프로필링크확인 링크확인 자세한내용 자세한건문의 자세한건디엠 디엠문의 댓글문의 택배배송 새벽배송 당일배송 오늘출발 빠른배송
정기구독 구독 첫구독 구독이벤트 신규가입 가입혜택 할인쿠폰 쿠폰증정
official offical ofcl ofc store shop now new best hot hit
""".split())

STOPWORDS = {w.strip().lower() for w in (BASE_STOP | EN_STOP | EXTRA_STOP) if w.strip()}

# 한 글자/숫자-only/일본어/불용어 제외
def is_valid_token(t: str) -> bool:
    if not t: return False
    if len(t) < 2: return False
    if jp_re.search(t): return False
    if t.lower() in STOPWORDS: return False
    if all(ch.isdigit() for ch in t): return False
    return True

# ===================== 형태소 분석기 (Okt 우선, 실패 시 단순 토크나이저) =====================
use_okt = False
try:
    from konlpy.tag import Okt
    okt = Okt()
    use_okt = True
except Exception as e:
    warnings.warn(f"konlpy/Okt 불러오기 실패: {e}\n간단 토크나이저로 대체합니다(정확도 낮음).")

# (중요) 영어도 살리기 위해 품사에 Alpha/Foreign 포함
ALLOWED_POS = {"Noun", "Adjective", "Verb", "Alpha", "Foreign"}

hangul_seq_re = re.compile(r"[A-Za-z\u3131-\u318E\uAC00-\uD7A3]+")

def tokenize_ko(text: str):
    text = str(text)
    if use_okt:
        pairs = okt.pos(text, norm=True, stem=True)
        toks = []
        for w, p in pairs:
            if p in ALLOWED_POS:
                w = w.strip().lower()
                if is_valid_token(w):
                    toks.append(w)
        return toks
    else:
        cand = hangul_seq_re.findall(text)
        toks = []
        for w in cand:
            w = w.strip().lower()
            if is_valid_token(w):
                toks.append(w)
        return toks

# ===================== 해시태그 파서(강화) =====================
hashtag_token_re = re.compile(r"#([0-9A-Za-z_\u3131-\u318E\uAC00-\uD7A3]+)")

def parse_hashtags(raw, fallback_text=""):
    tags = []
    if isinstance(raw, list):
        tags = raw
    elif isinstance(raw, str):
        s = raw.strip()
        # 리스트 문자열일 수 있음
        if (s.startswith("[") and s.endswith("]")) or (s.startswith("(") and s.endswith(")")):
            try:
                val = ast.literal_eval(s)
                if isinstance(val, (list, tuple)):
                    tags = list(val)
            except Exception:
                pass
        if not tags and s:
            # 쉼표/공백/세미콜론/슬래시/파이프/중점 등 다중 구분자
            parts = re.split(r"[,\s;/|·•\u3001\uFF0C]+", s)
            tags = [p.lstrip("#").strip() for p in parts if p.strip()]
    if not tags and fallback_text:
        tags = hashtag_token_re.findall(str(fallback_text))

    cleaned = []
    for t in tags:
        t = basic_clean(t)
        if not t or jp_re.search(t):   # 일본어 제외
            continue
        cleaned.extend(tokenize_ko(t))
    return cleaned

# ===================== 캡션 정제 & 토큰화 =====================
def clean_and_tokenize_caption(s: str):
    s = basic_clean(s)
    s_wo_hash = re.sub(r"#\S+", " ", s)   # 캡션 텍스트 분석에서는 해시태그 제거
    s_wo_hash = num_re.sub(" ", s_wo_hash)
    s_wo_hash = re.sub(r"\s+", " ", s_wo_hash).strip()
    toks = tokenize_ko(s_wo_hash)
    return s, toks

# ===================== 메인 처리 =====================
cap_texts, cap_tokens_list, tag_tokens_list = [], [], []

for _, row in df.iterrows():
    cap_raw = row[CAP_COL] if CAP_COL in df.columns else ""
    tag_raw = row[TAG_COL] if TAG_COL in df.columns else ""

    cap_cleaned, cap_toks = clean_and_tokenize_caption(cap_raw)
    tag_toks = parse_hashtags(tag_raw, fallback_text=cap_raw)

    cap_texts.append(cap_cleaned)
    cap_tokens_list.append(cap_toks)
    tag_tokens_list.append(tag_toks)

both_tokens_list = [ct + tt for ct, tt in zip(cap_tokens_list, tag_tokens_list)]

# ===================== 결과 DF 구성 =====================
df_out = df.copy()
df_out["caption_clean"]    = cap_texts
df_out["tokens_caption"]   = [json.dumps(toks, ensure_ascii=False) for toks in cap_tokens_list]
df_out["tokens_hashtags"]  = [json.dumps(toks, ensure_ascii=False) for toks in tag_tokens_list]
df_out["tokens_all"]       = [json.dumps(toks, ensure_ascii=False) for toks in both_tokens_list]

# ===================== 빈도 집계 =====================
def freq_df(token_lists, min_count=2):
    cnt = Counter()
    for toks in token_lists:
        cnt.update(toks)
    items = [(t, c) for t, c in cnt.items() if c >= min_count]
    items.sort(key=lambda x: (-x[1], x[0]))
    return pd.DataFrame(items, columns=["token", "freq"])

cap_freq  = freq_df(cap_tokens_list, min_count=MIN_COUNT)
tag_freq  = freq_df(tag_tokens_list, min_count=MIN_COUNT)
all_freq  = freq_df(both_tokens_list, min_count=MIN_COUNT)

# ===================== 저장 =====================
cleaned_path = os.path.join(OUTPUT_DIR, f"cleaned_instagram_captions_{STAMP}.csv")
cap_freq_path = os.path.join(OUTPUT_DIR, f"tokens_caption_freq_{STAMP}.csv")
tag_freq_path = os.path.join(OUTPUT_DIR, f"tokens_hashtag_freq_{STAMP}.csv")
all_freq_path = os.path.join(OUTPUT_DIR, f"tokens_all_freq_{STAMP}.csv")

df_out.to_csv(cleaned_path, index=False, encoding="utf-8-sig")
cap_freq.to_csv(cap_freq_path, index=False, encoding="utf-8-sig")
tag_freq.to_csv(tag_freq_path, index=False, encoding="utf-8-sig")
all_freq.to_csv(all_freq_path, index=False, encoding="utf-8-sig")

print("=== 저장 완료 ===")
print("정제 본문 & 토큰:", cleaned_path)
print("캡션 토큰 빈도:", cap_freq_path)
print("해시태그 토큰 빈도:", tag_freq_path)
print("전체(캡션+해시태그) 토큰 빈도:", all_freq_path)
print({"caption_docs_with_tokens": sum(1 for t in cap_tokens_list if t),
       "hashtag_docs_with_tokens": sum(1 for t in tag_tokens_list if t),
       "both_docs_with_tokens": sum(1 for t in both_tokens_list if t)})

# 상위 30 미리보기
print("\n[캡션 Top 30]\n", cap_freq.head(30))
print("\n[해시태그 Top 30]\n", tag_freq.head(30))
print("\n[전체(캡션+해시태그) Top 30]\n", all_freq.head(30))
```

```python
# -*- coding: utf-8 -*-
# Instagram 캡션/해시태그 형태소 분석 (한글만)
# - 영어/일본어 제거 → 한글만 분석
# - 안전한 이모지 정규식 (한글 영역 침범 X)
# - 캡션/해시태그/합본 빈도 저장

import os, re, ast, json, warnings
from datetime import datetime
from collections import Counter
import pandas as pd

# ===== 설정 =====
INPUT_PATH  = "instagram_captions_샐러드_600.csv"  # 같은 폴더에 둘 경우
OUTPUT_DIR  = "data"
MIN_COUNT   = 2
STAMP       = datetime.now().strftime("%Y%m%d_%H%M%S")
os.makedirs(OUTPUT_DIR, exist_ok=True)

# ===== 로딩 =====
def read_csv_safely(path):
    for enc in ["utf-8-sig", "utf-8", "cp949", "euc-kr"]:
        try:
            return pd.read_csv(path, encoding=enc)
        except Exception:
            pass
    # 마지막 시도
    return pd.read_csv(path, encoding="utf-8-sig", errors="ignore")

df = read_csv_safely(INPUT_PATH)

# 컬럼 자동 탐색
cap_candidates = ["caption", "text", "content", "contents", "내용", "캡션", "post", "body", "설명"]
tag_candidates = ["hashtags", "tags", "hash_tags", "해시태그", "태그"]
def find_col(cands):
    for c in df.columns:
        if str(c).strip().lower() in [x.lower() for x in cands]:
            return c
    for c in df.columns:
        lc = str(c).strip().lower()
        if any(x.lower() in lc for x in cands):
            return c
    return None
CAP_COL = find_col(cap_candidates) or df.columns[0]
TAG_COL = find_col(tag_candidates)

print({"rows": len(df), "columns": list(df.columns), "CAP_COL": CAP_COL, "TAG_COL": TAG_COL})

# ===== 정규식 유틸 =====
url_re      = re.compile(r"(https?://\S+|www\.\S+)")
mention_re  = re.compile(r"@[A-Za-z0-9_\.]+")
email_re    = re.compile(r"[A-Za-z0-9\._%+\-]+@[A-Za-z0-9\.\-]+\.[A-Za-z]{2,}")
# 안전한 이모지 범위(한글 포함 구간 제거)
emoji_re    = re.compile("[" 
    "\U0001F300-\U0001F5FF"
    "\U0001F600-\U0001F64F"
    "\U0001F680-\U0001F6FF"
    "\U0001F700-\U0001F77F"
    "\U0001F780-\U0001F7FF"
    "\U0001F800-\U0001F8FF"
    "\U0001F900-\U0001F9FF"
    "\U0001FA00-\U0001FA6F"
    "\U0001FA70-\U0001FAFF"
    "\u2600-\u26FF"
    "\u2700-\u27BF"
"]+")
# 일본어 제거
jp_re       = re.compile(r"[\u3040-\u309F\u30A0-\u30FF\uFF65-\uFF9F]")
# 한글만 남기기(영어/숫자/기타 제거, #는 해시태그 파싱 위해 유지)
non_korean_re = re.compile(r"[^ \t\n\r\u3131-\u318E\uAC00-\uD7A3#]")
# 한글 시퀀스(길이>=2)
korean_seq_re = re.compile(r"[\uAC00-\uD7A3\u3131-\u318E]{2,}")

def basic_clean_korean_only(s: str) -> str:
    if not isinstance(s, str):
        s = "" if pd.isna(s) else str(s)
    s = url_re.sub(" ", s)
    s = email_re.sub(" ", s)
    s = mention_re.sub(" ", s)
    s = emoji_re.sub(" ", s)
    s = jp_re.sub(" ", s)               # 일본어 제거
    s = non_korean_re.sub(" ", s)       # 영어/숫자/기타 제거 (한글 + # 유지)
    s = re.sub(r"\s+", " ", s).strip()
    return s

# ===== (가벼운) 한국어 불용어 =====
# 기능어/조사/서술어류 중심으로만 최소화
STOPWORDS = set("""
그리고 그러나 그런데 또한 또는 그래서 때문에 등의 즉 및 으로 로 은 는 이 가 을 를 과 와 하고 보다 에서 에게 에 에도 에는 에다가 으니까 면 도 만 까지 뿐 처럼 같은 듯 듯이 것 거 데 수 들 등 더 가장 제일 아주 매우 너무 정말 진짜 그냥 혹시 거의 대부분 여러 각각 모든 아무 이런 그런 저런 어떤 무슨 있다 없다 이다 아니다 하다 되다 같다
오늘 내일 지금 샐러드 맛있는 있는 좋은 메뉴 오늘도 오늘은 많이 맛있게 매일 먹고 만들어 샐러드맛집 요즘 가능 오늘의 함께
시간 맛집 먹는 있어요 저는 가득 이렇게 바로 넣고 만든 가능합니다 배달 샐러드를
""".split())

def tokenize_ko(text: str):
    toks = [w for w in korean_seq_re.findall(str(text)) if len(w) >= 2]
    return [w for w in toks if w not in STOPWORDS]

def parse_hashtags_korean(raw, fallback_text=""):
    if isinstance(raw, str) and raw.strip():
        s = raw
    elif isinstance(raw, list):
        s = " ".join(map(str, raw))
    else:
        s = str(fallback_text)
    s = basic_clean_korean_only(s)
    return tokenize_ko(s)

def clean_and_tokenize_caption_korean(s: str):
    s = basic_clean_korean_only(s)
    s_wo_hash = re.sub(r"#\S+", " ", s)     # 캡션 분석에서 해시태그 제거
    s_wo_hash = re.sub(r"\s+", " ", s_wo_hash).strip()
    return s, tokenize_ko(s_wo_hash)

# ===== 메인 처리 =====
cap_texts, cap_tokens_list, tag_tokens_list = [], [], []
for _, row in df.iterrows():
    cap_raw = row[CAP_COL] if CAP_COL in df.columns else ""
    tag_raw = row[TAG_COL] if TAG_COL and TAG_COL in df.columns else ""
    cap_cleaned, cap_toks = clean_and_tokenize_caption_korean(cap_raw)
    tag_toks = parse_hashtags_korean(tag_raw, fallback_text=cap_raw)
    cap_texts.append(cap_cleaned); cap_tokens_list.append(cap_toks); tag_tokens_list.append(tag_toks)

both_tokens_list = [ct + tt for ct, tt in zip(cap_tokens_list, tag_tokens_list)]

# ===== 빈도 집계 =====
def freq_df(token_lists, min_count=2):
    cnt = Counter()
    for toks in token_lists: cnt.update(toks)
    items = [(t, c) for t, c in cnt.items() if c >= min_count]
    items.sort(key=lambda x: (-x[1], x[0]))
    return pd.DataFrame(items, columns=["token", "freq"])

cap_freq  = freq_df(cap_tokens_list, min_count=MIN_COUNT)
tag_freq  = freq_df(tag_tokens_list, min_count=MIN_COUNT)
all_freq  = freq_df(both_tokens_list, min_count=MIN_COUNT)

# ===== 저장 =====
cleaned_path = os.path.join(OUTPUT_DIR, f"ko_only_cleaned_{STAMP}.csv")
cap_freq_path = os.path.join(OUTPUT_DIR, f"ko_only_caption_freq_{STAMP}.csv")
tag_freq_path = os.path.join(OUTPUT_DIR, f"ko_only_hashtag_freq_{STAMP}.csv")
all_freq_path = os.path.join(OUTPUT_DIR, f"ko_only_all_freq_{STAMP}.csv")

df_out = df.copy()
df_out["ko_caption_clean"]   = cap_texts
df_out["ko_tokens_caption"]  = [json.dumps(t, ensure_ascii=False) for t in cap_tokens_list]
df_out["ko_tokens_hashtags"] = [json.dumps(t, ensure_ascii=False) for t in tag_tokens_list]
df_out["ko_tokens_all"]      = [json.dumps(t, ensure_ascii=False) for t in both_tokens_list]

df_out.to_csv(cleaned_path, index=False, encoding="utf-8-sig")
cap_freq.to_csv(cap_freq_path, index=False, encoding="utf-8-sig")
tag_freq.to_csv(tag_freq_path, index=False, encoding="utf-8-sig")
all_freq.to_csv(all_freq_path, index=False, encoding="utf-8-sig")

print("=== 저장 완료 ===")
print("정제:", cleaned_path)
print("캡션:", cap_freq_path)
print("해시태그:", tag_freq_path)
print("합본:", all_freq_path)
print({"caption_docs_with_tokens": sum(1 for t in cap_tokens_list if t),
       "hashtag_docs_with_tokens": sum(1 for t in tag_tokens_list if t),
       "both_docs_with_tokens": sum(1 for t in both_tokens_list if t)})

# 미리보기
print("\n[캡션 Top 30]\n", cap_freq.head(30))
print("\n[해시태그 Top 30]\n", tag_freq.head(30))
print("\n[합본 Top 30]\n", all_freq.head(30))
```
