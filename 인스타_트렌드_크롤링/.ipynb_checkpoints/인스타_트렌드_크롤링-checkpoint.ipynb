```python
# -*- coding: utf-8 -*-
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.common.by import By
from selenium.webdriver.common.action_chains import ActionChains
from selenium.webdriver.common.keys import Keys
from selenium.common.exceptions import TimeoutException, WebDriverException
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

import pickle, time, random, re, csv
from urllib.parse import quote

# ==================== 설정 ====================
KEYWORD           = "여성영양제"
TARGET_CAPTIONS   = 1000       # 실전 수집량
MAX_SCROLLS       = 1500
VISIT_WAIT        = (0.5, 0.9)
SCROLL_SHORT      = (0.08, 0.18)
SCROLL_PAUSE      = (0.25, 0.55)
CSV_PATH          = "instagram_captions.csv"
INSTA_COOKIE_PKL  = "instagram_cookies.pkl"

FAST_HEADLESS     = False
BLOCK_MEDIA       = True
# =================================================

def make_driver():
    options = Options()
    options.page_load_strategy = "eager"
    options.add_argument("--start-maximized")
    options.add_argument("window-size=1920x1080")
    options.add_argument("lang=ko_KR")
    options.add_argument("--disable-blink-features=AutomationControlled")
    options.add_argument("--disable-infobars")
    options.add_argument("--no-sandbox")
    options.add_argument("--disable-dev-shm-usage")
    if FAST_HEADLESS:
        options.add_argument("--headless=new")
    if BLOCK_MEDIA:
        prefs = {
            "profile.managed_default_content_settings.images": 2,
            "profile.managed_default_content_settings.fonts": 2,
            "profile.managed_default_content_settings.video": 2,
            "profile.managed_default_content_settings.sound": 2,
            "intl.accept_languages": "ko_KR",
        }
        options.add_experimental_option("prefs", prefs)
        options.add_argument("--blink-settings=imagesEnabled=false")
    service = Service(ChromeDriverManager().install())
    drv = webdriver.Chrome(service=service, options=options)
    drv.set_page_load_timeout(50)
    return drv

driver = make_driver()
wait = WebDriverWait(driver, 10)
actions = ActionChains(driver)

# ---------------- 로그인/쿠키 ----------------
driver.get("https://www.instagram.com/")
time.sleep(1.0)
try:
    with open(INSTA_COOKIE_PKL, "rb") as f:
        cookies = pickle.load(f)
    for c in cookies:
        if "expiry" in c and isinstance(c["expiry"], float):
            c["expiry"] = int(c["expiry"])
        try:
            driver.add_cookie(c)
        except Exception:
            pass
    driver.refresh()
    time.sleep(1.0)
except FileNotFoundError:
    print("[경고] instagram_cookies.pkl 없음 → 비로그인일 수 있음")

for txt in ["필수 쿠키만 허용", "쿠키 허용", "Only allow essential cookies", "Allow all cookies",
            "나중에 하기", "Not now", "지금 아니오", "나중에"]:
    try:
        btn = driver.find_element(By.XPATH, f"//button[normalize-space()='{txt}']")
        driver.execute_script("arguments[0].click();", btn)
        time.sleep(0.2)
    except:
        pass

# ---------------- 태그 페이지 ----------------
def goto_hashtag(tag):
    enc = quote(tag)
    for url in [f"https://www.instagram.com/explore/tags/{enc}/",
                f"https://www.instagram.com/explore/tags/{tag}/"]:
        driver.get(url)
        try:
            wait.until(EC.presence_of_element_located((By.TAG_NAME, "a")))
            return
        except TimeoutException:
            continue
goto_hashtag(KEYWORD)

def human_scroll_cycle():
    actions.send_keys(Keys.END).perform()
    time.sleep(random.uniform(*SCROLL_PAUSE))
    for _ in range(1):
        dy = random.randint(350, 900)
        driver.execute_script("window.scrollBy(0, arguments[0]);", dy)
        time.sleep(random.uniform(*SCROLL_SHORT))

# ---------------- JS 파싱(모달/페이지 공통) ----------------
JS_EXTRACT = r"""
function extractFrom(doc) {
  function txt(el){ return el ? (el.textContent || '').trim() : ''; }
  const dialog = doc.querySelector('div[role="dialog"]');
  const art = (dialog && dialog.querySelector('article')) || doc.querySelector('article');
  if(!art) return {caption:'', hashtags:[], postPath:'', canonical:'', og:''};

  // 더보기
  const more = Array.from(art.querySelectorAll('span,button')).find(e=>{
    const t=(e.textContent||'').trim();
    return ['더 보기','더보기','More','more','… 더보기','… more'].includes(t);
  });
  if(more){ more.click(); }

  // 첫 li 블록
  const li1 = art.querySelector('ul li:first-child');
  const capNode = (li1 && (li1.querySelector('h1[dir="auto"]') || li1.querySelector('span[dir="auto"]')))
                  || art.querySelector('h1[dir="auto"]');
  let caption = txt(capNode);
  if(!caption && li1) caption = txt(li1);

  // permalink 추정
  let postPath = '';
  const timeA = art.querySelector('a[href^="/p/"] time, a[href^="/reel/"] time');
  if(timeA && timeA.parentElement) postPath = timeA.parentElement.getAttribute('href') || '';
  const canonical = (doc.querySelector('link[rel="canonical"]')||{}).href || '';
  const og = (doc.querySelector('meta[property="og:url"]')||{}).content || '';

  // 해시태그
  const seen = new Set(); const tags = [];
  if(li1){
    li1.querySelectorAll('a[href^="/explore/tags/"]').forEach(a=>{
      let raw = (a.textContent||'').trim().replace(/^#\s*/,'')
                 .replace(/[\u200b-\u200f\u202a-\u202e\u2060\s]+/g,'');
      if(raw && !seen.has(raw)){ seen.add(raw); tags.push(raw); }
    });
  }
  if(!tags.length && caption){
    (caption.match(/#([0-9A-Za-z_가-힣_]+)/g)||[]).forEach(s=>{
      const t = s.slice(1);
      if(t && !seen.has(t)){ seen.add(t); tags.push(t); }
    });
  }
  return {caption, hashtags:tags, postPath, canonical, og};
}
return extractFrom(document);
"""

def extract_current():
    d = driver.execute_script(JS_EXTRACT)
    cap = (d.get("caption") or "").strip()
    tags = d.get("hashtags") or []
    url = ""
    if d.get("postPath") and str(d["postPath"]).startswith("/"):
        url = "https://www.instagram.com" + d["postPath"]
    elif d.get("canonical"):
        url = d["canonical"]
    elif d.get("og"):
        url = d["og"]
    else:
        url = driver.current_url
    return url, cap, tags

def modal_present():
    try:
        return driver.execute_script("return !!document.querySelector('div[role=\"dialog\"] article');")
    except:
        return False

def open_first_modal():
    anchors = driver.find_elements(By.CSS_SELECTOR, "a[href*='/p/'], a[href*='/reel/']")
    for a in anchors:
        try:
            driver.execute_script("arguments[0].click();", a)
            time.sleep(random.uniform(*VISIT_WAIT))
            if modal_present():
                return True
        except:
            pass
    return False

# --- 핵심: “다음 포스트”만 클릭(캐러셀 제외) ---
def click_next_post():
    """
    모달 바깥(ancestor::article 아닌) '다음/Next' 컨트롤만 클릭.
    실패 시 JS로 후보를 찾아 클릭하고, 그래도 안되면 ArrowRight 키 사용.
    """
    # 1) XPath: article 바깥
    xps = [
        "//div[@role='dialog']//*[self::button or self::div[@role='button'] or self::a]"
        "[@aria-label='다음' or @aria-label='Next'][not(ancestor::article)]",
        "//div[@role='dialog']//*[name()='svg' and (@aria-label='다음' or @aria-label='Next')]"
        "/ancestor::*[self::button or self::div[@role='button'] or self::a][1][not(ancestor::article)]",
    ]
    for xp in xps:
        try:
            btn = driver.find_element(By.XPATH, xp)
            driver.execute_script("arguments[0].click();", btn)
            time.sleep(random.uniform(*VISIT_WAIT))
            return True
        except:
            continue

    # 2) JS로 후보 찾아 클릭(역시 article 바깥만)
    js = r"""
    const d=document.querySelector('div[role="dialog"]'); if(!d) return false;
    const art=d.querySelector('article');
    const cands=[...d.querySelectorAll('[aria-label="다음"],[aria-label="Next"]')];
    const target=cands.find(el=>!art || !el.closest('article'));
    if(target){ target.click(); return true; }
    return false;
    """
    try:
        ok = driver.execute_script(js)
        if ok:
            time.sleep(random.uniform(*VISIT_WAIT))
            return True
    except:
        pass

    # 3) 최후: ArrowRight (모달 컨테이너에 포커스)
    try:
        dlg = driver.find_element(By.CSS_SELECTOR, "div[role='dialog']")
        driver.execute_script("arguments[0].focus();", dlg)
        dlg.click()
        actions.send_keys(Keys.ARROW_RIGHT).perform()
        time.sleep(random.uniform(*VISIT_WAIT))
        return True
    except:
        return False

def close_modal_if_open():
    if not modal_present(): return
    xps = [
        "//div[@role='dialog']//button[@aria-label='닫기' or @aria-label='Close']",
        "//div[@role='dialog']//div[@role='button' and (@aria-label='닫기' or @aria-label='Close')]",
    ]
    for xp in xps:
        try:
            btn = driver.find_element(By.XPATH, xp)
            driver.execute_script("arguments[0].click();", btn)
            time.sleep(0.2)
            return
        except:
            pass
    try:
        actions.send_keys(Keys.ESCAPE).perform()
        time.sleep(0.2)
    except:
        pass

# ---------------- 빠른 순회 로직 ----------------
results, visited = [], set()

# 1) 모달 열기(없으면 스크롤하며 시도)
if not open_first_modal():
    tries = 0
    while tries < 8 and not modal_present():
        human_scroll_cycle()
        if open_first_modal(): break
        tries += 1

# 2) 모달에서 다음 "포스트"로만 순회
same_guard = 0
while modal_present() and len(results) < TARGET_CAPTIONS:
    try:
        url, cap, tags = extract_current()
        if url not in visited and (cap or tags):
            visited.add(url)
            results.append({"url": url, "caption": cap, "hashtags": tags})

        prev_url = url
        if not click_next_post():
            # 끝 도달 → 닫고 스크롤 후 재개
            close_modal_if_open()
            human_scroll_cycle()
            if not open_first_modal():
                human_scroll_cycle()
                if not open_first_modal():
                    break
            same_guard = 0
            continue

        # “다음” 눌렀는데 같은 포스트라면(캐러셀만 넘어간 경우) → 다시 시도
        url2, _, _ = extract_current()
        if url2 == prev_url:
            same_guard += 1
            if same_guard >= 2:
                # 캐러셀에 갇힘 → 강제로 닫고 재개
                close_modal_if_open()
                human_scroll_cycle()
                open_first_modal()
                same_guard = 0
            continue
        same_guard = 0

    except WebDriverException:
        close_modal_if_open()
        human_scroll_cycle()
        open_first_modal()

# 3) 백업: 모달이 막히면 링크 방문
if len(results) < TARGET_CAPTIONS:
    close_modal_if_open()
    have = set()
    for _ in range(20):
        anchors = driver.find_elements(By.CSS_SELECTOR, "a[href*='/p/'], a[href*='/reel/']")
        for a in anchors:
            href = a.get_attribute("href")
            if href and ("/p/" in href or "/reel/" in href):
                have.add(href)
        if len(have) >= TARGET_CAPTIONS*1.3: break
        human_scroll_cycle()
    links = [h for h in have if h not in visited]

    for url in links:
        if len(results) >= TARGET_CAPTIONS: break
        try:
            driver.get(url)
            time.sleep(random.uniform(*VISIT_WAIT))
            u, cap, tags = extract_current()
            if u not in visited and (cap or tags):
                visited.add(u)
                results.append({"url": u, "caption": cap, "hashtags": tags})
        except Exception:
            continue

# ---------------- CSV 저장 ----------------
with open(CSV_PATH, "w", newline="", encoding="utf-8-sig") as f:
    w = csv.writer(f)
    w.writerow(["url", "caption", "hashtags"])
    for r in results:
        w.writerow([r["url"], r["caption"], ";".join(r["hashtags"])])

print(f"[완료] 캡션 {len(results)}건 저장 → {CSV_PATH}")
driver.quit()
```

```python
# -*- coding: utf-8 -*-
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.common.by import By
from selenium.webdriver.common.action_chains import ActionChains
from selenium.webdriver.common.keys import Keys
from selenium.common.exceptions import TimeoutException, WebDriverException
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import pickle, time, random, re, csv
from urllib.parse import quote
# ==================== 설정 ====================
KEYWORD           = "도시락"
TARGET_CAPTIONS   = 2         # 실전 수집량
MAX_SCROLLS       = 2
VISIT_WAIT        = (0.5, 0.9)
SCROLL_SHORT      = (0.08, 0.18)
SCROLL_PAUSE      = (0.25, 0.55)
CSV_PATH          = "instagram_captions.csv"
INSTA_COOKIE_PKL  = "instagram_cookies.pkl"
FAST_HEADLESS     = False
BLOCK_MEDIA       = True
# =================================================
def make_driver():
    options = Options()
    options.page_load_strategy = "eager"
    options.add_argument("--start-maximized")
    options.add_argument("window-size=1920x1080")
    options.add_argument("lang=ko_KR")
    options.add_argument("--disable-blink-features=AutomationControlled")
    options.add_argument("--disable-infobars")
    options.add_argument("--no-sandbox")
    options.add_argument("--disable-dev-shm-usage")
    if FAST_HEADLESS:
        options.add_argument("--headless=new")
    if BLOCK_MEDIA:
        prefs = {
            "profile.managed_default_content_settings.images": 2,
            "profile.managed_default_content_settings.fonts": 2,
            "profile.managed_default_content_settings.video": 2,
            "profile.managed_default_content_settings.sound": 2,
            "intl.accept_languages": "ko_KR",
        }
        options.add_experimental_option("prefs", prefs)
        options.add_argument("--blink-settings=imagesEnabled=false")
    service = Service(ChromeDriverManager().install())
    drv = webdriver.Chrome(service=service, options=options)
    drv.set_page_load_timeout(50)
    return drv
driver = make_driver()
wait = WebDriverWait(driver, 10)
actions = ActionChains(driver)
# ---------------- 로그인/쿠키 ----------------
driver.get("https://www.instagram.com/")
time.sleep(1.0)
try:
    with open(INSTA_COOKIE_PKL, "rb") as f:
        cookies = pickle.load(f)
    for c in cookies:
        if "expiry" in c and isinstance(c["expiry"], float):
            c["expiry"] = int(c["expiry"])
        try:
            driver.add_cookie(c)
        except Exception:
            pass
    driver.refresh()
    time.sleep(1.0)
except FileNotFoundError:
    print("[경고] instagram_cookies.pkl 없음 → 비로그인일 수 있음")
for txt in ["필수 쿠키만 허용", "쿠키 허용", "Only allow essential cookies", "Allow all cookies",
            "나중에 하기", "Not now", "지금 아니오", "나중에"]:
    try:
        btn = driver.find_element(By.XPATH, f"//button[normalize-space()='{txt}']")
        driver.execute_script("arguments[0].click();", btn)
        time.sleep(0.2)
    except:
        pass
# ---------------- 태그 페이지 ----------------
def goto_hashtag(tag):
    enc = quote(tag)
    for url in [f"https://www.instagram.com/explore/tags/{enc}/",
                f"https://www.instagram.com/explore/tags/{tag}/"]:
        driver.get(url)
        try:
            wait.until(EC.presence_of_element_located((By.TAG_NAME, "a")))
            return
        except TimeoutException:
            continue
goto_hashtag(KEYWORD)
def human_scroll_cycle():
    actions.send_keys(Keys.END).perform()
    time.sleep(random.uniform(*SCROLL_PAUSE))
    for _ in range(1):
        dy = random.randint(350, 900)
        driver.execute_script("window.scrollBy(0, arguments[0]);", dy)
        time.sleep(random.uniform(*SCROLL_SHORT))
# ---------------- JS 파싱(모달/페이지 공통) ----------------
JS_EXTRACT = r"""
function extractFrom(doc) {
  function txt(el){ return el ? (el.textContent || '').trim() : ''; }
  const dialog = doc.querySelector('div[role="dialog"]');
  const art = (dialog && dialog.querySelector('article')) || doc.querySelector('article');
  if(!art) return {caption:'', hashtags:[], postPath:'', canonical:'', og:''};
  // 더보기
  const more = Array.from(art.querySelectorAll('span,button')).find(e=>{
    const t=(e.textContent||'').trim();
    return ['더 보기','더보기','More','more','… 더보기','… more'].includes(t);
  });
  if(more){ more.click(); }
  // 첫 li 블록
  const li1 = art.querySelector('ul li:first-child');
  const capNode = (li1 && (li1.querySelector('h1[dir="auto"]') || li1.querySelector('span[dir="auto"]')))
                  || art.querySelector('h1[dir="auto"]');
  let caption = txt(capNode);
  if(!caption && li1) caption = txt(li1);
  // permalink 추정
  let postPath = '';
  const timeA = art.querySelector('a[href^="/p/"] time, a[href^="/reel/"] time');
  if(timeA && timeA.parentElement) postPath = timeA.parentElement.getAttribute('href') || '';
  const canonical = (doc.querySelector('link[rel="canonical"]')||{}).href || '';
  const og = (doc.querySelector('meta[property="og:url"]')||{}).content || '';
  // 해시태그
  const seen = new Set(); const tags = [];
  if(li1){
    li1.querySelectorAll('a[href^="/explore/tags/"]').forEach(a=>{
      let raw = (a.textContent||'').trim().replace(/^#\s*/,'')
                 .replace(/[\u200b-\u200f\u202a-\u202e\u2060\s]+/g,'');
      if(raw && !seen.has(raw)){ seen.add(raw); tags.push(raw); }
    });
  }
  if(!tags.length && caption){
    (caption.match(/#([0-9A-Za-z_가-힣_]+)/g)||[]).forEach(s=>{
      const t = s.slice(1);
      if(t && !seen.has(t)){ seen.add(t); tags.push(t); }
    });
  }
  return {caption, hashtags:tags, postPath, canonical, og};
}
return extractFrom(document);
"""
def extract_current():
    d = driver.execute_script(JS_EXTRACT)
    cap = (d.get("caption") or "").strip()
    tags = d.get("hashtags") or []
    url = ""
    if d.get("postPath") and str(d["postPath"]).startswith("/"):
        url = "https://www.instagram.com" + d["postPath"]
    elif d.get("canonical"):
        url = d["canonical"]
    elif d.get("og"):
        url = d["og"]
    else:
        url = driver.current_url
    return url, cap, tags
def modal_present():
    try:
        return driver.execute_script("return !!document.querySelector('div[role=\"dialog\"] article');")
    except:
        return False
def open_first_modal():
    anchors = driver.find_elements(By.CSS_SELECTOR, "a[href*='/p/'], a[href*='/reel/']")
    for a in anchors:
        try:
            driver.execute_script("arguments[0].click();", a)
            time.sleep(random.uniform(*VISIT_WAIT))
            if modal_present():
                return True
        except:
            pass
    return False
# --- 핵심: “다음 포스트”만 클릭(캐러셀 제외) ---
def click_next_post():
    """
    모달 바깥(ancestor::article 아닌) '다음/Next' 컨트롤만 클릭.
    실패 시 JS로 후보를 찾아 클릭하고, 그래도 안되면 ArrowRight 키 사용.
    """
    # 1) XPath: article 바깥
    xps = [
        "//div[@role='dialog']//*[self::button or self::div[@role='button'] or self::a]"
        "[@aria-label='다음' or @aria-label='Next'][not(ancestor::article)]",
        "//div[@role='dialog']//*[name()='svg' and (@aria-label='다음' or @aria-label='Next')]"
        "/ancestor::*[self::button or self::div[@role='button'] or self::a][1][not(ancestor::article)]",
    ]
    for xp in xps:
        try:
            btn = driver.find_element(By.XPATH, xp)
            driver.execute_script("arguments[0].click();", btn)
            time.sleep(random.uniform(*VISIT_WAIT))
            return True
        except:
            continue
    # 2) JS로 후보 찾아 클릭(역시 article 바깥만)
    js = r"""
    const d=document.querySelector('div[role="dialog"]'); if(!d) return false;
    const art=d.querySelector('article');
    const cands=[...d.querySelectorAll('[aria-label="다음"],[aria-label="Next"]')];
    const target=cands.find(el=>!art || !el.closest('article'));
    if(target){ target.click(); return true; }
    return false;
    """
    try:
        ok = driver.execute_script(js)
        if ok:
            time.sleep(random.uniform(*VISIT_WAIT))
            return True
    except:
        pass
    # 3) 최후: ArrowRight (모달 컨테이너에 포커스)
    try:
        dlg = driver.find_element(By.CSS_SELECTOR, "div[role='dialog']")
        driver.execute_script("arguments[0].focus();", dlg)
        dlg.click()
        actions.send_keys(Keys.ARROW_RIGHT).perform()
        time.sleep(random.uniform(*VISIT_WAIT))
        return True
    except:
        return False
def close_modal_if_open():
    if not modal_present(): return
    xps = [
        "//div[@role='dialog']//button[@aria-label='닫기' or @aria-label='Close']",
        "//div[@role='dialog']//div[@role='button' and (@aria-label='닫기' or @aria-label='Close')]",
    ]
    for xp in xps:
        try:
            btn = driver.find_element(By.XPATH, xp)
            driver.execute_script("arguments[0].click();", btn)
            time.sleep(0.2)
            return
        except:
            pass
    try:
        actions.send_keys(Keys.ESCAPE).perform()
        time.sleep(0.2)
    except:
        pass
# ---------------- 빠른 순회 로직 ----------------
results, visited = [], set()
# 1) 모달 열기(없으면 스크롤하며 시도)
if not open_first_modal():
    tries = 0
    while tries < 8 and not modal_present():
        human_scroll_cycle()
        if open_first_modal(): break
        tries += 1
# 2) 모달에서 다음 "포스트"로만 순회
same_guard = 0
while modal_present() and len(results) < TARGET_CAPTIONS:
    try:
        url, cap, tags = extract_current()
        if url not in visited and (cap or tags):
            visited.add(url)
            results.append({"url": url, "caption": cap, "hashtags": tags})
        prev_url = url
        if not click_next_post():
            # 끝 도달 → 닫고 스크롤 후 재개
            close_modal_if_open()
            human_scroll_cycle()
            if not open_first_modal():
                human_scroll_cycle()
                if not open_first_modal():
                    break
            same_guard = 0
            continue
        # “다음” 눌렀는데 같은 포스트라면(캐러셀만 넘어간 경우) → 다시 시도
        url2, _, _ = extract_current()
        if url2 == prev_url:
            same_guard += 1
            if same_guard >= 2:
                # 캐러셀에 갇힘 → 강제로 닫고 재개
                close_modal_if_open()
                human_scroll_cycle()
                open_first_modal()
                same_guard = 0
            continue
        same_guard = 0
    except WebDriverException:
        close_modal_if_open()
        human_scroll_cycle()
        open_first_modal()
# 3) 백업: 모달이 막히면 링크 방문
if len(results) < TARGET_CAPTIONS:
    close_modal_if_open()
    have = set()
    for _ in range(20):
        anchors = driver.find_elements(By.CSS_SELECTOR, "a[href*='/p/'], a[href*='/reel/']")
        for a in anchors:
            href = a.get_attribute("href")
            if href and ("/p/" in href or "/reel/" in href):
                have.add(href)
        if len(have) >= TARGET_CAPTIONS*1.3: break
        human_scroll_cycle()
    links = [h for h in have if h not in visited]
    for url in links:
        if len(results) >= TARGET_CAPTIONS: break
        try:
            driver.get(url)
            time.sleep(random.uniform(*VISIT_WAIT))
            u, cap, tags = extract_current()
            if u not in visited and (cap or tags):
                visited.add(u)
                results.append({"url": u, "caption": cap, "hashtags": tags})
        except Exception:
            continue
# ---------------- CSV 저장 ----------------
with open(CSV_PATH, "w", newline="", encoding="utf-8-sig") as f:
    w = csv.writer(f)
    w.writerow(["url", "caption", "hashtags"])
    for r in results:
        w.writerow([r["url"], r["caption"], ";".join(r["hashtags"])])
print(f"[완료] 캡션 {len(results)}건 저장 → {CSV_PATH}")
driver.quit()






```

```python
# -*- coding: utf-8 -*-
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.common.by import By
from selenium.webdriver.common.action_chains import ActionChains
from selenium.webdriver.common.keys import Keys
from selenium.common.exceptions import TimeoutException, WebDriverException
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

import pickle, time, random, re, csv
from urllib.parse import quote

# ==================== 설정 ====================
KEYWORD           = "여자영양제"
TARGET_CAPTIONS   = 1000      # 수집 목표
MAX_SCROLLS       = 2000        # 태그 피드 스크롤 상한
VISIT_WAIT        = (0.55, 1.1)
SCROLL_SHORT      = (0.08, 0.18)
SCROLL_PAUSE      = (0.25, 0.55)
CSV_PATH          = "instagram_captions.csv"
INSTA_COOKIE_PKL  = "instagram_cookies.pkl"
FAST_HEADLESS     = False
BLOCK_MEDIA       = True       # 텍스트만 크롤링 → 미디어 차단으로 속도 ↑
# =================================================

def make_driver():
    options = Options()
    options.page_load_strategy = "eager"
    options.add_argument("--start-maximized")
    options.add_argument("window-size=1920x1080")
    options.add_argument("lang=ko_KR")
    options.add_argument("--disable-blink-features=AutomationControlled")
    options.add_argument("--disable-infobars")
    options.add_argument("--no-sandbox")
    options.add_argument("--disable-dev-shm-usage")
    if FAST_HEADLESS:
        options.add_argument("--headless=new")
    if BLOCK_MEDIA:
        prefs = {
            "profile.managed_default_content_settings.images": 2,
            "profile.managed_default_content_settings.fonts": 2,
            "profile.managed_default_content_settings.video": 2,
            "profile.managed_default_content_settings.sound": 2,
            "intl.accept_languages": "ko_KR",
        }
        options.add_experimental_option("prefs", prefs)
        options.add_argument("--blink-settings=imagesEnabled=false")
    service = Service(ChromeDriverManager().install())
    drv = webdriver.Chrome(service=service, options=options)
    drv.set_page_load_timeout(50)
    return drv

driver = make_driver()
wait = WebDriverWait(driver, 10)
actions = ActionChains(driver)

# ---------------- 로그인/쿠키 ----------------
driver.get("https://www.instagram.com/")
time.sleep(1.0)
try:
    with open(INSTA_COOKIE_PKL, "rb") as f:
        cookies = pickle.load(f)
    for c in cookies:
        if "expiry" in c and isinstance(c["expiry"], float):
            c["expiry"] = int(c["expiry"])
        try:
            driver.add_cookie(c)
        except Exception:
            pass
    driver.refresh()
    time.sleep(1.0)
except FileNotFoundError:
    print("[경고] instagram_cookies.pkl 없음 → 비로그인일 수 있음")

for txt in ["필수 쿠키만 허용", "쿠키 허용", "Only allow essential cookies", "Allow all cookies",
            "나중에 하기", "Not now", "지금 아니오", "나중에"]:
    try:
        btn = driver.find_element(By.XPATH, f"//button[normalize-space()='{txt}']")
        driver.execute_script("arguments[0].click();", btn)
        time.sleep(0.2)
    except:
        pass

# ---------------- 태그 페이지 ----------------
def goto_hashtag(tag):
    enc = quote(tag)
    for url in [f"https://www.instagram.com/explore/tags/{enc}/",
                f"https://www.instagram.com/explore/tags/{tag}/"]:
        driver.get(url)
        try:
            wait.until(EC.presence_of_element_located((By.TAG_NAME, "a")))
            return
        except TimeoutException:
            continue

goto_hashtag(KEYWORD)

def human_scroll_cycle():
    actions.send_keys(Keys.END).perform()
    time.sleep(random.uniform(*SCROLL_PAUSE))
    for _ in range(1):
        dy = random.randint(350, 900)
        driver.execute_script("window.scrollBy(0, arguments[0]);", dy)
        time.sleep(random.uniform(*SCROLL_SHORT))

# ---------------- JS 파싱(모달/페이지 공통) ----------------
JS_EXTRACT = r"""
function extractFrom(doc) {
  function txt(el){ return el ? (el.textContent || '').trim() : ''; }
  const dialog = doc.querySelector('div[role="dialog"]');
  const art = (dialog && dialog.querySelector('article')) || doc.querySelector('article');
  if(!art) return {caption:'', hashtags:[], postPath:'', canonical:'', og:'', publishedISO:'', publishedText:''};

  // 더보기
  const more = Array.from(art.querySelectorAll('span,button')).find(e=>{
    const t=(e.textContent||'').trim();
    return ['더 보기','더보기','More','more','… 더보기','… more'].includes(t);
  });
  if(more){ more.click(); }

  // 첫 li 블록
  const li1 = art.querySelector('ul li:first-child');
  const capNode = (li1 && (li1.querySelector('h1[dir="auto"]') || li1.querySelector('span[dir="auto"]')))
                  || art.querySelector('h1[dir="auto"]');
  let caption = txt(capNode);
  if(!caption && li1) caption = txt(li1);

  // permalink & 시간
  let postPath = '';
  let t = art.querySelector('a[href^="/p/"] time, a[href^="/reel/"] time') || art.querySelector('time');
  if(t && t.parentElement && t.parentElement.getAttribute('href')) {
    postPath = t.parentElement.getAttribute('href') || '';
  }
  const publishedISO  = (t && t.getAttribute('datetime')) || (doc.querySelector('meta[property="article:published_time"]')||{}).content || '';
  const publishedText = (t && (t.getAttribute('title') || t.textContent || '').trim()) || '';

  const canonical = (doc.querySelector('link[rel="canonical"]')||{}).href || '';
  const og = (doc.querySelector('meta[property="og:url"]')||{}).content || '';

  // 해시태그
  const seen = new Set(); const tags = [];
  if(li1){
    li1.querySelectorAll('a[href^="/explore/tags/"]').forEach(a=>{
      let raw = (a.textContent||'').trim().replace(/^#\s*/,'')
                 .replace(/[\u200b-\u200f\u202a-\u202e\u2060\s]+/g,'');
      if(raw && !seen.has(raw)){ seen.add(raw); tags.push(raw); }
    });
  }
  if(!tags.length && caption){
    (caption.match(/#([0-9A-Za-z_가-힣_]+)/g)||[]).forEach(s=>{
      const tt = s.slice(1);
      if(tt && !seen.has(tt)){ seen.add(tt); tags.push(tt); }
    });
  }
  return {caption, hashtags:tags, postPath, canonical, og, publishedISO, publishedText};
}
return extractFrom(document);
"""

def extract_current():
    """현재 모달/페이지에서 url, caption, tags, 날짜 추출"""
    d = driver.execute_script(JS_EXTRACT)
    cap = (d.get("caption") or "").strip()
    tags = d.get("hashtags") or []
    url = ""
    if d.get("postPath") and str(d["postPath"]).startswith("/"):
        url = "https://www.instagram.com" + d["postPath"]
    elif d.get("canonical"):
        url = d["canonical"]
    elif d.get("og"):
        url = d["og"]
    else:
        url = driver.current_url
    pub_iso  = (d.get("publishedISO") or "").strip()
    pub_text = (d.get("publishedText") or "").strip()
    return url, cap, tags, pub_iso, pub_text

def modal_present():
    try:
        return driver.execute_script("return !!document.querySelector('div[role=\"dialog\"] article');")
    except:
        return False

def open_first_modal():
    """피드의 첫 썸네일을 클릭해서 모달오픈"""
    anchors = driver.find_elements(By.CSS_SELECTOR, "a[href*='/p/'], a[href*='/reel/']")
    for a in anchors:
        try:
            driver.execute_script("arguments[0].click();", a)
            time.sleep(random.uniform(*VISIT_WAIT))
            if modal_present():
                return True
        except:
            pass
    return False

# --- 핵심: "다음 포스트" 전용 클릭 ---
def click_next_post_and_wait(prev_key, timeout=6):
    """
    1) article 바깥 '다음/Next'만 클릭
    2) 클릭 후 prev_key와 다른 포스트가 뜰 때까지 대기
    """
    clicked = False
    try:
        js = r"""
        const d=document.querySelector('div[role="dialog"]'); if(!d) return false;
        const art=d.querySelector('article');
        const cands=[...d.querySelectorAll('[aria-label="다음"],[aria-label="Next"]')];
        const t=cands.find(el=>!art || !el.closest('article'));
        if(t){ t.click(); return true; }
        return false;
        """
        clicked = driver.execute_script(js)
    except Exception:
        clicked = False

    if not clicked:
        try:
            dlg = driver.find_element(By.CSS_SELECTOR, "div[role='dialog']")
            driver.execute_script("arguments[0].focus();", dlg)
            dlg.click()
            actions.send_keys(Keys.ARROW_RIGHT).perform()
            clicked = True
        except:
            clicked = False

    if not clicked:
        return False

    end = time.time() + timeout
    while time.time() < end:
        try:
            url2, _, _, _, _ = extract_current()
            if url2 and url2 != prev_key:
                return True
        except:
            pass
        time.sleep(0.25)
    return False

def close_modal_if_open():
    if not modal_present(): return
    xps = [
        "//div[@role='dialog']//button[@aria-label='닫기' or @aria-label='Close']",
        "//div[@role='dialog']//div[@role='button' and (@aria-label='닫기' or @aria-label='Close')]",
    ]
    for xp in xps:
        try:
            btn = driver.find_element(By.XPATH, xp)
            driver.execute_script("arguments[0].click();", btn)
            time.sleep(0.2)
            return
        except:
            pass
    try:
        actions.send_keys(Keys.ESCAPE).perform()
        time.sleep(0.2)
    except:
        pass

# ---------------- 빠른 순회 로직 ----------------
results, visited = [], set()

# 1) 모달 열기(없으면 스크롤하며 시도)
if not open_first_modal():
    tries = 0
    while tries < 8 and not modal_present():
        human_scroll_cycle()
        if open_first_modal(): break
        tries += 1

# 2) 모달에서 "다음 포스트" 순회
stuck_guard = 0
hop_guard   = 0   # 안전 상한 (무한 루프 방지)

while modal_present() and len(results) < TARGET_CAPTIONS and hop_guard < TARGET_CAPTIONS * 3:
    hop_guard += 1
    try:
        url, cap, tags, pub_iso, pub_text = extract_current()
        if url not in visited and (cap or tags):
            visited.add(url)
            results.append({"url": url, "caption": cap, "hashtags": tags,
                            "date_iso": pub_iso, "date_text": pub_text})
        prev_key = url

        moved = click_next_post_and_wait(prev_key, timeout=6)
        if not moved:
            close_modal_if_open()
            human_scroll_cycle()
            if not open_first_modal():
                human_scroll_cycle()
                if not open_first_modal():
                    break
            stuck_guard = 0
            continue

        url2, _, _, _, _ = extract_current()
        if url2 == prev_key:
            stuck_guard += 1
            if stuck_guard >= 2:
                close_modal_if_open()
                human_scroll_cycle()
                open_first_modal()
                stuck_guard = 0
            continue
        stuck_guard = 0

    except WebDriverException:
        close_modal_if_open()
        human_scroll_cycle()
        open_first_modal()

# 3) 백업: 모달이 막히면 개별 링크 방문
if len(results) < TARGET_CAPTIONS:
    close_modal_if_open()
    have = set()
    feed_scrolls = 0
    while len(have) < TARGET_CAPTIONS*1.3 and feed_scrolls < MAX_SCROLLS:
        anchors = driver.find_elements(By.CSS_SELECTOR, "a[href*='/p/'], a[href*='/reel/']")
        for a in anchors:
            href = a.get_attribute("href")
            if href and ("/p/" in href or "/reel/" in href):
                have.add(href)
        human_scroll_cycle()
        feed_scrolls += 1
    links = [h for h in have if h not in visited]
    for url in links:
        if len(results) >= TARGET_CAPTIONS: break
        try:
            driver.get(url)
            time.sleep(random.uniform(*VISIT_WAIT))
            u, cap, tags, pub_iso, pub_text = extract_current()
            if u not in visited and (cap or tags):
                visited.add(u)
                results.append({"url": u, "caption": cap, "hashtags": tags,
                                "date_iso": pub_iso, "date_text": pub_text})
        except Exception:
            continue

# ---------------- CSV 저장 ----------------
with open(CSV_PATH, "w", newline="", encoding="utf-8-sig") as f:
    w = csv.writer(f)
    # 날짜 2개 컬럼만 **추가** (기존 컬럼 보존)
    w.writerow(["url", "caption", "hashtags", "date_iso", "date_text"])
    for r in results:
        w.writerow([r["url"], r["caption"], ";".join(r["hashtags"]), r.get("date_iso",""), r.get("date_text","")])

print(f"[완료] 캡션 {len(results)}건 저장 → {CSV_PATH}")
driver.quit()
```

```python
# -*- coding: utf-8 -*-
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.common.by import By
from selenium.webdriver.common.action_chains import ActionChains
from selenium.webdriver.common.keys import Keys
from selenium.common.exceptions import TimeoutException, WebDriverException
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

import pickle, time, random, re, csv
from urllib.parse import quote

# ==================== 설정 ====================
KEYWORD           = "여자영양제"
TARGET_CAPTIONS   = 30      # 수집 목표(캡션 있는 포스트 개수)
MAX_SCROLLS       = 30      # 피드 스크롤 상한(총합)
VISIT_WAIT        = (0.55, 1.1)
SCROLL_SHORT      = (0.08, 0.18)
SCROLL_PAUSE      = (0.25, 0.55)
CSV_PATH          = "instagram_captions.csv"
INSTA_COOKIE_PKL  = "instagram_cookies.pkl"
FAST_HEADLESS     = False
BLOCK_MEDIA       = True     # 텍스트 위주면 True 권장
# 중복 루프 탈출 민감도
SAME_POST_RETRIES = 2        # 모달에서 같은 포스트가 이 횟수 연속이면 강제 종료 후 피드로 복귀
FEED_OPEN_TRIES   = 8        # 피드에서 새 포스트 모달 열기 재시도 횟수
# =================================================

def make_driver():
    options = Options()
    options.page_load_strategy = "eager"
    options.add_argument("--start-maximized")
    options.add_argument("window-size=1920x1080")
    options.add_argument("lang=ko_KR")
    options.add_argument("--disable-blink-features=AutomationControlled")
    options.add_argument("--disable-infobars")
    options.add_argument("--no-sandbox")
    options.add_argument("--disable-dev-shm-usage")
    if FAST_HEADLESS:
        options.add_argument("--headless=new")
    if BLOCK_MEDIA:
        prefs = {
            "profile.managed_default_content_settings.images": 2,
            "profile.managed_default_content_settings.fonts": 2,
            "profile.managed_default_content_settings.video": 2,
            "profile.managed_default_content_settings.sound": 2,
            "intl.accept_languages": "ko_KR",
        }
        options.add_experimental_option("prefs", prefs)
        options.add_argument("--blink-settings=imagesEnabled=false")
    service = Service(ChromeDriverManager().install())
    drv = webdriver.Chrome(service=service, options=options)
    drv.set_page_load_timeout(50)
    return drv

driver = make_driver()
wait = WebDriverWait(driver, 10)
actions = ActionChains(driver)

# ---------------- 로그인/쿠키 ----------------
driver.get("https://www.instagram.com/")
time.sleep(1.0)
try:
    with open(INSTA_COOKIE_PKL, "rb") as f:
        cookies = pickle.load(f)
    for c in cookies:
        if "expiry" in c and isinstance(c["expiry"], float):
            c["expiry"] = int(c["expiry"])
        try:
            driver.add_cookie(c)
        except Exception:
            pass
    driver.refresh()
    time.sleep(1.0)
except FileNotFoundError:
    print("[경고] instagram_cookies.pkl 없음 → 비로그인일 수 있음")

for txt in ["필수 쿠키만 허용", "쿠키 허용", "Only allow essential cookies", "Allow all cookies",
            "나중에 하기", "Not now", "지금 아니오", "나중에"]:
    try:
        btn = driver.find_element(By.XPATH, f"//button[normalize-space()='{txt}']")
        driver.execute_script("arguments[0].click();", btn)
        time.sleep(0.2)
    except:
        pass

# ---------------- 태그 페이지 ----------------
def goto_hashtag(tag):
    enc = quote(tag)
    for url in [f"https://www.instagram.com/explore/tags/{enc}/",
                f"https://www.instagram.com/explore/tags/{tag}/"]:
        driver.get(url)
        try:
            wait.until(EC.presence_of_element_located((By.TAG_NAME, "a")))
            return
        except TimeoutException:
            continue
goto_hashtag(KEYWORD)

def human_scroll_once():
    actions.send_keys(Keys.END).perform()
    time.sleep(random.uniform(*SCROLL_PAUSE))
    dy = random.randint(350, 900)
    driver.execute_script("window.scrollBy(0, arguments[0]);", dy)
    time.sleep(random.uniform(*SCROLL_SHORT))

# ---------------- JS 파싱(모달/페이지 공통) ----------------
JS_EXTRACT = r"""
function extractFrom(doc) {
  function txt(el){ return el ? (el.textContent || '').trim() : ''; }
  const dialog = doc.querySelector('div[role="dialog"]');
  const art = (dialog && dialog.querySelector('article')) || doc.querySelector('article');
  if(!art) return {caption:'', hashtags:[], postPath:'', canonical:'', og:'', postKind:'', shortcode:''};

  // 더보기
  const more = Array.from(art.querySelectorAll('span,button')).find(e=>{
    const t=(e.textContent||'').trim();
    return ['더 보기','더보기','More','more','… 더보기','… more'].includes(t);
  });
  if(more){ more.click(); }

  // 첫 li 블록
  const li1 = art.querySelector('ul li:first-child');
  const capNode = (li1 && (li1.querySelector('h1[dir="auto"]') || li1.querySelector('span[dir="auto"]')))
                  || art.querySelector('h1[dir="auto"]');
  let caption = txt(capNode);
  if(!caption && li1) caption = txt(li1);

  // permalink 추정
  let anchor = art.querySelector('a[href^="/p/"], a[href^="/reel/"]');
  let postPath = anchor ? (anchor.getAttribute('href')||'') : '';
  if(!postPath){
    const t = art.querySelector('a[href^="/p/"] time, a[href^="/reel/"] time');
    if(t && t.parentElement) postPath = t.parentElement.getAttribute('href') || '';
  }
  const canonical = (doc.querySelector('link[rel="canonical"]')||{}).href || '';
  const og = (doc.querySelector('meta[property="og:url"]')||{}).content || '';

  // 해시태그
  const seen = new Set(); const tags = [];
  if(li1){
    li1.querySelectorAll('a[href^="/explore/tags/"]').forEach(a=>{
      let raw = (a.textContent||'').trim().replace(/^#\s*/,'')
                 .replace(/[\u200b-\u200f\u202a-\u202e\u2060\s]+/g,'');
      if(raw && !seen.has(raw)){ seen.add(raw); tags.push(raw); }
    });
  }
  if(!tags.length && caption){
    (caption.match(/#([0-9A-Za-z_가-힣_]+)/g)||[]).forEach(s=>{
      const t = s.slice(1);
      if(t && !seen.has(t)){ seen.add(t); tags.push(t); }
    });
  }

  // 고유키(Shortcode)와 타입
  function pickHref() {
    return postPath || canonical || og || (doc.location ? doc.location.href : '');
  }
  const href = pickHref();
  let postKind = ''; let shortcode = '';
  const m = href.match(/\/(p|reel)\/([^\/\?\#]+)/);
  if(m){ postKind = m[1]; shortcode = m[2]; }

  return {caption, hashtags:tags, postPath, canonical, og, postKind, shortcode};
}
return extractFrom(document);
"""

def _normalize_url(u: str) -> str:
    m = re.search(r"https://www\.instagram\.com/(p|reel)/([^/\?#]+)/?", str(u))
    if m:
        return f"https://www.instagram.com/{m.group(1)}/{m.group(2)}/"
    return str(u or "")

def _shortcode_from_url(u: str) -> str:
    m = re.search(r"/(p|reel)/([^/\?#]+)/?", str(u))
    return m.group(2) if m else ""

def extract_current():
    """현재 모달/페이지에서 url, caption, tags, shortcode 추출(iframe 백업 포함 안 해도 IG는 동일 출처)"""
    d = driver.execute_script(JS_EXTRACT)
    cap = (d.get("caption") or "").strip()
    tags = d.get("hashtags") or []
    # URL 우선순위
    if d.get("postPath") and str(d["postPath"]).startswith("/"):
        url = "https://www.instagram.com" + d["postPath"]
    elif d.get("canonical"):
        url = d["canonical"]
    elif d.get("og"):
        url = d["og"]
    else:
        url = driver.current_url
    url = _normalize_url(url)
    code = _shortcode_from_url(url) or (d.get("shortcode") or "")
    return url, cap, tags, code

def modal_present():
    try:
        return driver.execute_script("return !!document.querySelector('div[role=\"dialog\"] article');")
    except:
        return False

def click_next_post():
    """
    모달 바깥(ancestor::article 아닌) '다음/Next' 컨트롤만 클릭.
    실패 시 ArrowRight 키 사용.
    """
    # 1) XPath: article 바깥
    xps = [
        "//div[@role='dialog']//*[self::button or self::div[@role='button'] or self::a]"
        "[@aria-label='다음' or @aria-label='Next'][not(ancestor::article)]",
        "//div[@role='dialog']//*[name()='svg' and (@aria-label='다음' or @aria-label='Next')]"
        "/ancestor::*[self::button or self::div[@role='button'] or self::a][1][not(ancestor::article)]",
    ]
    for xp in xps:
        try:
            btn = driver.find_element(By.XPATH, xp)
            driver.execute_script("arguments[0].click();", btn)
            time.sleep(random.uniform(*VISIT_WAIT))
            return True
        except:
            continue

    # 2) JS 후보 클릭
    js = r"""
    const d=document.querySelector('div[role="dialog"]'); if(!d) return false;
    const art=d.querySelector('article');
    const cands=[...d.querySelectorAll('[aria-label="다음"],[aria-label="Next"]')];
    const target=cands.find(el=>!art || !el.closest('article'));
    if(target){ target.click(); return true; }
    return false;
    """
    try:
        ok = driver.execute_script(js)
        if ok:
            time.sleep(random.uniform(*VISIT_WAIT))
            return True
    except:
        pass

    # 3) ArrowRight
    try:
        dlg = driver.find_element(By.CSS_SELECTOR, "div[role='dialog']")
        driver.execute_script("arguments[0].focus();", dlg)
        dlg.click()
        actions.send_keys(Keys.ARROW_RIGHT).perform()
        time.sleep(random.uniform(*VISIT_WAIT))
        return True
    except:
        return False

def close_modal_if_open():
    if not modal_present(): return
    xps = [
        "//div[@role='dialog']//button[@aria-label='닫기' or @aria-label='Close']",
        "//div[@role='dialog']//div[@role='button' and (@aria-label='닫기' or @aria-label='Close')]",
    ]
    for xp in xps:
        try:
            btn = driver.find_element(By.XPATH, xp)
            driver.execute_script("arguments[0].click();", btn)
            time.sleep(0.2)
            return
        except:
            pass
    try:
        actions.send_keys(Keys.ESCAPE).perform()
        time.sleep(0.2)
    except:
        pass

def open_modal_from_feed(visited_codes, max_feed_scroll=6):
    """
    피드에서 아직 보지 않은 (p|reel) 링크를 클릭해 모달을 연다.
    - visited_codes: 본 적 있는 shortcode 집합
    """
    tries = 0
    while tries < max_feed_scroll:
        anchors = driver.find_elements(By.CSS_SELECTOR, "a[href*='/p/'], a[href*='/reel/']")
        clicked = False
        for a in anchors:
            href = a.get_attribute("href") or ""
            if href.startswith("/"):
                href = "https://www.instagram.com" + href
            code = _shortcode_from_url(href)
            if not code or code in visited_codes:
                continue
            try:
                driver.execute_script("arguments[0].click();", a)
                time.sleep(random.uniform(*VISIT_WAIT))
                if modal_present():
                    return True
            except Exception:
                continue
        # 새 링크 못 찾음 → 스크롤
        human_scroll_once()
        tries += 1
    return False

# ---------------- 메인 루프(모달 방식 + 중복 방지) ----------------
results = []
visited_codes = set()   # 수집 완료(또는 시도)한 포스트의 shortcode
total_feed_scrolls = 0

# 1) 첫 모달 열기
if not open_modal_from_feed(visited_codes, max_feed_scroll=FEED_OPEN_TRIES):
    # 피드에서 충분히 스크롤하며 다시 시도
    while total_feed_scrolls < MAX_SCROLLS and not modal_present():
        human_scroll_once()
        total_feed_scrolls += 1
        if open_modal_from_feed(visited_codes, max_feed_scroll=2):
            break

# 2) 모달 순회
same_guard = 0
while modal_present() and len(results) < TARGET_CAPTIONS and total_feed_scrolls < MAX_SCROLLS:
    try:
        url, cap, tags, code = extract_current()

        # 수집 & 중복체크
        if code and code not in visited_codes and (cap or tags):
            visited_codes.add(code)
            results.append({"url": url, "caption": cap, "hashtags": tags})
            same_guard = 0  # 정상 진행되면 가드 초기화
        else:
            # 중복/빈값이면 가드 증가
            same_guard += 1

        prev_code = code

        # 다음 포스트로 이동
        moved = click_next_post()
        if not moved:
            same_guard = SAME_POST_RETRIES  # 강제 재시작 트리거

        # 이동 후 코드 확인
        url2, _, _, code2 = extract_current()
        if not code2 or code2 == prev_code:
            same_guard += 1
        else:
            same_guard = 0

        # 같은 포스트/빈값 반복 → 모달 종료 후 피드 스크롤, 새 모달 열기
        if same_guard >= SAME_POST_RETRIES:
            close_modal_if_open()
            # 피드에서 새(미방문) 포스트를 찾아 모달 열기
            opened = open_modal_from_feed(visited_codes, max_feed_scroll=FEED_OPEN_TRIES)
            if not opened:
                # 스크롤을 넉넉히 돌며 다시 시도
                round_scrolls = 0
                while round_scrolls < FEED_OPEN_TRIES and len(results) < TARGET_CAPTIONS:
                    human_scroll_once()
                    total_feed_scrolls += 1
                    round_scrolls += 1
                    if open_modal_from_feed(visited_codes, max_feed_scroll=2):
                        same_guard = 0
                        break
                else:
                    # 더 이상 열 수 없으면 종료
                    break
            else:
                same_guard = 0

    except WebDriverException:
        # 경미한 에러 → 재시작
        close_modal_if_open()
        open_modal_from_feed(visited_codes, max_feed_scroll=FEED_OPEN_TRIES)

# ---------------- 백업: 피드에서 링크 방문(모달 불가 시) ----------------
if len(results) < TARGET_CAPTIONS:
    have = set()
    feed_rounds = 0
    while len(have) < TARGET_CAPTIONS*1.3 and feed_rounds < MAX_SCROLLS:
        anchors = driver.find_elements(By.CSS_SELECTOR, "a[href*='/p/'], a[href*='/reel/']")
        for a in anchors:
            href = a.get_attribute("href") or ""
            if href.startswith("/"):
                href = "https://www.instagram.com" + href
            code = _shortcode_from_url(href)
            if code and code not in visited_codes:
                have.add(_normalize_url(href))
        human_scroll_once()
        feed_rounds += 1

    for url in list(have):
        if len(results) >= TARGET_CAPTIONS: break
        try:
            driver.get(url)
            time.sleep(random.uniform(*VISIT_WAIT))
            u, cap, tags, code = extract_current()
            if code and code not in visited_codes and (cap or tags):
                visited_codes.add(code)
                results.append({"url": u, "caption": cap, "hashtags": tags})
        except Exception:
            continue

# ---------------- CSV 저장 ----------------
with open(CSV_PATH, "w", newline="", encoding="utf-8-sig") as f:
    w = csv.writer(f)
    w.writerow(["url", "caption", "hashtags"])
    for r in results:
        w.writerow([r["url"], r["caption"], ";".join(r["hashtags"])])

print(f"[완료] 캡션 {len(results)}건 저장 → {CSV_PATH}")
driver.quit()
```

```python
# -*- coding: utf-8 -*-
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.common.by import By
from selenium.webdriver.common.action_chains import ActionChains
from selenium.webdriver.common.keys import Keys
from selenium.common.exceptions import TimeoutException, WebDriverException, NoSuchElementException
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

import pickle, time, random, re, csv
from urllib.parse import quote

# ==================== 설정 ====================
KEYWORD           = "여자영양제"
TARGET_CAPTIONS   = 1000      # 수집 목표(캡션 있는 포스트 개수)
MAX_SCROLLS       = 2000      # 피드 스크롤 상한(총합)
VISIT_WAIT        = (0.55, 1.1)
SCROLL_SHORT      = (0.08, 0.18)
SCROLL_PAUSE      = (0.25, 0.55)
CSV_PATH          = "instagram_captions.csv"
INSTA_COOKIE_PKL  = "instagram_cookies.pkl"
FAST_HEADLESS     = False
BLOCK_MEDIA       = True     # 텍스트 위주면 True 권장

# 중복/무진행 가드
SAME_POST_RETRIES     = 2    # 모달 내 같은 포스트가 이 횟수 연속이면 모달 닫고 피드로 복귀
FEED_OPEN_TRIES       = 8    # 피드에서 새 모달 열기 재시도 횟수
NO_PROGRESS_MAX_ROUND = 8    # 결과 개수가 늘지 않는 라운드가 이 횟수면 종료
# =================================================

def make_driver():
    options = Options()
    options.page_load_strategy = "eager"
    options.add_argument("--start-maximized")
    options.add_argument("window-size=1920x1080")
    options.add_argument("lang=ko_KR")
    options.add_argument("--disable-blink-features=AutomationControlled")
    options.add_argument("--disable-infobars")
    options.add_argument("--no-sandbox")
    options.add_argument("--disable-dev-shm-usage")
    if FAST_HEADLESS:
        options.add_argument("--headless=new")
    if BLOCK_MEDIA:
        prefs = {
            "profile.managed_default_content_settings.images": 2,
            "profile.managed_default_content_settings.fonts": 2,
            "profile.managed_default_content_settings.video": 2,
            "profile.managed_default_content_settings.sound": 2,
            "intl.accept_languages": "ko_KR",
        }
        options.add_experimental_option("prefs", prefs)
        options.add_argument("--blink-settings=imagesEnabled=false")
    service = Service(ChromeDriverManager().install())
    drv = webdriver.Chrome(service=service, options=options)
    drv.set_page_load_timeout(50)
    return drv

driver = make_driver()
wait = WebDriverWait(driver, 10)
actions = ActionChains(driver)

# ---------------- 로그인/쿠키 ----------------
driver.get("https://www.instagram.com/")
time.sleep(1.0)
try:
    with open(INSTA_COOKIE_PKL, "rb") as f:
        cookies = pickle.load(f)
    for c in cookies:
        if "expiry" in c and isinstance(c["expiry"], float):
            c["expiry"] = int(c["expiry"])
        try:
            driver.add_cookie(c)
        except Exception:
            pass
    driver.refresh()
    time.sleep(1.0)
except FileNotFoundError:
    print("[경고] instagram_cookies.pkl 없음 → 비로그인일 수 있음")

for txt in ["필수 쿠키만 허용", "쿠키 허용", "Only allow essential cookies", "Allow all cookies",
            "나중에 하기", "Not now", "지금 아니오", "나중에"]:
    try:
        btn = driver.find_element(By.XPATH, f"//button[normalize-space()='{txt}']")
        driver.execute_script("arguments[0].click();", btn)
        time.sleep(0.2)
    except:
        pass

# ---------------- 태그 페이지 ----------------
def goto_hashtag(tag):
    enc = quote(tag)
    for url in [f"https://www.instagram.com/explore/tags/{enc}/",
                f"https://www.instagram.com/explore/tags/{tag}/"]:
        driver.get(url)
        try:
            wait.until(EC.presence_of_element_located((By.TAG_NAME, "a")))
            return
        except TimeoutException:
            continue
goto_hashtag(KEYWORD)

def human_scroll_once():
    actions.send_keys(Keys.END).perform()
    time.sleep(random.uniform(*SCROLL_PAUSE))
    dy = random.randint(350, 900)
    driver.execute_script("window.scrollBy(0, arguments[0]);", dy)
    time.sleep(random.uniform(*SCROLL_SHORT))

# ---------------- JS 파싱(모달/페이지 공통) ----------------
JS_EXTRACT = r"""
function extractFrom(doc) {
  function txt(el){ return el ? (el.textContent || '').trim() : ''; }
  const dialog = doc.querySelector('div[role="dialog"]');
  const art = (dialog && dialog.querySelector('article')) || doc.querySelector('article');
  if(!art) return {caption:'', hashtags:[], postPath:'', canonical:'', og:'', postKind:'', shortcode:''};

  // 더보기
  const more = Array.from(art.querySelectorAll('span,button')).find(e=>{
    const t=(e.textContent||'').trim();
    return ['더 보기','더보기','More','more','… 더보기','… more'].includes(t);
  });
  if(more){ more.click(); }

  // 첫 li 블록
  const li1 = art.querySelector('ul li:first-child');
  const capNode = (li1 && (li1.querySelector('h1[dir="auto"]') || li1.querySelector('span[dir="auto"]')))
                  || art.querySelector('h1[dir="auto"]');
  let caption = txt(capNode);
  if(!caption && li1) caption = txt(li1);

  // permalink 후보
  let anchor = art.querySelector('a[href^="/p/"], a[href^="/reel/"]');
  let postPath = anchor ? (anchor.getAttribute('href')||'') : '';
  if(!postPath){
    const t = art.querySelector('a[href^="/p/"] time, a[href^="/reel/"] time');
    if(t && t.parentElement) postPath = t.parentElement.getAttribute('href') || '';
  }
  const canonical = (doc.querySelector('link[rel="canonical"]')||{}).href || '';
  const og = (doc.querySelector('meta[property="og:url"]')||{}).content || '';

  // 해시태그
  const seen = new Set(); const tags = [];
  if(li1){
    li1.querySelectorAll('a[href^="/explore/tags/"]').forEach(a=>{
      let raw = (a.textContent||'').trim().replace(/^#\s*/,'')
                 .replace(/[\u200b-\u200f\u202a-\u202e\u2060\s]+/g,'');
      if(raw && !seen.has(raw)){ seen.add(raw); tags.push(raw); }
    });
  }
  if(!tags.length && caption){
    (caption.match(/#([0-9A-Za-z_가-힣_]+)/g)||[]).forEach(s=>{
      const t = s.slice(1);
      if(t && !seen.has(t)){ seen.add(t); tags.push(t); }
    });
  }

  // 고유키(Shortcode)
  function pickHref() {
    return postPath || canonical || og || (doc.location ? doc.location.href : '');
  }
  const href = pickHref();
  let postKind = ''; let shortcode = '';
  const m = href.match(/\/(p|reel)\/([^\/\?\#]+)/);
  if(m){ postKind = m[1]; shortcode = m[2]; }

  return {caption, hashtags:tags, postPath, canonical, og, postKind, shortcode};
}
return extractFrom(document);
"""

def _normalize_url(u: str) -> str:
    m = re.search(r"https://www\.instagram\.com/(p|reel)/([^/\?#]+)/?", str(u))
    if m:
        return f"https://www.instagram.com/{m.group(1)}/{m.group(2)}/"
    return str(u or "")

def _shortcode_from_url(u: str) -> str:
    m = re.search(r"/(p|reel)/([^/\?#]+)/?", str(u))
    return m.group(2) if m else ""

def extract_current():
    d = driver.execute_script(JS_EXTRACT)
    cap = (d.get("caption") or "").strip()
    tags = d.get("hashtags") or []
    if d.get("postPath") and str(d["postPath"]).startswith("/"):
        url = "https://www.instagram.com" + d["postPath"]
    elif d.get("canonical"):
        url = d["canonical"]
    elif d.get("og"):
        url = d["og"]
    else:
        url = driver.current_url
    url = _normalize_url(url)
    code = _shortcode_from_url(url) or (d.get("shortcode") or "")
    return url, cap, tags, code

def modal_present():
    try:
        return driver.execute_script("return !!document.querySelector('div[role=\"dialog\"] article');")
    except:
        return False

def click_next_post_and_check(prev_code, timeout=4):
    """
    모달에서 '다음' 버튼(캐러셀 제외)을 클릭하고 shortcode가 바뀔 때까지 짧게 대기.
    반환: (changed: bool)
    """
    changed = False

    # 1) 컴포넌트 클릭
    selectors = [
        "//div[@role='dialog']//*[self::button or self::div[@role='button'] or self::a]"
        "[@aria-label='다음' or @aria-label='Next'][not(ancestor::article)]",
        "//div[@role='dialog']//*[name()='svg' and (@aria-label='다음' or @aria-label='Next')]"
        "/ancestor::*[self::button or self::div[@role='button'] or self::a][1][not(ancestor::article)]",
    ]
    clicked = False
    for xp in selectors:
        try:
            btn = driver.find_element(By.XPATH, xp)
            driver.execute_script("arguments[0].click();", btn)
            clicked = True
            break
        except Exception:
            continue

    if not clicked:
        # JS 후보
        js = r"""
        const d=document.querySelector('div[role="dialog"]'); if(!d) return false;
        const art=d.querySelector('article');
        const cands=[...d.querySelectorAll('[aria-label="다음"],[aria-label="Next"]')];
        const t=cands.find(el=>!art || !el.closest('article'));
        if(t){ t.click(); return true; }
        return false;
        """
        try:
            clicked = bool(driver.execute_script(js))
        except Exception:
            clicked = False

    if not clicked:
        # ArrowRight
        try:
            dlg = driver.find_element(By.CSS_SELECTOR, "div[role='dialog']")
            driver.execute_script("arguments[0].focus();", dlg)
            dlg.click()
            actions.send_keys(Keys.ARROW_RIGHT).perform()
            clicked = True
        except Exception:
            clicked = False

    if not clicked:
        return False

    # 2) 변경 대기
    end = time.time() + timeout
    while time.time() < end:
        try:
            _, _, _, code2 = extract_current()
            if code2 and code2 != prev_code:
                changed = True
                break
        except Exception:
            pass
        time.sleep(0.2)

    return changed

def close_modal_if_open():
    if not modal_present(): return
    xps = [
        "//div[@role='dialog']//button[@aria-label='닫기' or @aria-label='Close']",
        "//div[@role='dialog']//div[@role='button' and (@aria-label='닫기' or @aria-label='Close')]",
    ]
    for xp in xps:
        try:
            btn = driver.find_element(By.XPATH, xp)
            driver.execute_script("arguments[0].click();", btn)
            time.sleep(0.2)
            return
        except:
            pass
    try:
        actions.send_keys(Keys.ESCAPE).perform()
        time.sleep(0.2)
    except:
        pass

def open_modal_from_feed_unseen(visited_codes, max_feed_scroll=6):
    """
    피드에서 아직 보지 않은 (p|reel) 링크를 찾아 모달을 연다.
    - visited_codes: 이미 수집한 shortcode 집합
    """
    tries = 0
    while tries < max_feed_scroll:
        anchors = driver.find_elements(By.CSS_SELECTOR, "a[href*='/p/'], a[href*='/reel/']")
        for a in anchors:
            href = a.get_attribute("href") or ""
            if href.startswith("/"):
                href = "https://www.instagram.com" + href
            code = _shortcode_from_url(href)
            if not code or code in visited_codes:
                continue
            try:
                driver.execute_script("arguments[0].click();", a)
                time.sleep(random.uniform(*VISIT_WAIT))
                if modal_present():
                    return True
            except Exception:
                continue
        human_scroll_once()
        tries += 1
    return False

# ---------------- 메인 루프(모달 방식 + '중복 시 모달 닫고 재시작') ----------------
results = []
visited_codes = set()
total_feed_scrolls = 0
no_progress_rounds = 0
last_result_count = 0

# 1) 첫 모달 열기
if not open_modal_from_feed_unseen(visited_codes, max_feed_scroll=FEED_OPEN_TRIES):
    while total_feed_scrolls < MAX_SCROLLS and not modal_present():
        human_scroll_once()
        total_feed_scrolls += 1
        if open_modal_from_feed_unseen(visited_codes, max_feed_scroll=2):
            break

# 2) 모달 순회
same_guard = 0
while modal_present() and len(results) < TARGET_CAPTIONS and total_feed_scrolls < MAX_SCROLLS:
    try:
        url, cap, tags, code = extract_current()
        collected_before = len(results)

        if code and code not in visited_codes and cap.strip():
            visited_codes.add(code)
            results.append({"url": url, "caption": cap, "hashtags": tags})
            same_guard = 0
        else:
            same_guard += 1  # 중복/빈캡션 카운트

        prev_code = code
        changed = click_next_post_and_check(prev_code, timeout=4)
        if not changed:
            same_guard += 1  # 다음 클릭했는데 코드가 그대로면 추가 가산

        # 같은 포스트/빈값 반복 → 모달 닫고 피드에서 재시작
        if same_guard >= SAME_POST_RETRIES:
            close_modal_if_open()
            opened = open_modal_from_feed_unseen(visited_codes, max_feed_scroll=FEED_OPEN_TRIES)
            if not opened:
                # 넉넉히 스크롤하며 다시 시도
                round_scrolls = 0
                while round_scrolls < FEED_OPEN_TRIES and len(results) < TARGET_CAPTIONS:
                    human_scroll_once()
                    total_feed_scrolls += 1
                    round_scrolls += 1
                    if open_modal_from_feed_unseen(visited_codes, max_feed_scroll=2):
                        same_guard = 0
                        break
                else:
                    break  # 더 이상 새 모달을 못 열면 종료
            else:
                same_guard = 0

        # 무진행(결과 증가 없음) 라운드 추적 → 하드 종료 가드
        if len(results) == last_result_count:
            no_progress_rounds += 1
            if no_progress_rounds >= NO_PROGRESS_MAX_ROUND:
                close_modal_if_open()
                break
        else:
            no_progress_rounds = 0
            last_result_count = len(results)

    except WebDriverException:
        close_modal_if_open()
        if not open_modal_from_feed_unseen(visited_codes, max_feed_scroll=FEED_OPEN_TRIES):
            break

# ---------------- 백업: 피드에서 링크 방문(모달 불가 시) ----------------
if len(results) < TARGET_CAPTIONS:
    have = set()
    feed_rounds = 0
    while len(have) < TARGET_CAPTIONS*1.3 and feed_rounds < MAX_SCROLLS:
        anchors = driver.find_elements(By.CSS_SELECTOR, "a[href*='/p/'], a[href*='/reel/']")
        for a in anchors:
            href = a.get_attribute("href") or ""
            if href.startswith("/"):
                href = "https://www.instagram.com" + href
            code = _shortcode_from_url(href)
            if code and code not in visited_codes:
                have.add(_normalize_url(href))
        human_scroll_once()
        feed_rounds += 1

    for url in list(have):
        if len(results) >= TARGET_CAPTIONS: break
        try:
            driver.get(url)
            time.sleep(random.uniform(*VISIT_WAIT))
            u, cap, tags, code = extract_current()
            if code and code not in visited_codes and cap.strip():
                visited_codes.add(code)
                results.append({"url": u, "caption": cap, "hashtags": tags})
        except Exception:
            continue

# ---------------- CSV 저장 ----------------
with open(CSV_PATH, "w", newline="", encoding="utf-8-sig") as f:
    w = csv.writer(f)
    w.writerow(["url", "caption", "hashtags"])
    for r in results:
        w.writerow([r["url"], r["caption"], ";".join(r["hashtags"])])

print(f"[완료] 캡션 {len(results)}건 저장 → {CSV_PATH}")
driver.quit()
```

```python
# -*- coding: utf-8 -*-
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.common.by import By
from selenium.webdriver.common.action_chains import ActionChains
from selenium.webdriver.common.keys import Keys
from selenium.common.exceptions import TimeoutException, WebDriverException
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

import pickle, time, random, re, csv
from urllib.parse import quote

# ==================== 설정 ====================
KEYWORD           = "여자영양제"
TARGET_CAPTIONS   = 1000      # 수집 목표(캡션 있는 포스트 개수)
MAX_SCROLLS       = 2000     # 피드 스크롤 상한(총합)
VISIT_WAIT        = (0.55, 1.1)
SCROLL_SHORT      = (0.08, 0.18)
SCROLL_PAUSE      = (0.25, 0.55)
CSV_PATH          = "instagram_captions.csv"
INSTA_COOKIE_PKL  = "instagram_cookies.pkl"
FAST_HEADLESS     = False
BLOCK_MEDIA       = True     # 텍스트 위주면 True 권장

# 중복/무진행 가드
SAME_POST_RETRIES     = 2    # 모달 내 같은 포스트가 이 횟수 연속이면 모달 닫고 피드로 복귀
FEED_OPEN_TRIES       = 8    # 피드에서 새 모달 열기 재시도 횟수
NO_PROGRESS_MAX_ROUND = 8    # 결과 개수가 늘지 않는 라운드가 이 횟수면 종료
# =================================================

def make_driver():
    options = Options()
    options.page_load_strategy = "eager"
    options.add_argument("--start-maximized")
    options.add_argument("window-size=1920x1080")
    options.add_argument("lang=ko_KR")
    options.add_argument("--disable-blink-features=AutomationControlled")
    options.add_argument("--disable-infobars")
    options.add_argument("--no-sandbox")
    options.add_argument("--disable-dev-shm-usage")
    if FAST_HEADLESS:
        options.add_argument("--headless=new")
    if BLOCK_MEDIA:
        prefs = {
            "profile.managed_default_content_settings.images": 2,
            "profile.managed_default_content_settings.fonts": 2,
            "profile.managed_default_content_settings.video": 2,
            "profile.managed_default_content_settings.sound": 2,
            "intl.accept_languages": "ko_KR",
        }
        options.add_experimental_option("prefs", prefs)
        options.add_argument("--blink-settings=imagesEnabled=false")
    service = Service(ChromeDriverManager().install())
    drv = webdriver.Chrome(service=service, options=options)
    drv.set_page_load_timeout(50)
    return drv

driver = make_driver()
wait = WebDriverWait(driver, 10)
actions = ActionChains(driver)

# ---------------- 로그인/쿠키 ----------------
driver.get("https://www.instagram.com/")
time.sleep(1.0)
try:
    with open(INSTA_COOKIE_PKL, "rb") as f:
        cookies = pickle.load(f)
    for c in cookies:
        if "expiry" in c and isinstance(c["expiry"], float):
            c["expiry"] = int(c["expiry"])
        try:
            driver.add_cookie(c)
        except Exception:
            pass
    driver.refresh()
    time.sleep(1.0)
except FileNotFoundError:
    print("[경고] instagram_cookies.pkl 없음 → 비로그인일 수 있음")

for txt in ["필수 쿠키만 허용", "쿠키 허용", "Only allow essential cookies", "Allow all cookies",
            "나중에 하기", "Not now", "지금 아니오", "나중에"]:
    try:
        btn = driver.find_element(By.XPATH, f"//button[normalize-space()='{txt}']")
        driver.execute_script("arguments[0].click();", btn)
        time.sleep(0.2)
    except:
        pass

# ---------------- 태그 페이지 ----------------
def goto_hashtag(tag):
    enc = quote(tag)
    for url in [f"https://www.instagram.com/explore/tags/{enc}/",
                f"https://www.instagram.com/explore/tags/{tag}/"]:
        driver.get(url)
        try:
            wait.until(EC.presence_of_element_located((By.TAG_NAME, "a")))
            return
        except TimeoutException:
            continue
goto_hashtag(KEYWORD)

def human_scroll_once():
    actions.send_keys(Keys.END).perform()
    time.sleep(random.uniform(*SCROLL_PAUSE))
    dy = random.randint(350, 900)
    driver.execute_script("window.scrollBy(0, arguments[0]);", dy)
    time.sleep(random.uniform(*SCROLL_SHORT))

# ---------------- JS 파싱(모달/페이지 공통) ----------------
JS_EXTRACT = r"""
function extractFrom(doc) {
  function txt(el){ return el ? (el.textContent || '').trim() : ''; }
  const dialog = doc.querySelector('div[role="dialog"]');
  const art = (dialog && dialog.querySelector('article')) || doc.querySelector('article');
  if(!art) return {caption:'', hashtags:[], postPath:'', canonical:'', og:'', postKind:'', shortcode:''};

  // 더보기
  const more = Array.from(art.querySelectorAll('span,button')).find(e=>{
    const t=(e.textContent||'').trim();
    return ['더 보기','더보기','More','more','… 더보기','… more'].includes(t);
  });
  if(more){ more.click(); }

  // 첫 li 블록
  const li1 = art.querySelector('ul li:first-child');
  const capNode = (li1 && (li1.querySelector('h1[dir="auto"]') || li1.querySelector('span[dir="auto"]')))
                  || art.querySelector('h1[dir="auto"]');
  let caption = txt(capNode);
  if(!caption && li1) caption = txt(li1);

  // permalink 후보
  let anchor = art.querySelector('a[href^="/p/"], a[href^="/reel/"]');
  let postPath = anchor ? (anchor.getAttribute('href')||'') : '';
  if(!postPath){
    const t = art.querySelector('a[href^="/p/"] time, a[href^="/reel/"] time');
    if(t && t.parentElement) postPath = t.parentElement.getAttribute('href') || '';
  }
  const canonical = (doc.querySelector('link[rel="canonical"]')||{}).href || '';
  const og = (doc.querySelector('meta[property="og:url"]')||{}).content || '';

  // 해시태그
  const seen = new Set(); const tags = [];
  if(li1){
    li1.querySelectorAll('a[href^="/explore/tags/"]').forEach(a=>{
      let raw = (a.textContent||'').trim().replace(/^#\s*/,'')
                 .replace(/[\u200b-\u200f\u202a-\u202e\u2060\s]+/g,'');
      if(raw && !seen.has(raw)){ seen.add(raw); tags.push(raw); }
    });
  }
  if(!tags.length && caption){
    (caption.match(/#([0-9A-Za-z_가-힣_]+)/g)||[]).forEach(s=>{
      const t = s.slice(1);
      if(t && !seen.has(t)){ seen.add(t); tags.push(t); }
    });
  }

  // 고유키(Shortcode)
  function pickHref() {
    return postPath || canonical || og || (doc.location ? doc.location.href : '');
  }
  const href = pickHref();
  let postKind = ''; let shortcode = '';
  const m = href.match(/\/(p|reel)\/([^\/\?\#]+)/);
  if(m){ postKind = m[1]; shortcode = m[2]; }

  return {caption, hashtags:tags, postPath, canonical, og, postKind, shortcode};
}
return extractFrom(document);
"""

def _normalize_url(u: str) -> str:
    m = re.search(r"https://www\.instagram\.com/(p|reel)/([^/\?#]+)/?", str(u))
    if m:
        return f"https://www.instagram.com/{m.group(1)}/{m.group(2)}/"
    return str(u or "")

def _shortcode_from_url(u: str) -> str:
    m = re.search(r"/(p|reel)/([^/\?#]+)/?", str(u))
    return m.group(2) if m else ""

def extract_current():
    d = driver.execute_script(JS_EXTRACT)
    cap = (d.get("caption") or "").strip()
    tags = d.get("hashtags") or []
    if d.get("postPath") and str(d["postPath"]).startswith("/"):
        url = "https://www.instagram.com" + d["postPath"]
    elif d.get("canonical"):
        url = d["canonical"]
    elif d.get("og"):
        url = d["og"]
    else:
        url = driver.current_url
    url = _normalize_url(url)
    code = _shortcode_from_url(url) or (d.get("shortcode") or "")
    return url, cap, tags, code

def modal_present():
    try:
        return driver.execute_script("return !!document.querySelector('div[role=\"dialog\"] article');")
    except:
        return False

def wait_article_loaded(timeout=6):
    end = time.time() + timeout
    js = "return !!document.querySelector('div[role=\"dialog\"] article time, div[role=\"dialog\"] article h1, div[role=\"dialog\"] article ul');"
    while time.time() < end:
        try:
            if driver.execute_script(js):
                return True
        except Exception:
            pass
        time.sleep(0.2)
    return False

def click_next_post_and_check(prev_code, timeout=4):
    """
    모달에서 '다음'(캐러셀 제외)을 클릭하고 shortcode가 바뀔 때까지 대기.
    """
    clicked = False
    xps = [
        "//div[@role='dialog']//*[self::button or self::div[@role='button'] or self::a]"
        "[@aria-label='다음' or @aria-label='Next'][not(ancestor::article)]",
        "//div[@role='dialog']//*[name()='svg' and (@aria-label='다음' or @aria-label='Next')]"
        "/ancestor::*[self::button or self::div[@role='button'] or self::a][1][not(ancestor::article)]",
    ]
    for xp in xps:
        try:
            btn = driver.find_element(By.XPATH, xp)
            driver.execute_script("arguments[0].click();", btn)
            clicked = True
            break
        except Exception:
            continue

    if not clicked:
        js = r"""
        const d=document.querySelector('div[role="dialog"]'); if(!d) return false;
        const art=d.querySelector('article');
        const cands=[...d.querySelectorAll('[aria-label="다음"],[aria-label="Next"]')];
        const t=cands.find(el=>!art || !el.closest('article'));
        if(t){ t.click(); return true; }
        return false;
        """
        try:
            clicked = bool(driver.execute_script(js))
        except Exception:
            clicked = False

    if not clicked:
        try:
            dlg = driver.find_element(By.CSS_SELECTOR, "div[role='dialog']")
            driver.execute_script("arguments[0].focus();", dlg)
            dlg.click()
            actions.send_keys(Keys.ARROW_RIGHT).perform()
            clicked = True
        except Exception:
            clicked = False

    if not clicked:
        return False

    # 변경 대기
    end = time.time() + timeout
    while time.time() < end:
        try:
            _, _, _, code2 = extract_current()
            if code2 and code2 != prev_code:
                return True
        except Exception:
            pass
        time.sleep(0.2)
    return False

def close_modal_if_open():
    if not modal_present(): return
    xps = [
        "//div[@role='dialog']//button[@aria-label='닫기' or @aria-label='Close']",
        "//div[@role='dialog']//div[@role='button' and (@aria-label='닫기' or @aria-label='Close')]",
    ]
    for xp in xps:
        try:
            btn = driver.find_element(By.XPATH, xp)
            driver.execute_script("arguments[0].click();", btn)
            time.sleep(0.2)
            return
        except:
            pass
    try:
        actions.send_keys(Keys.ESCAPE).perform()
        time.sleep(0.2)
    except:
        pass

# --- 스테일 회피용: 피드에서 JS로 href만 수집 ---
JS_GET_POST_URLS = r"""
const list = Array.from(document.querySelectorAll("a[href*='/p/'], a[href*='/reel/']"))
  .map(a => {
    const h = a.getAttribute('href') || '';
    if(!h) return '';
    return h.startsWith('/') ? location.origin + h : h;
  })
  .filter(Boolean);
return Array.from(new Set(list));
"""

def get_feed_urls_by_js():
    try:
        urls = driver.execute_script(JS_GET_POST_URLS) or []
        # 정규화
        norm = []
        for u in urls:
            m = re.search(r"https://www\.instagram\.com/(p|reel)/([^/\?#]+)/?", str(u))
            if m:
                norm.append(f"https://www.instagram.com/{m.group(1)}/{m.group(2)}/")
        return list(dict.fromkeys(norm))
    except Exception:
        return []

def click_feed_by_shortcode(sc):
    sel = f"a[href*='/{sc}/']"
    js = r"""
      const sel = arguments[0];
      const a = document.querySelector(sel);
      if(!a) return false;
      a.scrollIntoView({block:'center', inline:'center'});
      a.click();
      return true;
    """
    try:
        ok = driver.execute_script(js, sel)
        if ok:
            time.sleep(random.uniform(*VISIT_WAIT))
            return wait_article_loaded(timeout=6) and modal_present()
    except Exception:
        pass
    return False

def open_modal_from_feed_unseen(visited_codes, max_feed_scroll=6):
    tries = 0
    while tries < max_feed_scroll:
        urls = get_feed_urls_by_js()
        opened_any = False
        for u in urls:
            sc = _shortcode_from_url(u)
            if not sc or sc in visited_codes:
                continue
            # 1) 모달 시도
            if click_feed_by_shortcode(sc):
                return True
            # 2) 백업: 새 탭에서 직접 열어 수집
            try:
                driver.execute_script("window.open(arguments[0], '_blank');", u)
                time.sleep(random.uniform(*VISIT_WAIT))
                driver.switch_to.window(driver.window_handles[-1])
                return True  # 모달은 아니지만 방문 상태에서 extract_current 가능
            except Exception:
                # 실패면 피드로 복귀
                try:
                    if len(driver.window_handles) > 1:
                        driver.close()
                        driver.switch_to.window(driver.window_handles[0])
                except Exception:
                    pass
                continue
        # 새 링크 못 찾음 → 스크롤
        human_scroll_once()
        tries += 1
    return False

# ---------------- 메인 루프(모달 방식 + '중복 시 모달 닫고 재시작') ----------------
results = []
visited_codes = set()
total_feed_scrolls = 0
no_progress_rounds = 0
last_result_count = 0

# 1) 첫 모달/탭 열기
if not open_modal_from_feed_unseen(visited_codes, max_feed_scroll=FEED_OPEN_TRIES):
    while total_feed_scrolls < MAX_SCROLLS and not modal_present():
        human_scroll_once()
        total_feed_scrolls += 1
        if open_modal_from_feed_unseen(visited_codes, max_feed_scroll=2):
            break

# 2) 수집 루프
same_guard = 0
home_handle = driver.window_handles[0]

while len(results) < TARGET_CAPTIONS and total_feed_scrolls < MAX_SCROLLS:
    try:
        # 모달이 있으면 모달에서 추출, 없으면 현재 페이지(개별 포스트 탭)에서 추출
        url, cap, tags, code = extract_current()
        collected_before = len(results)

        if code and code not in visited_codes and cap.strip():
            visited_codes.add(code)
            results.append({"url": url, "caption": cap, "hashtags": tags})
            same_guard = 0
        else:
            same_guard += 1

        prev_code = code

        progressed = False
        if modal_present():
            # 모달인 경우 다음 포스트로 이동
            changed = click_next_post_and_check(prev_code, timeout=4)
            if not changed:
                same_guard += 1
            else:
                progressed = True
        else:
            # 개별 탭인 경우: 탭 닫고 피드로 복귀 → 새 모달/탭 열기
            try:
                if len(driver.window_handles) > 1:
                    driver.close()
            except Exception:
                pass
            try:
                driver.switch_to.window(home_handle)
            except Exception:
                driver.switch_to.window(driver.window_handles[0])
                home_handle = driver.current_window_handle
            opened = open_modal_from_feed_unseen(visited_codes, max_feed_scroll=FEED_OPEN_TRIES)
            if opened:
                same_guard = 0
                progressed = True
            else:
                # 스크롤 넉넉히 후 재시도
                for _ in range(FEED_OPEN_TRIES):
                    human_scroll_once()
                    total_feed_scrolls += 1
                    if open_modal_from_feed_unseen(visited_codes, max_feed_scroll=2):
                        same_guard = 0
                        progressed = True
                        break
                if not progressed:
                    break

        # 같은 포스트/빈값 반복 → 모달 닫고 새로 시작
        if same_guard >= SAME_POST_RETRIES:
            close_modal_if_open()
            opened = open_modal_from_feed_unseen(visited_codes, max_feed_scroll=FEED_OPEN_TRIES)
            if opened:
                same_guard = 0
            else:
                # 더 시도해도 없으면 종료
                break

        # 무진행 라운드 가드
        if len(results) == last_result_count:
            no_progress_rounds += 1
            if no_progress_rounds >= NO_PROGRESS_MAX_ROUND:
                close_modal_if_open()
                break
        else:
            no_progress_rounds = 0
            last_result_count = len(results)

    except WebDriverException:
        close_modal_if_open()
        if not open_modal_from_feed_unseen(visited_codes, max_feed_scroll=FEED_OPEN_TRIES):
            break

# ---------------- CSV 저장 ----------------
with open(CSV_PATH, "w", newline="", encoding="utf-8-sig") as f:
    w = csv.writer(f)
    w.writerow(["url", "caption", "hashtags"])
    for r in results:
        w.writerow([r["url"], r["caption"], ";".join(r["hashtags"])])

print(f"[완료] 캡션 {len(results)}건 저장 → {CSV_PATH}")
driver.quit()
```

```python
# -*- coding: utf-8 -*-
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.common.by import By
from selenium.webdriver.common.action_chains import ActionChains
from selenium.webdriver.common.keys import Keys
from selenium.common.exceptions import TimeoutException, WebDriverException
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

import pickle, time, random, re, csv
from urllib.parse import quote

# ==================== 설정 ====================
KEYWORD           = "여자영양제"
TARGET_CAPTIONS   = 1000      # 캡션 있는 포스트 목표
MAX_SCROLLS       = 2000      # 피드 스크롤 상한
VISIT_WAIT        = (0.55, 1.1)
SCROLL_SHORT      = (0.08, 0.18)
SCROLL_PAUSE      = (0.25, 0.55)
CSV_PATH          = "instagram_captions.csv"
INSTA_COOKIE_PKL  = "instagram_cookies.pkl"
FAST_HEADLESS     = False
BLOCK_MEDIA       = True

# 가드/재시도
SAME_POST_RETRIES     = 2     # 모달에서 같은 포스트가 이 횟수 연속이면 종료→피드 재시작
FEED_OPEN_TRIES       = 8     # 피드에서 새 모달 열기 재시도
NO_PROGRESS_MAX_ROUND = 8     # 결과가 늘지 않는 라운드 상한
NEXT_WAIT_TIMEOUT     = 8     # 다음 포스트 로딩 대기(초)
# =================================================

def make_driver():
    options = Options()
    options.page_load_strategy = "eager"
    options.add_argument("--start-maximized")
    options.add_argument("window-size=1920x1080")
    options.add_argument("lang=ko_KR")
    options.add_argument("--disable-blink-features=AutomationControlled")
    options.add_argument("--disable-infobars")
    options.add_argument("--no-sandbox")
    options.add_argument("--disable-dev-shm-usage")
    if FAST_HEADLESS:
        options.add_argument("--headless=new")
    if BLOCK_MEDIA:
        prefs = {
            "profile.managed_default_content_settings.images": 2,
            "profile.managed_default_content_settings.fonts": 2,
            "profile.managed_default_content_settings.video": 2,
            "profile.managed_default_content_settings.sound": 2,
            "intl.accept_languages": "ko_KR",
        }
        options.add_experimental_option("prefs", prefs)
        options.add_argument("--blink-settings=imagesEnabled=false")
    service = Service(ChromeDriverManager().install())
    drv = webdriver.Chrome(service=service, options=options)
    drv.set_page_load_timeout(50)
    return drv

driver = make_driver()
wait = WebDriverWait(driver, 10)
actions = ActionChains(driver)

# ---------------- 로그인/쿠키 ----------------
driver.get("https://www.instagram.com/")
time.sleep(1.0)
try:
    with open(INSTA_COOKIE_PKL, "rb") as f:
        cookies = pickle.load(f)
    for c in cookies:
        if "expiry" in c and isinstance(c["expiry"], float):
            c["expiry"] = int(c["expiry"])
        try:
            driver.add_cookie(c)
        except Exception:
            pass
    driver.refresh()
    time.sleep(1.0)
except FileNotFoundError:
    print("[경고] instagram_cookies.pkl 없음 → 비로그인일 수 있음")

for txt in ["필수 쿠키만 허용", "쿠키 허용", "Only allow essential cookies", "Allow all cookies",
            "나중에 하기", "Not now", "지금 아니오", "나중에"]:
    try:
        btn = driver.find_element(By.XPATH, f"//button[normalize-space()='{txt}']")
        driver.execute_script("arguments[0].click();", btn)
        time.sleep(0.2)
    except:
        pass

# ---------------- 태그 페이지 ----------------
def goto_hashtag(tag):
    enc = quote(tag)
    for url in [f"https://www.instagram.com/explore/tags/{enc}/",
                f"https://www.instagram.com/explore/tags/{tag}/"]:
        driver.get(url)
        try:
            wait.until(EC.presence_of_element_located((By.TAG_NAME, "a")))
            return
        except TimeoutException:
            continue
goto_hashtag(KEYWORD)

def human_scroll_once():
    actions.send_keys(Keys.END).perform()
    time.sleep(random.uniform(*SCROLL_PAUSE))
    dy = random.randint(350, 900)
    driver.execute_script("window.scrollBy(0, arguments[0]);", dy)
    time.sleep(random.uniform(*SCROLL_SHORT))

# ---------------- JS 파싱(모달/페이지 공통) ----------------
JS_EXTRACT = r"""
function extractFrom(doc) {
  function txt(el){ return el ? (el.textContent || '').trim() : ''; }

  const dialog = doc.querySelector('div[role="dialog"]');
  const art = (dialog && dialog.querySelector('article')) || doc.querySelector('article');
  if(!art) return {caption:'', hashtags:[], postPath:'', canonical:'', og:'', publishedISO:'', publishedText:'', postKind:'', shortcode:''};

  // 더보기 클릭
  const more = Array.from(art.querySelectorAll('span,button')).find(e=>{
    const t=(e.textContent||'').trim();
    return ['더 보기','더보기','More','more','… 더보기','… more'].includes(t);
  });
  if(more){ more.click(); }

  // 캡션
  const li1 = art.querySelector('ul li:first-child');
  const capNode = (li1 && (li1.querySelector('h1[dir="auto"]') || li1.querySelector('span[dir="auto"]')))
                  || art.querySelector('h1[dir="auto"]');
  let caption = txt(capNode);
  if(!caption && li1) caption = txt(li1);

  // permalink & 시간
  let anchor = art.querySelector('a[href^="/p/"], a[href^="/reel/"]');
  let postPath = anchor ? (anchor.getAttribute('href')||'') : '';
  let t = art.querySelector('a[href^="/p/"] time, a[href^="/reel/"] time') || art.querySelector('time');
  if(!postPath && t && t.parentElement && t.parentElement.getAttribute('href')) {
    postPath = t.parentElement.getAttribute('href') || '';
  }
  const publishedISO  = (t && t.getAttribute('datetime')) || (doc.querySelector('meta[property="article:published_time"]')||{}).content || '';
  const publishedText = (t && (t.getAttribute('title') || t.textContent || '').trim()) || '';

  const canonical = (doc.querySelector('link[rel="canonical"]')||{}).href || '';
  const og = (doc.querySelector('meta[property="og:url"]')||{}).content || '';

  // 해시태그
  const seen = new Set(); const tags = [];
  if(li1){
    li1.querySelectorAll('a[href^="/explore/tags/"]').forEach(a=>{
      let raw = (a.textContent||'').trim().replace(/^#\s*/,'')
                 .replace(/[\u200b-\u200f\u202a-\u202e\u2060\s]+/g,'');
      if(raw && !seen.has(raw)){ seen.add(raw); tags.push(raw); }
    });
  }
  if(!tags.length && caption){
    (caption.match(/#([0-9A-Za-z_가-힣_]+)/g)||[]).forEach(s=>{
      const tt = s.slice(1);
      if(tt && !seen.has(tt)){ seen.add(tt); tags.push(tt); }
    });
  }

  // 고유키(Shortcode)와 타입
  function pickHref() {
    return postPath || canonical || og || (doc.location ? doc.location.href : '');
  }
  const href = pickHref();
  let postKind = ''; let shortcode = '';
  const m = href.match(/\/(p|reel)\/([^\/\?\#]+)/);
  if(m){ postKind = m[1]; shortcode = m[2]; }

  return {caption, hashtags:tags, postPath, canonical, og, publishedISO, publishedText, postKind, shortcode};
}
return extractFrom(document);
"""

def _normalize_url(u: str) -> str:
    m = re.search(r"https://www\.instagram\.com/(p|reel)/([^/\?#]+)/?", str(u))
    if m:
        return f"https://www.instagram.com/{m.group(1)}/{m.group(2)}/"
    return str(u or "")

def _shortcode_from_url(u: str) -> str:
    m = re.search(r"/(p|reel)/([^/\?#]+)/?", str(u))
    return m.group(2) if m else ""

# ---- (선택) iframe 백업: IG는 보통 동일 문서지만, 가끔 광고/임베드가 섞일 수 있어 보강
def _exec_extract_in_this_frame():
    d = driver.execute_script(JS_EXTRACT)
    return d if isinstance(d, dict) else {}

def _exec_extract_any_frame():
    d = _exec_extract_in_this_frame()
    if d.get("shortcode") or d.get("caption") or d.get("hashtags"):
        return d
    frames = driver.find_elements(By.TAG_NAME, "iframe")
    for fr in frames:
        try:
            driver.switch_to.frame(fr)
            d = _exec_extract_in_this_frame()
            driver.switch_to.default_content()
            if d.get("shortcode") or d.get("caption") or d.get("hashtags"):
                return d
        except Exception:
            try: driver.switch_to.default_content()
            except: pass
            continue
    return {}

def extract_current():
    """url, caption, hashtags, date_iso, date_text, shortcode"""
    d = _exec_extract_any_frame()
    cap = (d.get("caption") or "").strip()
    tags = d.get("hashtags") or []
    if d.get("postPath") and str(d["postPath"]).startswith("/"):
        url = "https://www.instagram.com" + d["postPath"]
    elif d.get("canonical"):
        url = d["canonical"]
    elif d.get("og"):
        url = d["og"]
    else:
        url = driver.current_url
    url = _normalize_url(url)

    date_iso  = (d.get("publishedISO") or "").strip()
    date_text = (d.get("publishedText") or "").strip()
    code = _shortcode_from_url(url) or (d.get("shortcode") or "")
    return url, cap, tags, date_iso, date_text, code

def modal_present():
    try:
        return driver.execute_script("return !!document.querySelector('div[role=\"dialog\"] article');")
    except:
        return False

def wait_article_loaded(timeout=8):
    end = time.time() + timeout
    js = "return !!document.querySelector('div[role=\"dialog\"] article time, div[role=\"dialog\"] article h1, div[role=\"dialog\"] article ul');"
    while time.time() < end:
        try:
            if driver.execute_script(js):
                return True
        except Exception:
            pass
        time.sleep(0.2)
    return False

def click_next_post_and_wait(prev_code, timeout=NEXT_WAIT_TIMEOUT):
    """
    '다음'(Next) **포스트** 버튼만 눌러서 캐러셀 이동을 피하고,
    shortcode가 바뀌었을 때만 True.
    """
    clicked = False
    # 1) article 바깥의 Next만 클릭
    xps = [
        "//div[@role='dialog']//*[self::button or self::div[@role='button'] or self::a]"
        "[@aria-label='다음' or @aria-label='Next'][not(ancestor::article)]",
        "//div[@role='dialog']//*[name()='svg' and (@aria-label='다음' or @aria-label='Next')]"
        "/ancestor::*[self::button or self::div[@role='button'] or self::a][1][not(ancestor::article)]",
    ]
    for xp in xps:
        try:
            btn = driver.find_element(By.XPATH, xp)
            driver.execute_script("arguments[0].click();", btn)
            clicked = True
            break
        except Exception:
            continue

    # 2) JS 후보
    if not clicked:
        js = r"""
        const d=document.querySelector('div[role="dialog"]'); if(!d) return false;
        const art=d.querySelector('article');
        const cands=[...d.querySelectorAll('[aria-label="다음"],[aria-label="Next"]')];
        const t=cands.find(el=>!art || !el.closest('article'));
        if(t){ t.click(); return true; }
        return false;
        """
        try:
            clicked = bool(driver.execute_script(js))
        except Exception:
            clicked = False

    # 3) ArrowRight 백업
    if not clicked:
        try:
            dlg = driver.find_element(By.CSS_SELECTOR, "div[role='dialog']")
            driver.execute_script("arguments[0].focus();", dlg)
            dlg.click()
            actions.send_keys(Keys.ARROW_RIGHT).perform()
            clicked = True
        except Exception:
            clicked = False

    if not clicked:
        return False

    # 새 포스트 로딩 대기
    end = time.time() + timeout
    changed = False
    while time.time() < end:
        try:
            wait_article_loaded(2)
            _, _, _, _, _, code2 = extract_current()
            if prev_code and code2 and code2 != prev_code:
                changed = True
                break
            # prev_code 없으면 게시물 요소가 보이기만 해도 진행
            if not prev_code and code2:
                changed = True
                break
        except Exception:
            pass
        time.sleep(0.25)
    return changed

def close_modal_if_open():
    if not modal_present(): return
    xps = [
        "//div[@role='dialog']//button[@aria-label='닫기' or @aria-label='Close']",
        "//div[@role='dialog']//div[@role='button' and (@aria-label='닫기' or @aria-label='Close')]",
    ]
    for xp in xps:
        try:
            btn = driver.find_element(By.XPATH, xp)
            driver.execute_script("arguments[0].click();", btn)
            time.sleep(0.2)
            return
        except:
            pass
    try:
        actions.send_keys(Keys.ESCAPE).perform()
        time.sleep(0.2)
    except:
        pass

# --- 피드 URL 수집(스테일 방지: JS로 문자열만 가져옴) ---
JS_GET_POST_URLS = r"""
const list = Array.from(document.querySelectorAll("a[href*='/p/'], a[href*='/reel/']"))
  .map(a => {
    const h = a.getAttribute('href') || '';
    if(!h) return '';
    return h.startsWith('/') ? location.origin + h : h;
  })
  .filter(Boolean);
return Array.from(new Set(list));
"""

def get_feed_urls_by_js():
    try:
        urls = driver.execute_script(JS_GET_POST_URLS) or []
        norm = []
        for u in urls:
            m = re.search(r"https://www\.instagram\.com/(p|reel)/([^/\?#]+)/?", str(u))
            if m:
                norm.append(f"https://www.instagram.com/{m.group(1)}/{m.group(2)}/")
        # unique preserving order
        return list(dict.fromkeys(norm))
    except Exception:
        return []

def click_feed_by_shortcode(sc):
    sel = f"a[href*='/{sc}/']"
    js = r"""
      const sel = arguments[0];
      const a = document.querySelector(sel);
      if(!a) return false;
      a.scrollIntoView({block:'center', inline:'center'});
      a.click();
      return true;
    """
    try:
        ok = driver.execute_script(js, sel)
        if ok:
            time.sleep(random.uniform(*VISIT_WAIT))
            return wait_article_loaded(timeout=8) and modal_present()
    except Exception:
        pass
    return False

def open_modal_from_feed_unseen(visited_codes, max_feed_scroll=6):
    tries = 0
    while tries < max_feed_scroll:
        urls = get_feed_urls_by_js()
        for u in urls:
            sc = _shortcode_from_url(u)
            if not sc or sc in visited_codes:
                continue
            # 1) 모달 시도
            if click_feed_by_shortcode(sc):
                return True
            # 2) 백업: 새 탭에서 직접 열기(모달 막힐 때)
            try:
                driver.execute_script("window.open(arguments[0], '_blank');", u)
                time.sleep(random.uniform(*VISIT_WAIT))
                driver.switch_to.window(driver.window_handles[-1])
                return True
            except Exception:
                # 실패 시 피드로 복귀
                try:
                    if len(driver.window_handles) > 1:
                        driver.close()
                        driver.switch_to.window(driver.window_handles[0])
                except Exception:
                    pass
                continue
        human_scroll_once()
        tries += 1
    return False

# ---------------- 메인 루프 ----------------
results = []
visited_codes = set()
total_feed_scrolls = 0
no_progress_rounds = 0
last_result_count = 0

# 1) 첫 포스트 열기
if not open_modal_from_feed_unseen(visited_codes, max_feed_scroll=FEED_OPEN_TRIES):
    while total_feed_scrolls < MAX_SCROLLS and not modal_present():
        human_scroll_once()
        total_feed_scrolls += 1
        if open_modal_from_feed_unseen(visited_codes, max_feed_scroll=2):
            break

# 2) 수집 루프(모달 우선, 모달이 안보이면 개별 탭에서 수집 후 피드 복귀)
same_guard = 0
home_handle = driver.window_handles[0]

while len(results) < TARGET_CAPTIONS and total_feed_scrolls < MAX_SCROLLS:
    try:
        url, cap, tags, date_iso, date_text, code = extract_current()

        # 수집(캡션 기준) + 중복 방지
        if code and code not in visited_codes and cap.strip():
            visited_codes.add(code)
            results.append({
                "url": url,
                "caption": cap,
                "hashtags": tags,
                "date_iso": date_iso,
                "date_text": date_text
            })
            same_guard = 0
        else:
            same_guard += 1

        prev_code = code
        progressed = False

        if modal_present():
            changed = click_next_post_and_wait(prev_code, timeout=NEXT_WAIT_TIMEOUT)
            progressed = changed
            if not changed:
                same_guard += 1
        else:
            # 개별 탭인 경우: 탭 닫고 피드에서 새 포스트 열기
            try:
                if len(driver.window_handles) > 1:
                    driver.close()
            except Exception:
                pass
            try:
                driver.switch_to.window(home_handle)
            except Exception:
                driver.switch_to.window(driver.window_handles[0])
                home_handle = driver.current_window_handle
            if open_modal_from_feed_unseen(visited_codes, max_feed_scroll=FEED_OPEN_TRIES):
                progressed = True
                same_guard = 0
            else:
                for _ in range(FEED_OPEN_TRIES):
                    human_scroll_once()
                    total_feed_scrolls += 1
                    if open_modal_from_feed_unseen(visited_codes, max_feed_scroll=2):
                        progressed = True
                        same_guard = 0
                        break
                if not progressed:
                    break

        # 같은 포스트/빈값 반복 → 모달 닫고 피드 재시작
        if same_guard >= SAME_POST_RETRIES:
            close_modal_if_open()
            if open_modal_from_feed_unseen(visited_codes, max_feed_scroll=FEED_OPEN_TRIES):
                same_guard = 0
            else:
                break

        # 무진행 하드 가드
        if len(results) == last_result_count:
            no_progress_rounds += 1
            if no_progress_rounds >= NO_PROGRESS_MAX_ROUND:
                close_modal_if_open()
                break
        else:
            no_progress_rounds = 0
            last_result_count = len(results)

    except WebDriverException:
        close_modal_if_open()
        if not open_modal_from_feed_unseen(visited_codes, max_feed_scroll=FEED_OPEN_TRIES):
            break

# ---------------- CSV 저장 ----------------
with open(CSV_PATH, "w", newline="", encoding="utf-8-sig") as f:
    w = csv.writer(f)
    w.writerow(["url", "caption", "hashtags", "date_iso", "date_text"])
    for r in results:
        w.writerow([r["url"], r["caption"], ";".join(r["hashtags"]), r.get("date_iso",""), r.get("date_text","")])

print(f"[완료] 캡션 {len(results)}건 저장 → {CSV_PATH}")
driver.quit()
```

```python
# -*- coding: utf-8 -*-
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.common.by import By
from selenium.webdriver.common.action_chains import ActionChains
from selenium.webdriver.common.keys import Keys
from selenium.common.exceptions import TimeoutException, WebDriverException
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import pickle, time, random, re, csv
from urllib.parse import quote
# ==================== 설정 ====================
KEYWORD           = "여자영양제"
TARGET_CAPTIONS   = 100          # 실전 수집량
MAX_SCROLLS       = 100
VISIT_WAIT        = (0.5, 0.9)
SCROLL_SHORT      = (0.08, 0.18)
SCROLL_PAUSE      = (0.25, 0.55)
CSV_PATH          = "instagram_captions.csv"
INSTA_COOKIE_PKL  = "instagram_cookies.pkl"
FAST_HEADLESS     = False
BLOCK_MEDIA       = True
# =================================================
def make_driver():
    options = Options()
    options.page_load_strategy = "eager"
    options.add_argument("--start-maximized")
    options.add_argument("window-size=1920x1080")
    options.add_argument("lang=ko_KR")
    options.add_argument("--disable-blink-features=AutomationControlled")
    options.add_argument("--disable-infobars")
    options.add_argument("--no-sandbox")
    options.add_argument("--disable-dev-shm-usage")
    if FAST_HEADLESS:
        options.add_argument("--headless=new")
    if BLOCK_MEDIA:
        prefs = {
            "profile.managed_default_content_settings.images": 2,
            "profile.managed_default_content_settings.fonts": 2,
            "profile.managed_default_content_settings.video": 2,
            "profile.managed_default_content_settings.sound": 2,
            "intl.accept_languages": "ko_KR",
        }
        options.add_experimental_option("prefs", prefs)
        options.add_argument("--blink-settings=imagesEnabled=false")
    service = Service(ChromeDriverManager().install())
    drv = webdriver.Chrome(service=service, options=options)
    drv.set_page_load_timeout(50)
    return drv
driver = make_driver()
wait = WebDriverWait(driver, 10)
actions = ActionChains(driver)
# ---------------- 로그인/쿠키 ----------------
driver.get("https://www.instagram.com/")
time.sleep(1.0)
try:
    with open(INSTA_COOKIE_PKL, "rb") as f:
        cookies = pickle.load(f)
    for c in cookies:
        if "expiry" in c and isinstance(c["expiry"], float):
            c["expiry"] = int(c["expiry"])
        try:
            driver.add_cookie(c)
        except Exception:
            pass
    driver.refresh()
    time.sleep(1.0)
except FileNotFoundError:
    print("[경고] instagram_cookies.pkl 없음 → 비로그인일 수 있음")
for txt in ["필수 쿠키만 허용", "쿠키 허용", "Only allow essential cookies", "Allow all cookies",
            "나중에 하기", "Not now", "지금 아니오", "나중에"]:
    try:
        btn = driver.find_element(By.XPATH, f"//button[normalize-space()='{txt}']")
        driver.execute_script("arguments[0].click();", btn)
        time.sleep(0.2)
    except:
        pass
# ---------------- 태그 페이지 ----------------
def goto_hashtag(tag):
    enc = quote(tag)
    for url in [f"https://www.instagram.com/explore/tags/{enc}/",
                f"https://www.instagram.com/explore/tags/{tag}/"]:
        driver.get(url)
        try:
            wait.until(EC.presence_of_element_located((By.TAG_NAME, "a")))
            return
        except TimeoutException:
            continue
goto_hashtag(KEYWORD)
def human_scroll_cycle():
    actions.send_keys(Keys.END).perform()
    time.sleep(random.uniform(*SCROLL_PAUSE))
    for _ in range(1):
        dy = random.randint(350, 900)
        driver.execute_script("window.scrollBy(0, arguments[0]);", dy)
        time.sleep(random.uniform(*SCROLL_SHORT))
# ---------------- JS 파싱(모달/페이지 공통) ----------------
JS_EXTRACT = r"""
function extractFrom(doc) {
  function txt(el){ return el ? (el.textContent || '').trim() : ''; }
  const dialog = doc.querySelector('div[role="dialog"]');
  const art = (dialog && dialog.querySelector('article')) || doc.querySelector('article');
  if(!art) return {caption:'', hashtags:[], postPath:'', canonical:'', og:''};
  // 더보기
  const more = Array.from(art.querySelectorAll('span,button')).find(e=>{
    const t=(e.textContent||'').trim();
    return ['더 보기','더보기','More','more','… 더보기','… more'].includes(t);
  });
  if(more){ more.click(); }
  // 첫 li 블록
  const li1 = art.querySelector('ul li:first-child');
  const capNode = (li1 && (li1.querySelector('h1[dir="auto"]') || li1.querySelector('span[dir="auto"]')))
                  || art.querySelector('h1[dir="auto"]');
  let caption = txt(capNode);
  if(!caption && li1) caption = txt(li1);
  // permalink 추정
  let postPath = '';
  const timeA = art.querySelector('a[href^="/p/"] time, a[href^="/reel/"] time');
  if(timeA && timeA.parentElement) postPath = timeA.parentElement.getAttribute('href') || '';
  const canonical = (doc.querySelector('link[rel="canonical"]')||{}).href || '';
  const og = (doc.querySelector('meta[property="og:url"]')||{}).content || '';
  // 해시태그
  const seen = new Set(); const tags = [];
  if(li1){
    li1.querySelectorAll('a[href^="/explore/tags/"]').forEach(a=>{
      let raw = (a.textContent||'').trim().replace(/^#\s*/,'')
                 .replace(/[\u200b-\u200f\u202a-\u202e\u2060\s]+/g,'');
      if(raw && !seen.has(raw)){ seen.add(raw); tags.push(raw); }
    });
  }
  if(!tags.length && caption){
    (caption.match(/#([0-9A-Za-z_가-힣_]+)/g)||[]).forEach(s=>{
      const t = s.slice(1);
      if(t && !seen.has(t)){ seen.add(t); tags.push(t); }
    });
  }
  return {caption, hashtags:tags, postPath, canonical, og};
}
return extractFrom(document);
"""
def extract_current():
    d = driver.execute_script(JS_EXTRACT)
    cap = (d.get("caption") or "").strip()
    tags = d.get("hashtags") or []
    url = ""
    if d.get("postPath") and str(d["postPath"]).startswith("/"):
        url = "https://www.instagram.com" + d["postPath"]
    elif d.get("canonical"):
        url = d["canonical"]
    elif d.get("og"):
        url = d["og"]
    else:
        url = driver.current_url
    return url, cap, tags
def modal_present():
    try:
        return driver.execute_script("return !!document.querySelector('div[role=\"dialog\"] article');")
    except:
        return False
def open_first_modal():
    anchors = driver.find_elements(By.CSS_SELECTOR, "a[href*='/p/'], a[href*='/reel/']")
    for a in anchors:
        try:
            driver.execute_script("arguments[0].click();", a)
            time.sleep(random.uniform(*VISIT_WAIT))
            if modal_present():
                return True
        except:
            pass
    return False
# --- 핵심: “다음 포스트”만 클릭(캐러셀 제외) ---
def click_next_post():
    """
    모달 바깥(ancestor::article 아닌) '다음/Next' 컨트롤만 클릭.
    실패 시 JS로 후보를 찾아 클릭하고, 그래도 안되면 ArrowRight 키 사용.
    """
    # 1) XPath: article 바깥
    xps = [
        "//div[@role='dialog']//*[self::button or self::div[@role='button'] or self::a]"
        "[@aria-label='다음' or @aria-label='Next'][not(ancestor::article)]",
        "//div[@role='dialog']//*[name()='svg' and (@aria-label='다음' or @aria-label='Next')]"
        "/ancestor::*[self::button or self::div[@role='button'] or self::a][1][not(ancestor::article)]",
    ]
    for xp in xps:
        try:
            btn = driver.find_element(By.XPATH, xp)
            driver.execute_script("arguments[0].click();", btn)
            time.sleep(random.uniform(*VISIT_WAIT))
            return True
        except:
            continue
    # 2) JS로 후보 찾아 클릭(역시 article 바깥만)
    js = r"""
    const d=document.querySelector('div[role="dialog"]'); if(!d) return false;
    const art=d.querySelector('article');
    const cands=[...d.querySelectorAll('[aria-label="다음"],[aria-label="Next"]')];
    const target=cands.find(el=>!art || !el.closest('article'));
    if(target){ target.click(); return true; }
    return false;
    """
    try:
        ok = driver.execute_script(js)
        if ok:
            time.sleep(random.uniform(*VISIT_WAIT))
            return True
    except:
        pass
    # 3) 최후: ArrowRight (모달 컨테이너에 포커스)
    try:
        dlg = driver.find_element(By.CSS_SELECTOR, "div[role='dialog']")
        driver.execute_script("arguments[0].focus();", dlg)
        dlg.click()
        actions.send_keys(Keys.ARROW_RIGHT).perform()
        time.sleep(random.uniform(*VISIT_WAIT))
        return True
    except:
        return False
def close_modal_if_open():
    if not modal_present(): return
    xps = [
        "//div[@role='dialog']//button[@aria-label='닫기' or @aria-label='Close']",
        "//div[@role='dialog']//div[@role='button' and (@aria-label='닫기' or @aria-label='Close')]",
    ]
    for xp in xps:
        try:
            btn = driver.find_element(By.XPATH, xp)
            driver.execute_script("arguments[0].click();", btn)
            time.sleep(0.2)
            return
        except:
            pass
    try:
        actions.send_keys(Keys.ESCAPE).perform()
        time.sleep(0.2)
    except:
        pass
# ---------------- 빠른 순회 로직 ----------------
results, visited = [], set()
# 1) 모달 열기(없으면 스크롤하며 시도)
if not open_first_modal():
    tries = 0
    while tries < 8 and not modal_present():
        human_scroll_cycle()
        if open_first_modal(): break
        tries += 1
# 2) 모달에서 다음 "포스트"로만 순회
same_guard = 0
while modal_present() and len(results) < TARGET_CAPTIONS:
    try:
        url, cap, tags = extract_current()
        if url not in visited and (cap or tags):
            visited.add(url)
            results.append({"url": url, "caption": cap, "hashtags": tags})
        prev_url = url
        if not click_next_post():
            # 끝 도달 → 닫고 스크롤 후 재개
            close_modal_if_open()
            human_scroll_cycle()
            if not open_first_modal():
                human_scroll_cycle()
                if not open_first_modal():
                    break
            same_guard = 0
            continue
        # “다음” 눌렀는데 같은 포스트라면(캐러셀만 넘어간 경우) → 다시 시도
        url2, _, _ = extract_current()
        if url2 == prev_url:
            same_guard += 1
            if same_guard >= 2:
                # 캐러셀에 갇힘 → 강제로 닫고 재개
                close_modal_if_open()
                human_scroll_cycle()
                open_first_modal()
                same_guard = 0
            continue
        same_guard = 0
    except WebDriverException:
        close_modal_if_open()
        human_scroll_cycle()
        open_first_modal()
# 3) 백업: 모달이 막히면 링크 방문
if len(results) < TARGET_CAPTIONS:
    close_modal_if_open()
    have = set()
    for _ in range(20):
        anchors = driver.find_elements(By.CSS_SELECTOR, "a[href*='/p/'], a[href*='/reel/']")
        for a in anchors:
            href = a.get_attribute("href")
            if href and ("/p/" in href or "/reel/" in href):
                have.add(href)
        if len(have) >= TARGET_CAPTIONS*1.3: break
        human_scroll_cycle()
    links = [h for h in have if h not in visited]
    for url in links:
        if len(results) >= TARGET_CAPTIONS: break
        try:
            driver.get(url)
            time.sleep(random.uniform(*VISIT_WAIT))
            u, cap, tags = extract_current()
            if u not in visited and (cap or tags):
                visited.add(u)
                results.append({"url": u, "caption": cap, "hashtags": tags})
        except Exception:
            continue
# ---------------- CSV 저장 ----------------
with open(CSV_PATH, "w", newline="", encoding="utf-8-sig") as f:
    w = csv.writer(f)
    w.writerow(["url", "caption", "hashtags"])
    for r in results:
        w.writerow([r["url"], r["caption"], ";".join(r["hashtags"])])
print(f"[완료] 캡션 {len(results)}건 저장 → {CSV_PATH}")
driver.quit()
```
