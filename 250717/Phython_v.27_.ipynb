```python
### 문자열 관련 함수 리스트 / 문자열과 문자 different
- count : 문자열 객체 안에 있는 특정 문자 갯수 세기
- index : 문자열에 있는 특정 문자의 인덱스 위치를 알려주기 / 왼쪽부터 시작, 찾으면 끝
- find : 문자열에 있는 특정 문자의 인덱스를 찾아오긴 하나, 만약 존재하지 않는다면 에러X, -1을 반환한다
- join : 문자열 사이에 다른 문자를 끼워 넣기 / 인자값 사이에 문자열 append
- strip : 문자열 앞뒤에 공백을 지운다
- upper : 소문자를 대문자로 변경하기
- lower : 대문자르 소문자로 변경하기
- split : 특정 문자열 내 특정 문자열을 기준으로 나누고, list 저장
- replace : 문자열 중 일부를 다른 문자로 바꾸거나 혹은 삭제하기
```

```python
num1 = "10, 11, 12, 13, 14" 

list01 = []
for i in num1 :
    list01.append(i)
print(list01)
```

```python
num1 = "10, 11, 12, 13, 14" 
list02 = num1.split(",")
for index, list02_item in enumerate(list02) :
    list02[index] = int(list02_item)
print(list02)
```

```python
string08 = "Dave goes to Korea"
string08.replace("Dave", "David")

string09 = "(David)"
string09.replace("(", "")

string10 = "(David)"
string10.replace("()","")

string11 = "(David)"
string12 = string11.replace("(","")
string12.replace(")","")

# Method Chaining
string13 = string11.replace("(","").replace(")","")
string13
```
