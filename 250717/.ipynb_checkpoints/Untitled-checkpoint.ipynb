```python
#    substitute => 대체하다
#    re.sub(,""data.get_text().strip()) => 첫번쨰 패턴, 두번째 대체, 마지막 범위
#re.compile("[]")
#온점이 마지막에 붙음~ +\ 요소 반복~
### 정규표현식 문법 리스트

- [0-9] : 숫자를찾음 => \D
- [^0-9] : 숫자가 아닌 형태의 값을 찾을 때 (*문자, 특수문자, 공백 등)
- [\t\n] : 탭과 같은 여백, 줄바꿈 등의 문자를 찾고자 할 때
- [^\t\n] : 탭과 같은 여백, 줄바꿈 외의 값을 찾기 위해
- [A-Za-z0-9] : 문자 혹은 숫자를 찾음 => \w
- [^A-Za-z0-9] : 문자 혹은 숫자 외의 값을 찾고자 할 때 => \W
```

```python
### 정규표현식 문법 리스트 - 2
- Dot = . : 문자 1개
- ? : 앞에 있는 문자가 0번 또는 1번 표시되는 패턴 = 없어도 되고, 있다면 한 번만 있어야 하는 패턴
- + : 앞에 있는 문자가 1번 또는 그 이상 반복되는 패턴
- * : 앞에 있는 문자가 0번 또는 그 이상 반복되는 패턴
```

```python
### 정규표현식 문법 리스트 - 3
- {n} : 앞에 있는 문자가 n 번 반복되는 패턴
- {m, n} : 앞에 있는 문자가 m 번 반복되는 패턴부터 n 번 반복되는 패턴까지
```

```python
### 정규표현식 문법 리스트 - 4
- [] : 대괄호 내부에 입력한 값을 "또는"의 형태로 찾아올 수 있는 패턴
- ex) [abc] : a 또는 b 또는  c 중 하나가 들어가 있는 패턴
- "-" : 패턴 내 모든 문자 범위 지정
- [a-zA-Z] : 소문자 알파벳, 대문자 알파벳 모두
```

```python
### 정규표현식 문법 리스트 - 5
- 한글만 정규표현식의 패턴을 정의 : [가-힣]
```

```python
### 정규표현식 문법 리스트 - 5
- match(funtion) : 문자열의 처음부터 끝까지 정규표현식과 매칭되어지는 문자만을 찾아서 리턴
- search : 문자열 전체 중에서 정규표현식과 매칭되어지는 문자가 있으면 찾아서 리턴
- findall : 정규표현식 패턴과 매칭되는 모든 문자열을 리스트 자료형태로 리턴
- split 함수 : 정규표현식 패턴에서 구분자로 입력된 문자열을 기준을 문자열을 분리, 리스트 저장
```

```python




from urllib.request import urlopen
from bs4 import BeautifulSoup
import re

res = urlopen("https://crawlingtest.netlify.app/")
soup = BeautifulSoup(res, "html.parser")
datas = soup.select("ul#dev_course_list > li.course")
for data in datas :
    print(re.sub("\[[0-9]+\]", "", data.get_text().strip())) # escape sewuence => 독자적인 문자열 표현
    
    #print(data.get_text().strip())

```

```python
import re
pattern16 = re.compile("-")
subed = pattern16.sub("*",  "801210 - 101132")
print(subed)
```

```python
import openpyxl
import re

pattern17 = re.compile("-.......")
regex = re.compile("[0-9]+[-][0-9]+")

work_book = openpyxl.load_workbook("data_kr.xlsx")
work_sheet = work_book.active

for each_row in work_sheet :
    num2 = regex.findall(each_row[1].value)
    for item in num2 :
        item01 = pattern17.sub( "-******", item)
        print(item01)
```
