```python
import plotly.express as px
import pandas as pd
import numpy as np # 난수값

df = pd.DataFrame({
    "x열" : np.random.rand(100), # 0~1 미만의 실수자료형의 값을 랜덤으로 생성
    "y열" : np.random.rand(100),
    "색상기준열" : np.random.choice(["A", "B", "C"], 100)
})

fig = px.scatter(
    df,
    x="x열",
    y="y열",
    color="색상기준열",
    title="산점도 예제"
)

fig.show()
```

### 산점도 그래프 활용 방법
1) 관계분석 : 서로 다른 변수 사이에 상관관계가 존재하는지 파악하고 싶을 때
2) 패턴 및 경향성 파악 : 특정 데이터들이 어떤 패턴이나 경향성을 보이는지 확인하고자 할 때
3) 이상치 탐지(*Outlier) : 전체 데이터에서 특정 데이터가 다른 값들과 동떨어진 좌표값 = 이상치를 가지고 있을 시, 식별
4) 탐색적 데이터 분석(*EDA) 초반 단계 : 전체 총 데이터안에 있는 feature들 간 상관관계 = 상호작용 여부를 판단하고자 할 

```python
import plotly.express as px
import pandas as pd
import numpy as np 

df = pd.DataFrame({
    "x열" : np.arange(50),
    "y열" : np.random.rand(50)
})

fig = px.line(
    df,
    x="x열",
    y="y열",
    title="선 그래프 예제"
)

fig.update_xaxes(
    title="시간(초)",
    title_font={
        "size" : 10,
        "color" : "purple"
    },
    tickangle = 45,
    tickfont= {
        "size" : 12
    },
    dtick=2
)

import plotly.graph_objects as go

go.Figure()
fig.show()
```

### 선그래프 : 순차적인 순서에 따른 값의 변화를 선으로 연결해서 보고 싶을 때

```python
import plotly.express as px
import pandas as pd

df = pd.DataFrame({
    "범주열": ["a","b","c","d"],
    "값열" : [23, 15, 21, 33]
})

fig = px.bar(
    df,
    x = "범주열",
    y = "값열",
    title = "막대그래프 예제"
)

fig.show()
```

### 히스토그램 => 연속형 데이터의 분포를 구간별 막대로 나타내고자 할 때


```python
import plotly.express as px
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Data": np.random.rand(500)
})

fig = px.histogram(
    df,
    x="Data",
    nbins=30, # number of bins => 구간 | 막대 하나
    title="simple histogram example"
)

fig.show()
```

### 상자그림 그래프 (Box Plot)
- 여러개의 열의 데이터 분포, 중앙값(median), 사분위수 및 이상치를 한 눈에 확인 및 비교하고자 할 때
- 사분위수 계산 :
  Q1 : 중앙값을 기준으로 하위 25% 값
  Q3 : 데이터의 상위 25% 값
  이상치 : 최저값, 최대값 벗어난 수
  => 이상치 값을 계산하기 위한 공식
  => Q3- Q1 = IQR = interquartile Range
  => 하한 이상치, 상한 이상치
  => Q1 - (IQR * 1.5)
  => Q3 + (IQR * 1.5)
  ~> 하한보다 작은 값, 상한보다 높은 값이 나옴 => 이상치
  ~> 해당 특이점으로부터 발견해야하는 인사이트, 도출해야하는 것

```python
# 각 개별 그래프의 세부요소 변경!!
# 옵션 => 다양 // 업데이트
# 1) update_layout() : 해당 그래프에 대한 전체 레이아웃 수정 => 제목, 범례, 여백
# 2) update_xaxes() : x축에 해당되는 속성 요소 수정 => x축 제목, 글꼴, 눈금
# 3) update_yaxes() : y축에 해당되는 속성 요소 수정 => y축 제목, 글꼴, 눈금
```

```python
import plotly.express as px
import pandas as pd
import numpy as np # 난수값

df = pd.DataFrame({
    "x열" : np.random.rand(100), # 0~1 미만의 실수자료형의 값을 랜덤으로 생성
    "y열" : np.random.rand(100),
    "색상기준열" : np.random.choice(["A", "B", "C"], 100)
})

fig = px.scatter(
    df,
    x="x열",
    y="y열",
    color="색상기준열",
    title="산점도 예제"
)

fig.update_layout(
    title={
        "text": "새로운 그래프 제목",
        "font": {"size": 12, "color": "#537895"},
        "x": 0.5
    },
    legend={
        "title" : "제목",
        "x" : 1.02,
        "y" : 1,
        "bordercolor" : "#000",
        "borderwidth" : 1
    }
    # legend=dict(
    #     title="제목",
    #     x=1,
    #     y=1
    # )
)


fig.show()
```
